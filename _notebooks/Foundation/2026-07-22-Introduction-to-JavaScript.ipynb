## 1. Reference Guide

| Term | Meaning | Example |
|------|---------|---------|
| Object literal | Related information grouped inside `{ }`. | `const player = { name: "Alex", health: 100 };` |
| Property | A key and its value, separated by a colon. | `health: 100` |
| String / number | Text in quotes / a numeric value. | `"Alex"` / `100` |
| Dot notation | Reads or changes a property. | `player.health` |
| `console.log` | Prints a value in the Code Runner output. | `console.log(player.health);` |
| Nested object | An object stored as a property of another object. | `position: { x: 10, y: 20 }` |

Separate properties with commas. `const` keeps the name bound to the same object; you can still change its properties. A JavaScript object literal is related to JSON, but it is not the same as JSON text.
