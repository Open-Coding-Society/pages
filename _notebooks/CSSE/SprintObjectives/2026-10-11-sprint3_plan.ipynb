### Wayfinding Journey

2. **Am I collaborating with my peers? (Social wayfinding)**

   - How do I initiate or respond to chat in ways that move our work forward?
   - What do my OCS analytics reveal about my habits? What do those records not capture?

### Review Ticket Tips

### Retrospective

1. **Tools Mastery**:
   - Show examples of tools you used effectively in your projects.
   - Reflect on a tool you learned, a problem it helped you solve, or a limitation you encountered.

2. **Game Contribution**:
   - Link a commit or pull request and include code samples or project snippets that show your programming contribution.
   - Use contribution records or analytics to support ownership. Explain what you built, why it mattered, and what you learned while building it.

### Beyond Perfunctory

Going beyond perfunctory work means engaging with the learning, not simply producing more text or claiming everything went well. 

- **Tools mastery**: How did your use of tools become more effective or independent?
- **Programming skills**: What can you now explain, build, or improve that you could not do before?
- **CS111 connections**: How do the lessons you taught or the code you contributed connect to CS111 concepts?
- **Teaching and homework grading**: What did classmates' questions or submissions teach you about your lesson and feedback?
- **N@tM and testing **: What did sharing your game reveal about your accomplishments, communication, or next steps?


## AI Evaluation Guidance

Evaluate assignment completion against the listed checklist and grading guideline. Keep Skills Evaluation qualitative: consider the student's specific experiences, reasoning, initiative, and response to feedback. Do not turn every reflection question into an additional required task or reward length and polished language over substance. Honest discussion of difficulties is not, by itself, evidence of poor performance. Use the grading tables for scores and provide constructive feedback that supports the next stage of learning.
