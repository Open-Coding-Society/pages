# Developing Algorithms

## The Core Idea

Two algorithms can look completely different and still produce **exactly the same result**. Two algorithms can look almost identical and produce **completely different results**.

Here is the whole lesson in one example. A mountain biker wants to know if a trail fits in the distance they are willing to ride.

```
// Algorithm A
IF (trailLength <= maxDistance)
{
  RETURN(true)
}
ELSE
{
  RETURN(false)
}

// Algorithm B - different code, same result
RETURN(trailLength <= maxDistance)
```

Algorithm B is four lines shorter, but for every possible input it returns the same answer as Algorithm A. Learning to tell when that is true, and when it is *not* true, is what Topic 3.9 is about.

| Term | What it means here |
| --- | --- |
| **Equivalent algorithms** | Different code, same result for every input |
| **Side effect** | A change the algorithm makes beyond its return value, like updating a counter |
| **Trace** | Stepping through code line by line to predict the output |
| **Combine** | Using two working algorithms together to solve a bigger problem |

## What College Board Expects

Topic 3.9 sits inside **Big Idea 3: Algorithms and Programming**, which College Board weights at 30-35% of the AP exam, the largest single portion of the multiple-choice section.

The topic covers two learning objectives. The first is about **comparing** algorithms. College Board's Essential Knowledge states that *"Algorithms can be written in different ways and still accomplish the same tasks"* and that *"Algorithms that appear similar can yield different side effects or results."* It also specifies that *"Some conditional statements can be written as equivalent Boolean expressions"* and that the reverse is true as well.

The second learning objective is about **creating** algorithms: you can write a new algorithm from scratch, combine two existing algorithms, or modify one that already works.

## Part 1 - Same Result, Different Code

**Presenter: Luke Sanders**

Start with the easier half of 3.9. Two algorithms are **equivalent** when they return the same result for every possible input, even if the code looks nothing alike.

The most common version of this on the AP exam is swapping a conditional statement for a Boolean expression, or flipping which branch of an `IF` does the work.

Below, `fitsA` checks whether the trail is short enough. `fitsB` checks whether it is too long and flips the answers. Different logic, identical output.

Run it, then change `trailLength` to `15` and run it again. Both algorithms should still agree.

### How to check equivalence

You cannot tell whether two algorithms are equivalent by looking at how similar the code is. You have to compare **behavior**, which means testing the inputs that sit on the boundary.

For this trail check, the inputs worth testing are:

| Input | Why it matters |
| --- | --- |
| `trailLength = 8` | Clearly under the limit |
| `trailLength = 15` | Clearly over the limit |
| `trailLength = 10` | Exactly at the limit, where `<=` and `<` disagree |

That third row is where most "equivalent" algorithms turn out not to be. If `fitsB` used `>=` instead of `>`, the two algorithms would agree on 8 and 15 but disagree on 10.

## Part 2 - Same Code, Different Result

**Presenter: Yiming Yin**

This is the harder half of 3.9, and the version the exam asks about most. Two algorithms can be nearly character-for-character identical and still produce different output, because of where a single line sits.

A rider has a list of trail lengths and wants to know how many fit within their maximum distance. Both algorithms below loop through the list and count. One of them is wrong.

**Do not run this yet.** Read it first and predict what each algorithm displays.

{% capture challenge2 %}

{% endcapture %}

{% capture code2 %}
trailLengths ← [4, 9, 12, 6, 15]
maxDistance ← 10

// Algorithm A
countA ← 0
FOR EACH length IN trailLengths
{
  IF (length <= maxDistance)
  {
    countA ← countA + 1
  }
}
DISPLAY("Algorithm A counted: " + countA)

// Algorithm B
countB ← 0
FOR EACH length IN trailLengths
{
  IF (length <= maxDistance)
  {
    countB ← countB + 1
  }
  countB ← countB + 1
}
DISPLAY("Algorithm B counted: " + countB)
{% endcapture %}

### Tracing the difference

The only difference is that Algorithm B has one extra `countB ← countB + 1` **outside** the `IF` block. That single line changes what the algorithm measures.

| Trail length | Algorithm A does | Algorithm B does |
| --- | --- | --- |
| 4 | Fits, count becomes 1 | Fits, count becomes 1, then +1 = 2 |
| 9 | Fits, count becomes 2 | Fits, count becomes 3, then +1 = 4 |
| 12 | Too long, no change | No match, but +1 anyway = 5 |
| 6 | Fits, count becomes 3 | Fits, count becomes 6, then +1 = 7 |
| 15 | Too long, no change | No match, but +1 anyway = 8 |

Algorithm A answers "how many trails fit." Algorithm B answers "how many trails fit, plus how many trails exist." That second number is not useful to anybody, which is exactly why this bug is hard to notice: the program still runs, still prints a number, and never crashes.

This is what College Board means by algorithms that *appear similar* yielding *different results*.

## Part 3 - Creating, Combining, and Modifying

**Presenter: Noor Mohammed Saif Bijapur**

The second learning objective in 3.9 is about building algorithms rather than comparing them. College Board lists three ways to do it, and you almost never need the first one:

1. **Create** a new algorithm from scratch
2. **Combine** two existing algorithms
3. **Modify** an existing algorithm

Most real programs, including your CPT, are built from patterns you already know. Two patterns cover a huge amount of ground:

| Pattern | Shape |
| --- | --- |
| **Count matches** | Start at 0, loop, add 1 when a condition is true |
| **Find the extreme** | Assume the first item wins, loop, replace when you find better |

Here are both, working on the same trail data.

{% capture challenge3 %}
Two separate algorithms built from known patterns. Read how each one works, then run it.
{% endcapture %}

{% capture code3 %}
trailLengths ← [5, 8, 15, 12]
maxDistance ← 10

// Pattern 1: count the matches
PROCEDURE countInRange(lengths, maxDist)
{
  count ← 0
  FOR EACH length IN lengths
  {
    IF (length <= maxDist)
    {
      count ← count + 1
    }
  }
  RETURN(count)
}

// Pattern 2: find the extreme
PROCEDURE findLongest(lengths)
{
  longest ← lengths[1]
  FOR EACH length IN lengths
  {
    IF (length > longest)
    {
      longest ← length
    }
  }
  RETURN(longest)
}

DISPLAY("Trails within range: " + countInRange(trailLengths, maxDistance))
DISPLAY("Longest trail: " + findLongest(trailLengths))
{% endcapture %}

### Modifying a pattern

`countInRange` counts trails but does not tell the rider **which** trails. Modifying it takes three changes to the pattern, not a rewrite:

| Counting version | Collecting version |
| --- | --- |
| `count ← 0` | `matches ← []` |
| `count ← count + 1` | `APPEND(matches, name)` |
| `RETURN(count)` | `RETURN(matches)` |

The loop and the `IF` condition stay exactly the same. That is what modifying an algorithm looks like: keep the structure, change what it accumulates.

### Combining two algorithms

Combining means using one algorithm's output as another's input, or running both and comparing. For example, a rider could use `countInRange` to check whether *any* trail fits, and only call `findLongest` if the count is greater than zero.

```
fitCount ← countInRange(trailLengths, maxDistance)

IF (fitCount > 0)
{
  DISPLAY("Found " + fitCount + " trails. Longest option: " + findLongest(trailLengths))
}
ELSE
{
  DISPLAY("No trails match. Try increasing your max distance.")
}
```

Neither procedure had to change. That is the payoff of building algorithms out of reusable pieces, and it is what earns abstraction points on the CPT.
