    diceHelpBtn:    $("diceHelpButton"),
    dicePanel:      $("diceExplanationPanel"),
    diceCube:       $("diceCube"),
    diceVal:        $("diceDisplayValue"),
    lastRoll:       $("lastRollText"),
    rollOnce:       $("rollOnceButton"),
    rollTen:        $("rollTenButton"),
    rollHundred:    $("rollHundredButton"),
    customInput:    $("customRollInput"),
    customBtn:      $("customRollButton"),
    resetDice:      $("resetDiceButton"),
    totalRolls:     $("totalRollsText"),
    mostCommon:     $("mostCommonText"),
    leastCommon:    $("leastCommonText"),
    barChart:       $("diceBarChart"),
    tableBody:      $("diceResultsTableBody"),
    copyBtn:        $("copyDiceSummaryButton"),
    dicePrompt:     $("diceReflectionPrompt"),
    newDicePrompt:  $("newDicePromptButton"),
    flightHelpBtn:  $("flightHelpButton"),
    flightPanel:    $("flightExplanationPanel"),
    resetFlight:    $("resetFlightButton"),
    runFlight:      $("runFlightButton"),
    randomize:      $("randomizeFlightButton"),
    flightScene:    $("flightScene"),
    plane:          $("animatedPlane"),
    peopleCont:     $("fallingPeopleContainer"),
    explosion:      $("explosionEffect"),
    safeScore:      $("safetyScoreText"),
    randRoll:       $("randomRollText"),
    outcome:        $("outcomeText"),
    resultBox:      $("flightResultBox"),
    riskList:       $("riskBreakdownList"),
    flightPrompt:   $("flightReflectionPrompt"),
    newFlightPr:    $("newFlightPromptButton"),
    weatherIn:      $("weatherInput"),
    windIn:         $("windInput"),
    visIn:          $("visibilityInput"),
    planeStyleIn:   $("planeStyleInput"),
    planeAgeIn:     $("planeAgeInput"),
    maintIn:        $("maintenanceInput"),
    passIn:         $("passengerInput"),
    pilotIn:        $("pilotInput"),
    copilotIn:      $("copilotInput"),
  };

  const rndInt  = (a,b) => Math.floor(Math.random()*(b-a+1))+a;
  const pick    = arr  => arr[Math.floor(Math.random()*arr.length)];
  const clamp   = (v,a,b) => Math.max(a,Math.min(b,v));
  const pct     = v    => `${v.toFixed(1)}%`;

  /* ── VOCAB ── */
  el.vocabChips.forEach(btn => btn.addEventListener("click", () => {
    el.vocabBox.textContent = vocabDefs[btn.dataset.vocab] || "";
    el.vocabChips.forEach(b => b.classList.toggle("active", b===btn));
  }));

  /* ── DICE ── */
  function rollOnce() {
    const r = rndInt(1,6);
    state.diceCounts[r-1]++;
    state.totalRolls++;
    state.lastRoll = r;
    return r;
  }

  function animDice(r) {
    el.diceCube.classList.remove("is-rolling");
    void el.diceCube.offsetWidth;
    el.diceCube.classList.add("is-rolling");
    el.diceVal.textContent = r;
  }

  function prob(i) { return state.totalRolls ? state.diceCounts[i]/state.totalRolls : 0; }

  function renderDice() {
    el.totalRolls.textContent = state.totalRolls;
    if (state.totalRolls > 0) {
      const max = state.diceCounts.indexOf(Math.max(...state.diceCounts));
      const min = state.diceCounts.indexOf(Math.min(...state.diceCounts));
      el.mostCommon.textContent  = max+1;
      el.leastCommon.textContent = min+1;
    } else {
      el.mostCommon.textContent = el.leastCommon.textContent = "N/A";
    }
    el.lastRoll.textContent = state.lastRoll ? `Rolled a ${state.lastRoll}` : "No rolls yet";
    renderChart(); renderTable();
  }

  function renderChart() {
    el.barChart.innerHTML = "";
    for (let v=1;v<=6;v++) {
      const p = prob(v-1)*100;
      const row = document.createElement("div"); row.className="dice-bar-row";
      row.innerHTML = `<div class="dice-bar-label">Side ${v}</div>
        <div class="dice-bar-track"><div class="dice-bar-fill" style="width:${p}%"></div></div>
        <div class="dice-bar-percent">${pct(p)}</div>`;
      el.barChart.appendChild(row);
    }
  }

  function renderTable() {
    el.tableBody.innerHTML = "";
    for (let v=1;v<=6;v++) {
      const p = prob(v-1)*100;
      const diff = p - 100/6;
      const tr = document.createElement("tr");
      tr.innerHTML = `<td>${v}</td><td>${state.diceCounts[v-1]}</td><td>${pct(p)}</td>
        <td>${diff>=0?"+":""}${diff.toFixed(1)}%</td>
        <td><div class="mini-bar"><div class="mini-bar-fill" style="width:${p}%"></div></div></td>`;
      el.tableBody.appendChild(tr);
    }
  }

  function runRolls(n, anim=true) {
    n = clamp(Number(n)||1,1,10000);
    let last=null;
    for(let i=0;i<n;i++) last=rollOnce();
    if(anim&&last) animDice(last);
    renderDice();
  }

  function runAnimBatch(n) {
    n = clamp(Number(n)||1,1,10000);
    if(n>25){ runRolls(n,true); return; }
    let done=0;
    const iv = setInterval(()=>{
      const r=rollOnce(); animDice(r); renderDice();
      if(++done>=n) clearInterval(iv);
    },135);
  }

  el.diceHelpBtn.addEventListener("click",()=>el.dicePanel.classList.toggle("is-visible"));
  el.rollOnce.addEventListener("click",()=>runAnimBatch(1));
  el.rollTen.addEventListener("click",()=>runAnimBatch(10));
  el.rollHundred.addEventListener("click",()=>runRolls(100,true));
  el.customBtn.addEventListener("click",()=>runRolls(el.customInput.value,true));
  el.resetDice.addEventListener("click",()=>{
    state.diceCounts=[0,0,0,0,0,0]; state.totalRolls=0; state.lastRoll=null;
    el.diceVal.textContent="?"; renderDice();
  });
  el.copyBtn.addEventListener("click", async ()=>{
    let txt=`Dice Simulation Summary\nTotal rolls: ${state.totalRolls}\n`;
    for(let v=1;v<=6;v++) txt+=`Side ${v}: ${state.diceCounts[v-1]} rolls, ${pct(prob(v-1)*100)}\n`;
    try{ await navigator.clipboard.writeText(txt); el.copyBtn.textContent="Copied!"; setTimeout(()=>el.copyBtn.textContent="Copy Summary",1200); }
    catch(e){ alert(txt); }
  });
  el.newDicePrompt.addEventListener("click",()=>el.dicePrompt.textContent=pick(dicePrompts));

  /* ── FLIGHT ── */
  function getInputs() {
    return {
      weather:    el.weatherIn.value,
      wind:       el.windIn.value,
      visibility: el.visIn.value,
      planeStyle: el.planeStyleIn.value,
      planeAge:   Number(el.planeAgeIn.value),
      maintenance:el.maintIn.value,
      passengers: Number(el.passIn.value),
      pilot:      el.pilotIn.value,
      copilot:    el.copilotIn.value,
    };
  }

  function calcScore(inp) {
    const bd=[];
    let score=100;
    const add=(lbl,amt)=>{ if(amt){ bd.push({lbl,amt}); score-=amt; } };
    add("Weather condition",    {clear:0,rain:5,fog:9,storm:14}[inp.weather]??0);
    add("Wind speed",           {low:0,medium:4,high:10,extreme:16}[inp.wind]??0);
    add("Visibility",           {excellent:0,normal:2,poor:8,dangerous:13}[inp.visibility]??0);
    add("Plane style",          {modern:0,standard:2,small:5,cargo:4,old:8}[inp.planeStyle]??0);
    let age=0; if(inp.planeAge>12)age+=3; if(inp.planeAge>25)age+=5; if(inp.planeAge>40)age+=7;
    add("Plane age",age);
    add("Maintenance quality",  {excellent:0,average:3,delayed:8,poor:13}[inp.maintenance]??0);
    let pax=0; if(inp.passengers>180)pax+=2; if(inp.passengers>260)pax+=3; if(inp.passengers>340)pax+=3;
    add("Passenger load",pax);
    add("Pilot experience",     {expert:0,trained:3,new:10}[inp.pilot]??0);
    add("Copilot experience",   {expert:0,trained:2,new:7}[inp.copilot]??0);
    return { score: clamp(score,18,98), breakdown: bd };
  }

  function renderRisk(bd) {
    el.riskList.innerHTML = "";
    if(!bd.length){ el.riskList.innerHTML='<div class="risk-item"><span>No major risk deductions.</span><strong>-0</strong></div>'; return; }
    bd.forEach(r=>{ const d=document.createElement("div"); d.className="risk-item"; d.innerHTML=`<span>${r.lbl}</span><strong>-${r.amt}</strong>`; el.riskList.appendChild(d); });
  }

  function resetFlightAnim() {
    el.plane.classList.remove("takeoff-success","takeoff-fail","shaking");
    el.peopleCont.innerHTML="";
    el.explosion.classList.remove("show-explosion");
    el.resultBox.className="flight-result-box";
    el.resultBox.textContent="Choose inputs and run the flight simulation.";
    el.safeScore.textContent="--%";
    el.randRoll.textContent="--";
    el.outcome.textContent="Waiting";
  }

  function updateWeather(w) {
    el.flightScene.classList.toggle("stormy", w==="storm"||w==="fog");
  }

  function previewFlight() {
    const inp=getInputs(); updateWeather(inp.weather);
    const r=calcScore(inp); renderRisk(r.breakdown);
    el.safeScore.textContent=pct(r.score);
  }

  el.flightHelpBtn.addEventListener("click",()=>el.flightPanel.classList.toggle("is-visible"));
  el.resetFlight.addEventListener("click", resetFlightAnim);
  el.newFlightPr.addEventListener("click",()=>el.flightPrompt.textContent=pick(flightPrompts));
  el.randomize.addEventListener("click",()=>{
    el.weatherIn.value    = pick(["clear","rain","fog","storm"]);
    el.windIn.value       = pick(["low","medium","high","extreme"]);
    el.visIn.value        = pick(["excellent","normal","poor","dangerous"]);
    el.planeStyleIn.value = pick(["modern","standard","small","cargo","old"]);
    el.planeAgeIn.value   = rndInt(1,55);
    el.maintIn.value      = pick(["excellent","average","delayed","poor"]);
    el.passIn.value       = rndInt(30,360);
    el.pilotIn.value      = pick(["expert","trained","new"]);
    el.copilotIn.value    = pick(["expert","trained","new"]);
    previewFlight();
  });

  [el.weatherIn,el.windIn,el.visIn,el.planeStyleIn,el.planeAgeIn,el.maintIn,el.passIn,el.pilotIn,el.copilotIn]
    .forEach(i=>i&&i.addEventListener("change", previewFlight));

  el.runFlight.addEventListener("click",()=>{
    resetFlightAnim();
    const inp=getInputs(); updateWeather(inp.weather);
    const {score,breakdown}=calcScore(inp);
    const roll=Math.random()*100;
    const ok=roll<=score;
    renderRisk(breakdown);
    el.safeScore.textContent=pct(score);
    el.randRoll.textContent=roll.toFixed(1);
    el.outcome.textContent=ok?"Success":"Failed";
    el.plane.classList.add("shaking");
    setTimeout(()=>{
      el.plane.classList.remove("shaking");
      el.plane.classList.add(ok?"takeoff-success":"takeoff-fail");
      if(!ok){
        const people=["🧍","🧍‍♀️","🧍‍♂️","🧑"];
        el.peopleCont.innerHTML="";
        people.forEach(p=>{ const s=document.createElement("span"); s.className="falling-person"; s.textContent=p; el.peopleCont.appendChild(s); });
        el.explosion.classList.add("show-explosion");
      }
      el.resultBox.className = ok ? "flight-result-box success-result" : "flight-result-box fail-result";
      el.resultBox.innerHTML = ok
        ? `✅ Flight succeeded.<br>Safety score: ${pct(score)} · Uncertainty factor: ${roll.toFixed(1)}<br>Simulations include randomness to model unpredictable real-world factors. Here, uncertainty (${roll.toFixed(1)}) was within the safety margin, so the flight succeeded.`
        : `⚠️ Flight failed.<br>Safety score: ${pct(score)} · Uncertainty factor: ${roll.toFixed(1)}<br>Simulations include randomness to model unpredictable real-world factors. Here, uncertainty (${roll.toFixed(1)}) exceeded the safety margin, so the flight failed.`;
      state.flightHistory.push({inp,score,roll,ok});
    },140);
  });

  /* ── INIT ── */
  renderDice();
  previewFlight();
})();
