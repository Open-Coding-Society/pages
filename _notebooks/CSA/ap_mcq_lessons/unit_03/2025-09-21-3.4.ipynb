## Constructors

A constructor runs the moment you write `new`. It takes your inputs and fills in the object's fields so it comes out ready to use — not blank.

```java
public class Student {
    String name;
    int grade;
    public Student(String name, int grade) {
        this.name  = name;
        this.grade = grade;
    }
}
Student s = new Student("Alex", 11);  // constructor fires here
```

## Constructor Anatomy

| Part | What it is |
|---|---|
| `public Student(...)` | The constructor — must have the same name as the class |
| `String name, int grade` | Parameters — the values you pass in when using `new` |
| `this.name` | The field on the object being built |
| `name` (no `this.`) | The parameter that was passed in |
| `new Student(...)` | The keyword that triggers the constructor |

## 1. LxD Cycle Process

- **Learning Goal:** Students will write a constructor that sets fields using parameters.

## 2. Lesson Plan

**Learning Objective:** Understand what a constructor is, why it exists, and how to write one.

**Success Criteria:** You can write a constructor for a given class that takes parameters and sets fields correctly.

### Tech Talk (3 minutes)

A class is just a blueprint — it describes what an object *will* have. A constructor is what actually builds the object. It runs once, automatically, the moment you write `new`.

Without a constructor, every field is empty (`null` for text, `0` for numbers). The constructor is what fills them in.

✅ **Do this** — let the constructor handle setup:
```java
Student s = new Student("Alex", 11);
```

❌ **Don't do this** — manually setting fields after creation is messy and easy to forget:
```java
Student s = new Student();
s.name  = "Alex";
s.grade = 11;
```

**Class Activity — "You ARE the Object"**

Teacher picks a volunteer and types their name and grade on screen. The class calls `new Student("name", grade)` out loud. The volunteer stands up and says:

> *"My name is [X] and my grade is [Y]!"*

That's exactly what the constructor did — it took the parameters and set the fields. The volunteer IS the object.

### Code Examples

#### A. Simple — the problem without a constructor

````python
public class Student {
    String name;
    int grade;
    // no constructor — fields stay empty
}

Student s = new Student();
System.out.println(s.name);   // null
System.out.println(s.grade);  // 0
````

**Output:**
```
null
0
```

#### B. Intermediate — adding a constructor

````python
public class Student {
    String name;
    int grade;

    // =============================================
    //  CONSTRUCTOR
    //  Purpose: guarantee every Student starts with
    //  a real name and grade, not null and 0
    // =============================================
    public Student(String name, int grade) {
        this.name  = name;
        this.grade = grade;
    }

}

Student s = new Student("Alex", 11);
System.out.println(s.name);
System.out.println(s.grade);
````

**Output:**
```
Alex
11
```

#### C. Complex — two objects from the same constructor

````python
// The same constructor builds every Student —
// each object gets its own copy of the fields.

Student s1 = new Student("Alex", 11);
Student s2 = new Student("Jordan", 10);

System.out.println(s1.name + " " + s1.grade);
System.out.println(s2.name + " " + s2.grade);
````

**Output:**
```
Alex 11
Jordan 10
```
