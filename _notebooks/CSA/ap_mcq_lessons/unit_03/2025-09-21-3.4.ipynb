## Constructors

A constructor runs automatically when you write `new`. It sets the object's starting values.

> **Pizza analogy:** `new Student("Alex", 11)` → you pass in toppings (params), the kitchen (constructor) builds your pizza (object).

```python
public class Student {

    String name;   // field
    int grade;     // field

    // constructor — same name as the class, runs on "new"
    public Student(String name, int grade) {
        this.name  = name;   // "this." = the field on THIS object
        this.grade = grade;
    }
}

Student s = new Student("Alex", 11);
System.out.println(s.name);    // Alex
System.out.println(s.grade);   // 11
```

function conBuild() {
    var name  = document.getElementById('conName').value.trim();
    var grade = document.getElementById('conGrade').value.trim();
    if (!name || !grade) { alert('Fill in both the name and grade first!'); return; }

    document.getElementById('conVName').textContent  = '"' + name + '"';
    document.getElementById('conVGrade').textContent = grade;
    document.getElementById('conSName').textContent  = name;

        document.getElementById('conFName').classList.add('show');

        document.getElementById('conFGrade').classList.add('show');
    }, 950);

        var s = document.getElementById('conShout');
        s.style.display = 'block';
        setTimeout(function() { s.classList.add('show'); }, 50);
        document.getElementById('conReset').style.display = 'block';
    }, 1700);
}
function conReset() {
    ['conName','conGrade'].forEach(function(id){ document.getElementById(id).value = ''; });
    document.getElementById('conObj').style.display = 'none';
    document.getElementById('conFName').classList.remove('show');
    document.getElementById('conFGrade').classList.remove('show');
    var s = document.getElementById('conShout');
    s.classList.remove('show');
    s.style.display = 'none';
    document.getElementById('conReset').style.display = 'none';
}
</script>

```python
// no constructor written → Java gives you a free no-args default
Student s1 = new Student();
System.out.println(s1.name);   // null  ← not useful

// write your own → you control the starting values
Student s2 = new Student("Alex", 11);
System.out.println(s2.name);   // Alex
```

## Key Vocab

- **constructor** — method that runs on `new`, sets starting values
- **parameter** — value you pass in when creating the object
- **field** — variable that belongs to the object (every instance gets its own)
- **instantiation** — creating an object from a class
- **`new`** — keyword that builds the object and calls the constructor
