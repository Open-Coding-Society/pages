## Step 1: A class is just a blueprint

This is a `Student` class. It says every student *will have* a name and a grade. But at this point, it's just a description sitting in memory. No student exists yet.

````python
public class Student {
    String name;
    int grade;
}
````

## Step 2: So what's the problem?

Try creating one without a constructor:

```java
Student s = new Student();
```

It compiles. But `s.name` is `null` and `s.grade` is `0`. The object was created with nothing in it — because nothing told it what values to start with.

That's the whole point of a constructor. **Its job is to make sure every object starts out with real values, not empty ones.** You write it once, and every time someone calls `new`, the object comes out ready to use.

## Step 3: Add a constructor

The highlighted block below is the constructor. Notice it has the same name as the class. That's how Java knows it's a constructor and not a regular method.

````python
public class Student {
    String name;
    int grade;

    // =============================================
    //  CONSTRUCTOR
    //  Purpose: guarantee every Student starts with
    //  a real name and grade, not null and 0
    // =============================================
    public Student(String name, int grade) {
        this.name  = name;   // this.name  = the field | name  = what was passed in
        this.grade = grade;  // this.grade = the field | grade = what was passed in
    }
    // =============================================

}

// Every time you write "new", the constructor above runs.
// It takes your inputs and fills in the fields before handing you the object.
Student s = new Student("Alex", 11);
System.out.println(s.name);   // Alex  (not null)
System.out.println(s.grade);  // 11    (not 0)
````

## What if you skip the constructor?

Java won't crash. It just makes a blank object with nothing filled in.

Think of it like ordering a pizza but never picking your toppings. You still get a box — it's just empty inside.

That's why you write a constructor: so the object comes out with real values already in it, the moment it's created.

````python
// no constructor written — object is empty
Student s1 = new Student();
System.out.println(s1.name);   // null

// constructor written — object is ready to use
Student s2 = new Student("Alex", 11);
System.out.println(s2.name);   // Alex
````

## Key Vocab

- **constructor** — code that runs on `new` and sets the object's starting values
- **field** — a variable that belongs to the object (`name`, `grade`)
- **parameter** — the value you pass in when creating the object (`"Alex"`, `11`)
- **`this.`** — refers to the specific object being built right now
- **`new`** — triggers the constructor and hands back a finished object
