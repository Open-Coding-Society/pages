### What Are Toggles Good For?

Toggles let users quickly turn a setting on or off. They are commonly used in apps and websites for things like:

- Notifications
- Dark Mode
- Privacy settings
- Battery settings
- Website features

Toggles work best when there are two choices, such as on/off or enabled/disabled.

### Toggle in a UI Runner!



````python
%%html
<!-- UI_RUNNER: Toggle example | show_code: true -->
<!-- //Fill in the blanks! -->
 
<style>
.switch {
    position: relative;
    display: flex;
    width: 54px;
    height: 31px;
}

.switch input[type="checkbox"] {
    opacity: 0;
    width: 0;
    height: 0;
}

.slider {
    position: absolute;
    inset: 0;
    border: 1px solid var(--ui-border, #666);
    border-radius: 100px;
    background: var(--panel-mid, gray);
    cursor: pointer;
}

.slider:before {
    content: "";
    position: absolute;
    width: 23px;
    height: 23px;
    top: 3px;
    left: 4px;
    bottom: 0;
    border-radius: 50%;
    background: var(--pref-text-color, white);
    transition: transform 0.2s;
}

input[type="checkbox"]:checked + .slider {
    background: var(--pref-accent-color, blue);
}

input[type="checkbox"]:checked + .slider:before {
    transform: translateX(22px);
    outline: 2px solid var(--pref-accent-color, blue);
}
</style>

<h3>Checkbox</h3>
<label>
  <input type="checkbox">
</label>

<h3>Slider</h3>
<label class="switch">
  <input type="checkbox">
  <span class="slider" aria-hidden="true"></span>
  <span>Toggle</span>
</label>
````

   - The `.green` class can change the checked color to `$green1`.

7. Add Focus and Transitions
   - Add a focus shadow with `&:focus + .slider`.
   - Use `transition` to make the toggle changes smooth.

## The Beauty of SASS

To use checkboxes or sliders all you need to do is use the class from SASS!  This lets you focus on logic instead of repeating style over and over.

````python
%%html
<!-- UI_RUNNER: Toggle example using SASS | show_code: true -->

<label class="ocs__toggle ocs__toggle--green">
  <input id="feature-toggle" class="ocs__toggle-input" type="checkbox" checked>
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">My Label</span>
</label>

Status: <output id="feature-toggle-status" aria-live="polite"></output>

<script>
{
  const toggle = document.getElementById("feature-toggle");
  const status = document.getElementById("feature-toggle-status");

  const updateStatus = () => {
    status.value = String(toggle.checked);
  };

  toggle.addEventListener("change", updateStatus);
  updateStatus();
}
</script>
````

## Implementation
### Step One 
First, create the `.switch` class. This is the main container for the toggle and controls where it sits and how large it is.

- `position: relative` sets the toggle as a reference point for the slider.
- `width` and `height` control the size of the toggle.
- `padding` adds space inside the toggle.
- `justify-content` and `align-items` help position the toggle's contents.


```scss
.switch {
    position: relative;
    display: flex;
    width: 54px;
    height: 31px;
    padding: 2px 2px 2px 6px;
    justify-content: flex-end;
    align-items: center;
}
```

### 2. Set Up the Checkbox

Next, style the checkbox that controls whether the toggle is on or off.

- `opacity: 0` makes the checkbox invisible.
- `width: 0` and `height: 0` hide the checkbox.
- The checkbox still controls the toggle even though the user sees the slider instead.

```scss
input[type="checkbox"] {
    opacity: 0;
    width: 0;
    height: 0;
}
```


### 3. Slider
The `.slider` is the part of the toggle that the user actually sees.

- `position: absolute` lets the slider fill the switch container.
- `inset: 0` makes it fit inside the switch.
- `border-radius: 100px` gives it rounded edges.
- `background: $gray4` makes the toggle gray when it is off.
- `cursor: pointer` makes it look clickable.
- `transition: 0.4s` makes changes happen smoothly.



```scss
.slider {
    position: absolute;
    cursor: pointer;
    inset: 0;
    border-radius: 100px;
    background: $gray4;
    transition: 0.4s;
}
```

### 4. Create the Slider Button

The `&:before` creates the white circle inside the slider.

- `content: ""` allows the circle to be created.
- `width` and `height` set the size of the circle.
- `border-radius: 50%` makes it circular.
- `background: #fff` makes the circle white.
- `left` and `bottom` position the circle inside the slider.
- `box-shadow` adds a small shadow around the circle.
- `transition: 0.4s` makes the circle move smoothly.

```scss
&:before {
    content: "";
    position: absolute;
    width: 27px;
    height: 27px;
    left: 4px;
    bottom: 4px;
    border-radius: 50%;
    background: #fff;
}
```

### 5. Add the Checked State

Now, make the toggle look different when the checkbox is checked.

- `&:checked+.slider` selects the slider when the checkbox is checked.
- `background: $blue1` changes the slider to blue.
- `&.green` allows the slider to use `$green1` when the `green` class is added.

```scss
&:checked + .slider {
    background: $blue1;

    &.green {
        background: $green1;
    }
}
```

### 6. Move the Slider Button

When the checkbox is checked, move the white circle to the other side.

- `transform: translateX(27px)` moves the circle 27 pixels to the right.
- This makes it clear that the toggle has been turned on.

```scss
&:checked + .slider:before {
    transform: translateX(27px);
}
```

### 7. Add Focus and Transitions

Lastly, add a focus effect and make the toggle move smoothly.

- `&:focus+.slider` adds a small shadow when the checkbox is focused.
- `$blueShadow1` sets the color of the focus shadow.
- `transition: 0.4s` makes the toggle's changes look smooth instead of happening instantly.
---

```scss
&:focus + .slider {
    box-shadow: 0 0 1px $blueShadow1;
}
```
