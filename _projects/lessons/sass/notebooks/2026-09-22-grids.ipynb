### Try changing the `fr` to see what happens

Under the hood, OCS grid classes use a unit called `fr` (short for **fraction**) to size columns. Each `fr` value makes one column. For example, `cols-2` is the same as `1fr 1fr`.

The runner below has 2 beakers in a row using `grid-template-columns: 1fr 1fr`. Change the `fr` values and press **Run**:

1. Change `1fr 1fr` to `2fr 1fr`. Still 2 boxes, but Beaker 1 is now **twice as wide**.
2. Change it to `1fr 3fr`. Beaker 2 now gets 3 of the 4 shares.
3. Change it to `1fr 1fr 1fr` and add a third beaker cell. Now there are **3 boxes** in the row.

**Remember:** adding an `fr` value adds a box. A bigger `fr` number makes that box wider. On real pages, let the OCS classes write this for you.

### Grid Structure

Every OCS grid is built from the same layers, from the outside in.

| Class | What it is | Lab tray example |
| --- | --- | --- |
| `ocs__card` | A framed panel around the grid | The lab bench the tray sits on |
| `ocs__grid` | The grid itself. Always the outer element | The lab tray |
| `ocs__grid--*` | The grid **variant**. Pick exactly one | What kind of tray it is |
| `ocs__grid-cell` | One box. Every item inside the grid gets this | One test beaker |
| `ocs__grid-cell--*` | A **cell type** that changes one box | A label on one beaker |
| `<h3>` | A title inside a cell | "Hypothesis" |
| `<p>` | Normal text inside a cell | A beaker reading |

### OCS Grid Grammar

These are all the classes you can use. You never need to write your own CSS.

| Class | What it does |
| --- | --- |
| `ocs__grid--standard` | 3 boxes per row, same as `1fr 1fr 1fr` (2 on phones) |
| `cols-2`, `cols-4` | Changes a standard grid to 2 or 4 boxes per row (`1fr 1fr` or four `1fr`) |
| `ocs__grid--card` | Fits as many cards as it can, each at least 200px wide |
| `cols-3` | Makes a card grid exactly 3 cards per row |
| `ocs__grid--color` | Each box shows its own color when you hover |
| `ocs__grid--holographic` | Boxes cycle through rainbow colors in a wave |
| `ocs__grid--calculator` | 4 square buttons per row, like a keypad |
| `ocs__grid--gallery` | Fits as many square tiles as it can |
| `ocs__grid-cell--header` | Stretches across the **whole row** (a title) |
| `ocs__grid-cell--wide` | Stretches across **2 columns** |
| `ocs__grid-cell--accent` | Accent color so it stands out (an outlier beaker) |
| `ocs__grid-cell--muted` | Faded so it stays in the background (the control beaker) |

## 1. LxD Cycle Process

**Empathize:** When students post lab results, they often build the layout by hand with inline styles (like `<div style="display:flex; width:33%">`) or made up classes like `my-row`. These layouts break on phones, ignore the site theme, and look different on every page.

**Define:**

* **POV:** CSP students need a simple, reusable way to put lab data into rows and columns, because hand written layout CSS breaks on small screens.
* **Learning Goal:** Students will build grids using only OCS classes: `ocs__grid`, one variant, `ocs__grid-cell`, and cell types.

**Ideate:**

* **HMW Question:** How might we help students trust the OCS grid classes instead of writing their own layout CSS?
* **Activity:** Change `fr` values and OCS classes in live runners, then fix a badly built layout using only OCS classes.

**Prototype & Test:** We ran a practice version with our project team. They said the first homework used too many grid variants at once, so we cut it down to two grids: one standard grid of beaker readings and one calculator keypad.

## 2. Lesson Plan

**Learning Objective:** By the end of this lesson, you will be able to lay out lab data with OCS grid classes so it works on phones and follows the site theme, without writing any CSS.

**Success Criteria:** You can take a badly built layout, replace it with the right grid, variant, cells, and cell types, and it looks right on both a computer and a phone.

### Tech Talk (3 minutes)

The site already has a shared SASS grid system in `mixins/_grid.scss`. It handles columns, spacing, phone sizes, hover effects, and theme colors. **You just pick the class that describes the grid you want.**

* **The Rule:** Use OCS grid classes. Let the system handle the layout.
* ✅ **Do this:** `<div class="ocs__grid ocs__grid--standard">`
* ❌ **Don't do this:** `<div style="display:grid; grid-template-columns:1fr 1fr 1fr">` or `<div class="my-row">`

**Why?** Every lab page looks the same, your beaker data fits on phones automatically, and it follows each user's color settings.
