# Authoring skeleton - backend focus (CS113 backend graphs)

## Three short topics
1. Model dependencies (tasks, assignments, services) as a directed graph.
2. Adjacency list representation in Java.
3. Traversal (BFS/DFS) and ordering with cycle detection.

## PBL context
Represent dependencies or routes as a graph and compare traversal and path-finding strategies. Context: an ordered set of assignments or service calls.

## Backend layers to author
- Build an adjacency list from a small dependency list.
- BFS/DFS to find everything a node depends on.
- Topological order and cycle detection.
- Directed/weighted edge tests.

## Evidence checklist to author
