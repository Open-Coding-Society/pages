# Authoring skeleton - backend focus (CS113 backend structures)

## Three short topics
1. Set for deduplication of incoming records.
2. Map for keyed lookup and missing-key behavior.
3. Graph for relationships between records.

## PBL context
Use a set for deduplication, a map for keyed records, and a graph for relationships; explain each choice. Context: a small API service handling a batch of records.

## Backend layers to author
- Deduplicate a batch with a set; define equals/hashCode behavior.
- Index records by key in a map; define missing-key policy.
- Link related records in a graph and query neighbors.
- Justify each representation with expected lookup costs.

## Evidence checklist to author
Test duplicate records, missing keys, equal keys with different values, and disconnected graphs; verify counts and lookups.
