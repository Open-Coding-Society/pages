## 1. Reference Guide

### Key Topics

<br>

| Term | Definition | Example |
|---|---|---|
| **Linear search** | Check each element in order until a match is found or the list ends. | `for (int i = 0; i < arr.length; i++)` |
| **Sentinel** | A value that cannot be a real index, returned for "not found". | `return -1;` |
| **`==` vs `.equals()`** | `==` for primitives; `.equals()` for objects such as `String`. | `roster[i].equals(name)` |
| **Logical size** | How many slots hold real data, which may be less than `.length`. | `i < numSignedIn` |

- Linear search works on any list, sorted or not. That generality is why it can't be fast: the best case is O(1) and the worst case is O(n).
- Searching by a condition uses the same loop as searching by equality. Only the `if` changes.
- "First match" and "most extreme value" are different questions, and both compile.

### Picking an Algorithm

- **First match** → `return i` inside the `if`, and `return -1` only after the loop
- **Every match** → collect into an `ArrayList<Integer>` and drop the early `return`
- **Partially filled array** → loop to the logical size, never a blind `.length`
- **Compare every pair** → nested loop with `j` starting at `i + 1`; it is O(n²)

## 5. Code Examples

### A. Comparing names.  The NameSearch class searches the same roster with `.equals()` and with `==`

Common starting mistake:

<pre><code class="language-java">
if (roster[i] == targetName)        // same object in memory?
if (roster[i].equals(targetName))   // same text?
</code></pre>

Terminology:

`int` is a primitive, so `==` compares values. `String` is an object, so `==` compares references. Anything capitalized uses `.equals()`.

---

````java
// CODE_RUNNER: Predict both results before running, then change new String("Marcus Webb") to just "Marcus Webb"
// == vs. .equals() when searching names
public class NameSearch {
    public static int indexWithEquals(String[] roster, String target) {
        for (int i = 0; i < roster.length; i++) {
            if (roster[i].equals(target)) return i;
        }
        return -1;
    }

    public static int indexWithDoubleEquals(String[] roster, String target) {
        for (int i = 0; i < roster.length; i++) {
            if (roster[i] == target) return i;
        }
        return -1;
    }

    public static void main(String[] args) {
        String[] names = {"Maya Chen", "Diego Ruiz", "Marcus Webb"};
        String typed = new String("Marcus Webb");   // same text, different object

        System.out.println(".equals: " + indexWithEquals(names, typed));
        System.out.println("==:      " + indexWithDoubleEquals(names, typed));
    }
}

NameSearch.main(null);
````

### B. First match or every match.  The FirstOrEvery class finds failing quiz scores two ways

Common starting mistake:

<pre><code class="language-java">
if (scores[i] < threshold) return i;        // stops at the first match
if (scores[i] < threshold) matches.add(i);  // keeps going and collects them all
</code></pre>

Terminology:

An early `return` means "find the first". No early `return` means "find every". The first failing score is not the lowest one: Diego Ruiz (54) comes before Priya Nair (46).

---

````java
// CODE_RUNNER: Predict both outputs before running, then change the threshold to 50
// First match vs. every match
import java.util.ArrayList;

public class FirstOrEvery {
    public static int indexOfFirstBelow(int[] scores, int threshold) {
        for (int i = 0; i < scores.length; i++) {
            if (scores[i] < threshold) return i;
        }
        return -1;
    }

    public static ArrayList<Integer> indexesBelow(int[] scores, int threshold) {
        ArrayList<Integer> matches = new ArrayList<Integer>();
        for (int i = 0; i < scores.length; i++) {
            if (scores[i] < threshold) matches.add(i);
        }
        return matches;
    }

    public static void main(String[] args) {
        String[] names = {"Maya Chen", "Diego Ruiz", "Aaliyah Brooks", "Sam Okafor", "Jordan Blake",
                          "Priya Nair", "Liam Carter", "Zoe Fischer", "Noah Kim", "Ines Delgado",
                          "Marcus Webb", "Grace Liu"};
        int[] quizScores = {87, 54, 92, 71, 88, 46, 79, 95, 62, 83, 58, 90};

        int first = indexOfFirstBelow(quizScores, 60);
        System.out.println("first: " + names[first] + " (" + quizScores[first] + ")");

        System.out.println("every:");
        for (int i : indexesBelow(quizScores, 60)) {
            System.out.println("  " + names[i] + " (" + quizScores[i] + ")");
        }
    }
}

FirstOrEvery.main(null);
````

### C. The sign-in sheet.  The SignInSheet class searches a partly filled array two ways

Common starting mistake:

<pre><code class="language-java">
for (int i = 0; i < sheet.length; i++)         // includes the unused slots
for (int i = 0; i < numSignedIn; i++)          // only real check-ins
</code></pre>

Terminology:

`.length` is the **capacity**. The **logical size** is how many slots hold real data. Unused slots hold Java's default `0`, which a search can mistake for a real ID.

---

````java
// CODE_RUNNER: Predict both lines before running, then also search for 40940 in both
// Searching to .length vs. the logical size
public class SignInSheet {
    public static boolean searchToLength(int[] sheet, int target) {
        for (int i = 0; i < sheet.length; i++) {
            if (sheet[i] == target) return true;
        }
        return false;
    }

    public static boolean searchToSize(int[] sheet, int numSignedIn, int target) {
        for (int i = 0; i < numSignedIn; i++) {
            if (sheet[i] == target) return true;
        }
        return false;
    }

    public static void main(String[] args) {
        int[] sheet = new int[15];
        int numSignedIn = 8;
        int[] realIDs = {40217, 40593, 40108, 40772, 40364, 40881, 40255, 40940};
        for (int i = 0; i < realIDs.length; i++) sheet[i] = realIDs[i];

        System.out.println("to length, ID 0: " + searchToLength(sheet, 0));
        System.out.println("to size,   ID 0: " + searchToSize(sheet, numSignedIn, 0));
    }
}

SignInSheet.main(null);
````

````java
// CODE_RUNNER: Replace the sample values with your own, then run it
// Practice #1 - Try it yourself first!
public class MyStudyLog {
    public static int indexOfFirstAtLeast(int[] arr, int min) {
        for (int i = 0; i < arr.length; i++) {
            if (arr[i] >= min) return i;
        }
        return -1;
    }

    public static int indexOf(int[] arr, int target) {
        for (int i = 0; i < arr.length; i++) {
            if (arr[i] == target) return i;
        }
        return -1;
    }

    public static void main(String[] args) {
        // Try writing your own code here first!

        // Sample answer:
        int[] minutes = {95, 140, 80, 180, 120};

        System.out.println("First day at 120+: index " + indexOfFirstAtLeast(minutes, 120));
        System.out.println("A day of exactly 50: index " + indexOf(minutes, 50));
    }
}

MyStudyLog.main(null);
````
