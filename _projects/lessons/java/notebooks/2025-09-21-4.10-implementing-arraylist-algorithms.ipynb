## 1. Reference Guide

### Key Topics

<br>

| Term | Definition | Example |
|---|---|---|
| **Traversal** | Visiting every element of a collection, usually with a loop. | `for (int i = 0; i < nums.size(); i++)` |
| **Accumulator** | A variable updated across iterations to build a running result. | `max`, `count`, `kept` |
| **indexOf(obj)** | Returns the index of the first match, or `-1` if none. | `roster.indexOf("Amir")` |
| **contains(obj)** | Returns `true` if any element matches. | `roster.contains("Jordan")` |
| **ConcurrentModificationException** | Runtime error from adding/removing during an enhanced `for` loop. | Removing inside `for (int n : nums)` |

- The traversal shape is the same as for arrays; only the bound (`size()`) and the access (`get(i)`) change.
- `indexOf` and `contains` compare with `.equals`, so they match content, not object identity.

### Safe vs. Unsafe Traversal for Removal

| Traversal | Safe to remove while traversing? |
|---|---|
| Indexed `for` loop, forward (`i` increasing) | Only one removal, or accept skipped elements |
| Indexed `for` loop, backward (`i` decreasing to `0`) | Yes — earlier indices are never shifted |
| Enhanced `for` loop (`for (int n : nums)`) | No — throws `ConcurrentModificationException` if the list changes size |

### Picking an Algorithm

- **Largest or smallest value** → start at `nums.get(0)`, loop from index `1`
- **Is a value there? Where?** → `contains(obj)` or `indexOf(obj)`
- **How many match?** → a counting loop with `count++`
- **Keep only some values** → build a new list with `add`
- **Delete values in place** → indexed loop from `size() - 1` down to `0`

### Tech Talk (5 minutes)

The traversal shape carries over directly from arrays; only the bound and the access method change:

```java
for (int i = 0; i < scores.size(); i++) {   // size(), not .length

}
```

### A. Finding a maximum.  The MaxFinder class compares a correct start to a buggy one

Common starting mistake:

<pre><code class="language-java">
int max = 0;             // assumes some element is at least 0
int max = nums.get(0);   // starts with a real element, then loop from index 1
</code></pre>

Terminology:

`max` is an **accumulator**: it holds the best value seen so far. Starting it at a real element works for any list, even one where every value is negative.

---

````java
// CODE_RUNNER: Predict both results before running, then add -1 to the negatives list and run again
// Correct vs. buggy maximum
import java.util.ArrayList;

public class MaxFinder {
    public static int maxValue(ArrayList<Integer> nums) {
        int max = nums.get(0);
        for (int i = 1; i < nums.size(); i++) {
            if (nums.get(i) > max) {
                max = nums.get(i);
            }
        }
        return max;
    }

    public static int maxValueBuggy(ArrayList<Integer> nums) {
        int max = 0;   // bug: 0 may not be in the list
        for (int i = 0; i < nums.size(); i++) {
            if (nums.get(i) > max) {
                max = nums.get(i);
            }
        }
        return max;
    }

    public static void main(String[] args) {
        ArrayList<Integer> negatives = new ArrayList<Integer>();
        negatives.add(-4);
        negatives.add(-9);
        negatives.add(-2);

        System.out.println("maxValue = " + maxValue(negatives));
        System.out.println("maxValueBuggy = " + maxValueBuggy(negatives));
    }
}

MaxFinder.main(null);
````

### B. Searching and counting.  The RosterSearch class uses indexOf, contains, and a counting loop

Common search mix-up:

<pre><code class="language-java">
roster.indexOf("Amir")         // position of the FIRST match, or -1
roster.contains("Amir")        // true or false, stops at the first match
countMatches(roster, "Amir")   // a loop that checks EVERY element
</code></pre>

Terminology:

`indexOf` and `contains` stop at the first match, so they can't tell you how many matches there are. Counting needs a loop, and the loop compares Strings with `.equals`, not `==`.

---

````java
// CODE_RUNNER: Run it, then search for "Jordan" in all three lines and explain each result
// Searching and counting
import java.util.ArrayList;

public class RosterSearch {
    public static int countMatches(ArrayList<String> names, String target) {
        int count = 0;
        for (int i = 0; i < names.size(); i++) {
            if (names.get(i).equals(target)) {   // .equals compares content
                count++;
            }
        }
        return count;
    }

    public static void main(String[] args) {
        ArrayList<String> roster = new ArrayList<String>();
        roster.add("Amir");
        roster.add("Priya");
        roster.add("Amir");

        System.out.println("indexOf Amir = " + roster.indexOf("Amir"));
        System.out.println("contains Jordan = " + roster.contains("Jordan"));
        System.out.println("count of Amir = " + countMatches(roster, "Amir"));
    }
}

RosterSearch.main(null);
````

### C. Building a filtered list.  The ScoreFilter class keeps only the values that pass a test

Filtering pattern:

<pre><code class="language-java">
ArrayList&lt;Integer&gt; kept = new ArrayList&lt;Integer&gt;();   // new list to fill
for (int num : nums) {                                  // read the original
    if (num >= threshold) { kept.add(num); }            // add to the NEW list
}
</code></pre>

Terminology:

An enhanced `for` loop is safe here because it only reads `nums`. The original list keeps its size and order; only `kept` changes.

---

````java
// CODE_RUNNER: Run it, then change the threshold to 9 and predict the new list
// Filtering into a new list
import java.util.ArrayList;

public class ScoreFilter {
    public static ArrayList<Integer> keepAtLeast(ArrayList<Integer> nums, int threshold) {
        ArrayList<Integer> kept = new ArrayList<Integer>();
        for (int num : nums) {
            if (num >= threshold) {
                kept.add(num);
            }
        }
        return kept;
    }

    public static void main(String[] args) {
        ArrayList<Integer> nums = new ArrayList<Integer>();
        nums.add(3);
        nums.add(8);
        nums.add(1);
        nums.add(10);
        nums.add(6);

        System.out.println("kept = " + keepAtLeast(nums, 6));
        System.out.println("original = " + nums);
    }
}

ScoreFilter.main(null);
````

### D. The removal trap.  The RemoveTwos class removes while traversing, forward and backward

Common removal mistake:

<pre><code class="language-java">
for (int i = 0; i < list.size(); i++)        // forward: skips the element after each removal
for (int i = list.size() - 1; i >= 0; i--)   // backward: never skips
</code></pre>

Terminology:

`remove(i)` shifts every later element one place left. Going forward, the next element slides into index `i` right as the loop moves on to `i + 1`, so it never gets checked. Going backward, a removal only shifts elements the loop has already visited.

---

````java
// CODE_RUNNER: Predict both lists before running, then trace the forward loop on paper to find the skipped 2
// Forward vs. backward removal
import java.util.ArrayList;

public class RemoveTwos {
    public static ArrayList<Integer> makeList() {
        ArrayList<Integer> values = new ArrayList<Integer>();
        values.add(5);
        values.add(2);
        values.add(2);
        values.add(9);
        return values;
    }

    public static void main(String[] args) {
        ArrayList<Integer> forward = makeList();
        for (int i = 0; i < forward.size(); i++) {
            if (forward.get(i) == 2) {
                forward.remove(i);
            }
        }
        System.out.println("forward  = " + forward);

        ArrayList<Integer> backward = makeList();
        for (int i = backward.size() - 1; i >= 0; i--) {
            if (backward.get(i) == 2) {
                backward.remove(i);
            }
        }
        System.out.println("backward = " + backward);
    }
}

RemoveTwos.main(null);
````
