## Idea 1: Reuse It

<div class="ocs__card">
<span class="ocs__badge">Tech Talk · Idea 1 of 3</span>
<p class="ocs__lead"><strong>A class is a reference. Pasted CSS is a copy.</strong></p>
<div class="ocs__keypoints">
<div class="ocs__keypoint"><span class="ocs__check">✓</span><span><code>style="..."</code> pastes a copy of a look into one tag. The next tag needs its own copy.</span></div>
<div class="ocs__keypoint"><span class="ocs__check">✓</span><span><code>class="ocs__btn medium"</code> points at one look that lives in one SASS file. Every tag that names it shares it.</span></div>
<div class="ocs__keypoint"><span class="ocs__check">✓</span><span>Change the shared look once and every page updates. Change a copy and only that one tag changes.</span></div>
</div>
<div class="ocs__callout">Pasted CSS adds up. More than a quarter of the page files in the OCS pages repo contain some: 217 of 788 when we counted on October 1, 2026, with at least 571 different hex colors.</div>
</div>

The OCS SASS guide, `_sass/open-coding/README.md`, sets the goal: "OCS grammar is intended to help you build interesting pages with zero custom CSS." Today you will see why.

<details markdown="1">
<summary>How we counted</summary>

We checked every `.md` and `.ipynb` file that Git tracks under `_notebooks`, `_posts`, `_projects`, and `navigation` on the main branch of the OCS pages repo: 788 files. Copies made by the site build are not tracked, so they are not counted. A file counts if it has a `style="..."` attribute or a `<style>` block anywhere, even inside a code example. Colors are the six-digit hex codes inside that CSS, like `#3b82f6`.

</details>

### Java Twin: You Already Know References

In Java, "two different variables can hold references to the same object" (College Board, 2025, 2.6.B.1). Change the object through one variable, and the other variable sees the change.

A `class` attribute works like a shared reference. Pasted CSS works like giving every tag its own separate object. **Predict both lines before you press Run.**

```java
// CODE_RUNNER: Predict both lines, then run. Then change inlineButtons[0] to inlineButtons[2] and predict again.
public class ReferenceVsCopy {
    public static void main(String[] args) {
        // class="..." : three buttons share ONE Look object
        Look shared = new Look("blue");
        Look[] classButtons = {shared, shared, shared};

        // style="..." : three buttons, three separate copies
        Look[] inlineButtons = {new Look("blue"), new Look("blue"), new Look("blue")};

        shared.setColor("green");            // edit the shared look once
        inlineButtons[0].setColor("green");  // edit one copy

        System.out.print("class buttons:  ");
        for (Look button : classButtons) {
            System.out.print(button.getColor() + " ");
        }
        System.out.print("\ninline buttons: ");
        for (Look button : inlineButtons) {
            System.out.print(button.getColor() + " ");
        }
        System.out.println();
    }
}

// A tiny POJO: one private field, a constructor, a getter, and a setter.
// Not public: a .java source can have only one public class, and that is ReferenceVsCopy.
class Look {
    private String color;

    public Look(String color) {
        this.color = color;
    }

    public String getColor() {
        return color;
    }

    public void setColor(String color) {
        this.color = color;
    }
}
ReferenceVsCopy.main(null);
```

**Notice:** all three class buttons turned green from one `setColor` call, because they hold the same reference. Only one inline button changed, because each one is its own object. Change `.ocs__btn` in one SASS file, and every `ocs__btn` button on OCS changes the same way.

## Idea 2: Theme It

<div class="ocs__card">
<span class="ocs__badge">Tech Talk · Idea 2 of 3</span>
<p class="ocs__lead"><strong>One value from a Java POJO colors every component that reads it.</strong></p>
<div class="ocs__text">Pick a theme in Dashboard → Preferences, and your accent color takes a three-step trip.</div>
<div class="ocs__keypoints">
<div class="ocs__keypoint"><span class="ocs__check">1</span><span><strong>Java stores it.</strong> The Spring backend, the Java server behind OCS, keeps your theme in a POJO named <code>UserPreferences</code>, in a field named <code>accentColor</code>.</span></div>
<div class="ocs__keypoint"><span class="ocs__check">2</span><span><strong>The page turns it into a CSS variable.</strong> Dashboard → Preferences gets your theme from the backend and saves a copy in your browser. Every OCS page reads that copy and puts the color in <code>--pref-accent-color</code>.</span></div>
<div class="ocs__keypoint"><span class="ocs__check">3</span><span><strong>Classes read the variable.</strong> <code>.ocs__btn</code> uses <code>var(--pref-accent-color)</code>, so every component that reads it changes at once.</span></div>
</div>
</div>

**Step 1, the POJO**, trimmed from `UserPreferences.java` in the OCS Spring backend (comments added):

~~~java
@Data      // Lombok writes the getters and setters when this compiles
@Entity    // Spring saves this object in the database
public class UserPreferences {
    // ...
    private String accentColor = "#3b82f6";
    // ...more fields, like backgroundColor, textColor, and fontSize
}
~~~

**Step 3, the class**, from `_sass/open-coding/elements/buttons/_api.scss`:

~~~scss
.ocs__btn {
    @include text-button;
    --ocs-btn-tone: var(--pref-accent-color);
    border-color: var(--pref-accent-color);
    text-decoration: none;
}
~~~

<details markdown="1">
<summary>See every file on the trip</summary>

1. [`UserPreferences.java`](https://github.com/Open-Coding-Society/spring/blob/master/src/main/java/com/open/spring/mvc/userPreferences/UserPreferences.java) stores `private String accentColor = "#3b82f6";`.
2. [`UserPreferencesApiController.java`](https://github.com/Open-Coding-Society/spring/blob/master/src/main/java/com/open/spring/mvc/userPreferences/UserPreferencesApiController.java) answers `GET /api/user/preferences`. It builds JSON from the POJO's getters, like `getAccentColor()`, which Lombok wrote. An API is a URL the page can ask for data. JSON is text that lists an object's fields.
3. [`assets/js/pages/dashboard/preferences.js`](https://github.com/Open-Coding-Society/pages/blob/main/assets/js/pages/dashboard/preferences.js) runs on Dashboard → Preferences. It reads the JSON, `accent: backendPrefs.accentColor || d.accent,` (the `||` means "use the saved color, or a default if there is none"), and saves a copy in your browser.
4. [`assets/js/user-preferences.js`](https://github.com/Open-Coding-Society/pages/blob/main/assets/js/user-preferences.js) runs on every page. It reads that copy and sets the CSS variable: `set('--pref-accent-color', accent);`.
5. [`_sass/open-coding/elements/buttons/_api.scss`](https://github.com/Open-Coding-Society/pages/blob/main/_sass/open-coding/elements/buttons/_api.scss) is where `.ocs__btn` reads `var(--pref-accent-color)`.

</details>

<div class="ocs__callout">A border or background color typed as a hex code never asks the POJO, so it looks the same in every theme. Themes let students make OCS easier to read, like a light background or a bigger font, and a typed color ignores that choice. That is why the impostor did not change.</div>

Not every class follows your theme, and that is on purpose. Alert tones like `alert-red` use site colors such as `--red`, which are the same for everyone, because red should always mean danger. They are still CSS variables, defined once for the whole site.

## Idea 3: Generate It

<div class="ocs__card">
<span class="ocs__badge">Tech Talk · Idea 3 of 3</span>
<p class="ocs__lead"><strong>Need a new style? Change the data, not the page.</strong></p>
<div class="ocs__text">Nobody typed <code>.ocs__btn.small</code>, <code>.medium</code>, and <code>.large</code> by hand. The OCS SASS has a table of numbers and a loop. SASS did the typing.</div>
</div>

**The data.** `$sizes` is a SASS map: a table of named rows, from `_sass/open-coding/elements/buttons/_tokens.scss`. SASS variables like `$sizes` start with `$`. They are used up when the site is built, so they never reach your browser.

~~~scss
$sizes: (
    /* height, vertical padding, horizontal padding, border radius, text size, line height */
    small: (28px, 4px, 10px, 40px, 15px, 20px, -0.23px),
    medium: (35px, 7px, 14px, 12px, 15px, 20px, -0.23px),
    large: (50px, 14px, 20px, 12px, 17px, 22px, -0.43px)
);
~~~

**The loop**, from `_utilities.scss`. `@each` works like Java's `for (String name : names)`: it runs once per row. `#{$name}` pastes the row's name into the selector, the part before `{` that says which tags the rule styles. `nth($values, 1)` reads the row's first value. SASS counts from 1, not 0 like Java.

~~~scss
@each $name, $values in $sizes {
    .ocs__btn.#{$name} {
        height: nth($values, 1);
        padding: nth($values, 2) nth($values, 3);
        border-radius: nth($values, 4);
        border-style: solid;
        border-width: 1px;
        font-size: nth($values, 5);
        line-height: nth($values, 6);
        letter-spacing: nth($values, 7);
    }
}
~~~

**The output.** Each time GitHub rebuilds the site, SASS writes plain CSS into `style.css`. This is the rule it wrote for the first row, copied from the live site. `.ocs__btn.small` means "any tag with both the `ocs__btn` and `small` classes."

~~~css
.ocs__btn.small, .small.iridescent {
  height: 28px;
  padding: 4px 10px;
  border-radius: 40px;
  border-style: solid;
  border-width: 1px;
  font-size: 15px;
  line-height: 20px;
  letter-spacing: -0.23px; }
~~~

<details markdown="1">
<summary>Fine print, for the curious</summary>

- The `$sizes` comment names six columns. The seventh value is the letter spacing.
- `_utilities.scss` loops over `$sizes` twice: once for plain `.small`, `.medium`, and `.large` classes any control can use, and once for the `ocs__btn` versions shown above.
- The extra `.small.iridescent` appears because another style reuses every `ocs__btn` rule with SASS `@extend`.
- SASS list positions start at 1 (Sass, n.d.-b), and `@each` can walk a map's keys and values together (Sass, n.d.-a).
- The word `small` names a whole row of values. College Board calls this data abstraction: "giving data a name without referencing the specific details of the representation" (College Board, 2025, 3.1.A.2).

</details>

<div class="ocs__split">
<div class="ocs__visual"><strong>Tailwind does this too.</strong> Tailwind builds its classes the same way: a table of values plus a loop. Its version 3 palette has 22 colors in 11 shades each, which makes 242 shaded colors, and each one gets classes like <code>bg-blue-500</code> (Tailwind Labs, n.d.-b). One difference: a Tailwind class like <code>px-4</code> does one job, left and right padding. An OCS word like <code>small</code> sets a whole row.</div>
<div class="ocs__visual"><strong>So does Lombok, a Java library.</strong> The <code>@Data</code> on <code>UserPreferences</code> makes Lombok write the getters and setters from the fields when the code compiles (Project Lombok, n.d.). You write the data. A tool writes the repetitive code.</div>

### Lab C: Model a Button as a POJO

<span class="ocs__status">Java POJO</span> `UserPreferences` stores a theme. Your `ButtonStyle` POJO stores one button's look. Finish two parts:

1. **The constructor** allows only `small`, `medium`, or `large`. Any other size becomes `medium`. Compare Strings with `.equals`, not `==`.

function loadScssMode() {

// The body of the rule whose selector list starts with this exact selector, or null.
function ruleFor(css, selector) {
  const start = css.indexOf(selector + ' {');
  if (start === -1) { return null; }
  return css.slice(start, css.indexOf('}', start) + 1);
}

function px(rule, property) {
  const match = rule && rule.match(new RegExp(property + ':\\s*([0-9.]+)px'));
  return match ? parseFloat(match[1]) : NaN;
}

function names(css, pattern) {
  const found = [];
  let match;
  while ((match = pattern.exec(css)) !== null) {
    if (!found.includes(match[1])) { found.push(match[1]); }
  }
  return found;
}

function button(classes, label) {
  return '<button type="button" class="' + classes + '">' + label + '</button>';
}

function addedTones(css) {
  return names(css, /\.ocs__btn\.alert-([A-Za-z0-9_-]+) \{/g).filter((n) => !BUILT_IN_TONES.includes(n));
}

// The CSS variable a tone uses, like --teal, and whether OCS defines it on this page.
function toneVariable(rule) {
  const match = rule && rule.match(/var\((--[A-Za-z0-9_-]+)\)/);
  if (!match) { return { name: null, defined: false }; }
  const value = getComputedStyle(document.documentElement).getPropertyValue(match[1]).trim();
  return { name: match[1], defined: value !== '' };
}

async function setUpSassRunner(container) {
  const lab = LABS[container.dataset.sassLab];
  const output = container.querySelector('[data-sass-output]');
  const preview = container.querySelector('[data-sass-preview]');
  const checks = container.querySelector('[data-sass-checks]');
  const status = container.querySelector('[data-sass-status]');
  const starter = container.querySelector('.editor-textarea').value.trim();
  // The compiled CSS is applied only inside this runner's preview box.
  const scopedStyle = document.createElement('style');
  container.appendChild(scopedStyle);

  await loadScssMode();
  const runner = new BaseRunner(container, { storageKey: container.dataset.storageKey });
  runner.initializeEditor({
    defaultCode: starter,
    editorHeight: '320px',
    fallbackCode: starter,
    codeMirrorOptions: { lineNumbers: true, mode: 'text/x-scss', theme: 'darcula', indentUnit: 4, tabSize: 4, styleSelectedText: true }
  });

  async function compile() {
    const src = runner.getValue();
    output.textContent = 'Compiling...';
    let sass;
    try {
      sass = await loadSass();
    } catch (error) {
      output.textContent = 'Could not load the SASS compiler. Check your internet connection and press Run again.';
      return;
    }
    // OCS builds with Ruby Sass, which needs global functions like nth(), so skip Dart Sass's deprecation notes.
    const quiet = { logger: sass.Logger.silent };
    let css;
    try {
      css = sass.compileString(src, quiet).css.replace(/^@charset "UTF-8";\n/, '');
    } catch (error) {
      output.textContent = 'SASS error. Read the line number, fix it, and run again. Check for a missing comma or quote.\n\n' + error.message;
      status.textContent = 'does not compile';
      scopedStyle.textContent = '';
      preview.innerHTML = '';
      showChecks(checks, [['Compiles with no errors', false]]);
      return;
    }
    output.textContent = css || '(Your SASS compiled, but it wrote no CSS.)';
    // Compile again inside #preview-id { ... } so the student's rules style only this preview.
    // Code that is fine at the top level (like @use) can fail once nested, so guard it.
    try {
      scopedStyle.textContent = sass.compileString('#' + preview.id + ' {\n' + src + '\n}', quiet).css.replace(/^@charset "UTF-8";\n/, '');
      preview.innerHTML = lab.preview(css);
      bindPreviewThemes(preview);
    } catch (error) {
      scopedStyle.textContent = '';
      preview.textContent = 'The preview could not be built from this SASS: ' + error.message;
    }
    const results = [['Compiles with no errors', true]].concat(lab.checks(css, src));
    showChecks(checks, results);
    const passed = results.every((result) => result[1]);
    status.textContent = passed ? 'all checks PASS' : 'keep going';
    status.className = 'ocs__status-pill ' + (passed ? 'ocs__status-pill--good' : 'ocs__status-pill--neutral');
  }

  runner.bindEditorButtons({
    resetValue: starter,
    onClear: () => {
      output.textContent = 'Code reset to the starter. Press Run to compile.';
      scopedStyle.textContent = '';
      preview.innerHTML = '';
      preview.style.removeProperty('--pref-accent-color');
      checks.innerHTML = '';
      status.textContent = 'not compiled yet';
      status.className = 'ocs__status-pill ocs__status-pill--neutral';
    }
  });
  runner.bindButton('', compile, 'run');
  runner.bindShortcut(compile);
  const copyOutput = container.querySelector('[data-hook="copy-output"]');
  copyOutput.addEventListener('click', () => {
    navigator.clipboard.writeText(output.textContent).then(() => runner.flashButton(copyOutput, '✔'));
  });
}

document.querySelectorAll('[data-sass-lab]').forEach((container) => { setUpSassRunner(container); });

**"Isn't SASS just CSS?"** SASS is a language that turns into CSS when the site is built. The real choice is a shared, named, theme-aware class versus a pasted, one-off style.

**"Why didn't the alert buttons change with the theme?"** They use site colors like `--red` on purpose, so red means danger in every theme.

**"Why does `ocs__btn medium` have a white border?"** This shows without a saved theme. The plain `.medium` class, from the first `$sizes` loop, sets `border: none`. It comes after `.ocs__btn` in `style.css` with the same specificity, so the border falls back to the text color. Adding `accent` brings the theme border back.

**"Where is `accent` in the grammar?"** It is used all over OCS, but `button-grammar.md` does not list it yet.

**"What if the system does not have what I need?"** Follow the three tweak rules, then open an issue or a pull request for Mr. Mortensen to review.

**"Why not just use Tailwind?"** Both generate classes from data. `ocs__` classes read OCS CSS variables, so they follow OCS themes with far fewer patches, while Tailwind classes read Tailwind's own theme. In September 2026, the Capstone page was rebuilt with `ocs__` grammar.

**"Can I use a little CSS in the challenge?"** No. That is the rule from the survey: CSS means no candy. If the class menu does not have what you need, OCS needs a new class, and Idea 3 shows how to add one.

**"Why does my card follow my theme?"** The HTML runner draws your HTML right into this page, so it reads the same CSS variables as everything else.

**"Does SASS run in my browser?"** Not normally. It runs once when the site is built. The lab loads a compiler into this page only so you can practice.
