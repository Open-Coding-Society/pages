## Quick Reference

* `Math` is in `java.lang`, so no import is needed.
* The `Math` class contains **static methods**, so call them with `Math.methodName()`.
* `Math.abs(x)` → absolute value.
* `Math.pow(base, exponent)` → power.
* `Math.sqrt(x)` → nonnegative square root.
* `Math.random()` → a `double` from **0.0 inclusive to 1.0 exclusive**.
* Random values can be changed with arithmetic and casting to make a defined range.

## LxD Cycle: Math Class

**Empathize — student learning issues:** Students commonly expect `Math` to require an object or import, overlook that `pow` and `sqrt` return `double`, and treat `Math.random()` as if it could return `1.0`. Random-integer expressions create additional trouble when students cast before multiplying, omit the `+ 1` needed for an inclusive range, or cannot explain which endpoints are possible.

**Define:**
- **POV:** Students need to reason from each `Math` method's return type and range because memorizing a random-number formula does not reveal off-by-one or misplaced-cast errors.
- **Learning issue:** Students need to trace scaling, casting, and shifting as separate transformations of the interval `0.0 <= value < 1.0`.

**Ideate:**
- **HMW Question:** How might we turn a random-range expression into visible interval transformations rather than a formula to copy?
- **HMW Question:** How might repeated trials help students detect an impossible endpoint without suggesting that a short test proves correctness?
- **Activity:** Annotate the range after each operation in `(int)(Math.random() * 6) + 1`, repair two off-by-one variants, and compare observed rolls with the mathematically possible set.

**Prototype & Test:** Ask a partner to build a range from 10 through 20 and justify both endpoints before running it. Record where the interval reasoning breaks, then revise the diagram or prompt and test again.

### What is the Math Class?

The `Math` class is part of `java.lang`, which is available automatically. Its methods are static, so we call them with `Math.methodName()`.

### College Board Essential Knowledge

| Method | What it does | Example |
| :--- | :--- | :--- |
| `Math.abs(int)` / `Math.abs(double)` | Absolute value | `Math.abs(-5)` → `5` |
| `Math.pow(double, double)` | Raises a number to a power | `Math.pow(2, 3)` → `8.0` |
| `Math.sqrt(double)` | Nonnegative square root | `Math.sqrt(25)` → `5.0` |
| `Math.random()` | Random `double` from 0.0 inclusive to 1.0 exclusive | `Math.random()` → varies |



### Examples

```java
System.out.println(Math.abs(-12));       // 12
System.out.println(Math.pow(2, 3));      // 8.0
System.out.println(Math.sqrt(25));       // 5.0
System.out.println(Math.random());       // 0.0 <= value < 1.0
```

You can combine Math methods in an expression:

```java
double distance = Math.sqrt(Math.pow(3, 2) + Math.pow(4, 2));
System.out.println(distance); // 5.0
```


````java
public class MathDemo {
    public static void main(String[] args) {
        System.out.println(Math.abs(-12));
        System.out.println(Math.pow(2, 3));
        System.out.println(Math.sqrt(25));
        System.out.println(Math.random());
    }
}
````

## The Important Part: `Math.random()`

`Math.random()` gives a `double` such that:

```text
0.0 <= Math.random() < 1.0
```

To get a random **integer from `min` to `max`, inclusive**:

```java
(int)(Math.random() * (max - min + 1)) + min
```

Example: a number from 1 through 6:

```java
int roll = (int)(Math.random() * 6) + 1;
```

Why?

1. `Math.random()` gives `0.0` up to but not including `1.0`.
2. Multiplying by `6` gives `0.0` up to but not including `6.0`.
3. Casting to `int` gives `0` through `5`.
4. Adding `1` gives `1` through `6`.
