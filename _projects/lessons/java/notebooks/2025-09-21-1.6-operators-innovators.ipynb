## 1. Reference Guide

| Term | Definition / Example | College Board Requirement |
|---|---|---|
| **Compound assignment** | Does the math with the variable's current value, then stores the result back in it: `score += 10;` | Unit 1, Topic 1.6, p. 38 |
| **`+=` `-=` `*=` `/=` `%=`** | Shorthand for `x = x op y`: `health -= 15;` is `health = health - 15;` | Unit 1, Topic 1.6, p. 38 |
| **Post-increment** | `x++;` adds 1 to `x` | Unit 1, Topic 1.6, p. 38 |
| **Post-decrement** | `x--;` subtracts 1 from `x` | Unit 1, Topic 1.6, p. 38 |
| **Tracing** | Writing the variable's value after every line to predict the final value | — |

- `/=` on an `int` is integer division: `x = 7; x /= 2;` leaves `3`
- `=+` is not an operator: `x =+ 5;` means `x = +5;` and throws away the old value

### Compound Assignment Operators

| Operator | Full Form | Use Case | Example |
| --- | --- | --- | --- |
| `+=` | `x = x + y` | Add to a total, or add text to a String | `score += 10` |
| `-=` | `x = x - y` | Apply a penalty or a cost | `health -= 15` |
| `*=` | `x = x * y` | Scale or apply a multiplier | `price *= 1.15` |
| `/=` | `x = x / y` | Split or scale down (int division for `int`) | `total /= count` |
| `%=` | `x = x % y` | Wrap around a board or clock | `position %= 10` |

### Increment and Decrement

| Operator | Same As | On the AP Exam? |
| --- | --- | --- |
| `x++;` | `x += 1;` | Yes, as its own statement |
| `x--;` | `x -= 1;` | Yes, as its own statement |
| `++x`, `--x` | prefix forms | No |
| `y = x++;` or `x++ + x++` | increments inside another expression | No |

### Tracing a Sequence

| Line | `score` after the line |
| --- | --- |
| `int score = 100;` | 100 |
| `score += 50;` | 150 |
| `score -= 25;` | 125 |
| `score *= 2;` | 250 |
| `score /= 4;` | 62 (int division) |
| `score %= 10;` | 2 |

### Predicting the Result

- **Each line uses the value left by the line before it** → trace one line at a time, in order
- **The variable is an `int`** → `/=` drops the decimal, and `%=` keeps only the remainder
- **The variable is a `double`** → `*=` and `/=` keep the decimal
- **`++` or `--` appears inside a bigger expression** → rewrite it as its own statement; that form is not tested

### College Board Requirements

AP CSA Unit 1, Topic 1.6 Compound Assignment Operators. Quoted from the course and exam description (College Board, 2025, p. 38):

- "Compound assignment operators +=, -=, *=, /=, and %= can be used in place of the assignment operator in numeric expressions. A compound assignment operator performs the indicated arithmetic operation between the value on the left and the value on the right and then assigns the result to the variable on the left."
- "The post-increment operator ++ and post-decrement operator -- are used to add 1 or subtract 1 from the stored value of a numeric variable. The new value is assigned to the variable."

The course leaves out the prefix forms (`++x`, `--x`) and the use of `++` and `--` inside other expressions, so this lesson uses them only as statements.

The Java language specification gives the exact meaning: "A compound assignment expression of the form E1 op= E2 is equivalent to E1 = (T) ((E1) op (E2)), where T is the type of E1, except that E1 is evaluated only once" (Gosling et al., 2023, section 15.26.2).

## 2. Real-Life Application
### The Arcade Level Counter That Ran Out of Room

A famous 1980 arcade maze game keeps its level number in one byte, which can only count up to 255. Every time you clear a level, the game adds one to that counter, like `level++`. On level 256, the code that draws the fruit icons along the bottom of the screen goes wrong, and half of the maze fills with garbled symbols. Nobody can finish that level, and players call it the "kill screen." Everyone guess: why did the game's designers not run into it during testing?

| Real Case | What Happened | Java Idea in This Lesson |
|:---:|:---|:---|
| Arcade level counter (1980) | Adding 1 to a full one-byte counter wrapped it around, and the fruit-drawing code broke | `level++` repeated many times; every counter has a limit |
| Bank balance | A deposit adds to the balance and a fee takes from it | `balance += deposit;` and `balance -= fee;` |
| Sale prices | A 10% discount scales the price | `price *= 0.9;` |
| Board games | Moving past the last space wraps to the start | `position += roll; position %= 10;` |

**The takeaway:** a compound operator changes the variable it is applied to. To predict a program, trace the value after every update, in order.

## 3. Tech Talk

**Learning Objective:** Use compound assignment operators and `++`/`--` to update a variable, and trace the variable's value after each line.

**Success Criteria:** You can rewrite long-form updates as compound operators, predict a variable's value after a sequence of updates, and spot when `/=` or `%=` on an `int` drops the decimal.

A compound operator writes the variable name once instead of twice:

<pre><code class="language-java">
score = score + 10;   // long form
score += 10;          // compound form: the same thing
count++;              // count += 1
</code></pre>

The type of the variable still matters:

<pre><code class="language-java">
int x = 7;
x /= 2;               // 3, because x is an int
double y = 7;
y /= 2;               // 3.5
</code></pre>

On the AP exam, `++` and `--` appear only as their own statements, like `count++;`.

### Code Examples

#### A. One operator at a time. The CompoundTrace class prints the value after every update

Trace it before you run it. Write the value of `num` after each line, then check:

<pre><code class="language-java">
num += 5;    // 15
num /= 4;    // ?
</code></pre>

Terminology:

**Compound assignment** reads the current value, does one operation with the value on the right, and stores the result back in the same variable.

---

```java
// CODE_RUNNER: Trace num on paper first, then run it and compare
public class CompoundTrace {
    public static void main(String[] args) {
        int num = 10;

        num += 5;
        System.out.println("After += 5: " + num);
        num -= 3;
        System.out.println("After -= 3: " + num);
        num *= 2;
        System.out.println("After *= 2: " + num);
        num /= 4;
        System.out.println("After /= 4: " + num);   // int division
        num %= 5;
        System.out.println("After %= 5: " + num);
        num++;
        System.out.println("After ++:   " + num);
        num--;
        System.out.println("After --:   " + num);
    }
}

CompoundTrace.main(null);
```

#### B. Game state. The GameState class models a game turn with several variables

Real programs update several variables per event. Each update is its own statement, so it is easy to trace:

<pre><code class="language-java">
health -= 15;        // the enemy hits
enemiesDefeated++;   // one more defeated
damage *= 1.5;       // power-up: damage is a double
</code></pre>

Terminology:

**State** is the set of values a program is tracking right now. Each compound operator changes exactly one piece of the state.

---

```java
// CODE_RUNNER: Predict the final state, run it, then add one more event
public class GameState {
    public static void main(String[] args) {
        int health = 100;
        int xp = 0;
        int level = 1;
        int enemiesDefeated = 0;
        double damage = 10.0;

        // Event 1: the enemy attacks
        health -= 15;

        // Event 2: you win the fight
        xp += 120;
        enemiesDefeated++;

        // Event 3: level up after 100 XP
        level++;
        xp -= 100;
        damage *= 1.5;

        // Event 4: a potion heals 30, capped at 100 by the next line
        health += 30;
        health = Math.min(health, 100);

        System.out.println("Health: " + health);
        System.out.println("XP: " + xp);
        System.out.println("Level: " + level);
        System.out.println("Enemies: " + enemiesDefeated);
        System.out.println("Damage: " + damage);
    }
}

GameState.main(null);
```

## 6. LxD Cycle Process

**Empathize:** Students write `score = score + 10` when `score += 10` says the same thing. More importantly, they lose track of a variable after several updates, forget that `/=` on an `int` drops the decimal, and mix up `+=` with `=+`.

**Define:**
- **POV:** CSA students need to trace a variable through a sequence of compound updates, because the AP exam tests the final value after several lines, not the syntax alone.
- **Learning Goal:** Students will apply every compound assignment operator, use `++` and `--` as statements, and predict a variable's value after a sequence of updates.

**Ideate:**
- **HMW Question:** How might we get students to trace one line at a time instead of guessing the final value?
- **HMW Question:** How might we keep the practice about operators instead of about a complicated scenario?
- **Activity:** Rewrite long-form game code with compound operators, trace a six-line sequence, then fill in the updates for a step tracker and a board game.

**Prototype:**
- A reference guide with a trace table, runnable trace and game-state examples, a rewrite Popcorn Hack with a checker, an MCQ knowledge check, three scaffolded homework hacks with self-checking output, and a grading rubric.
- A trial run with open-ended scenarios showed students adding features instead of practicing operators, so each hack now gives the events in comments and asks for one operator per update.
- Excellence means tracing each update correctly, including integer division with `/=` and wrap-around with `%=`.

**Test:**
- Ask peers in peer review to complete the practice without additional explanation.
- Observe whether peers trace each line before running, and whether they predict the effect of `/=` on an `int`.
- Compare your lesson with another that is posted.
- Use the findings to revise any instruction or rubric criterion that did not guide students clearly.
- On submission, collect evidence from runner output, MCQ results, AI grading and student explanations.
- After teaching, grading and analysis, come back and revise lesson to complete teaching cycle for continuous improvement.
