## Quick Reference

* **Class method** = a method associated with the class, not a specific object.
* A class method uses the **`static`** keyword in its header.
* Call a class method with `ClassName.methodName()`.
* Inside the class that defines the method, the class name is optional: `methodName()`.
* A class method cannot directly use an instance variable because there is no specific object attached to the call.

## LxD Cycle: Calling Class Methods

**Empathize — student learning issues:** Students may read `static` as meaning "constant" or "global," assume every method call needs an object, or call an instance method with a class name. Because Java permits a static method to be called through an object reference, code can compile while reinforcing the mistaken idea that the method belongs to that object. Students may also expect a class method to access instance fields without selecting an instance.

**Define:**
- **POV:** Students learning class methods need to connect `static` with the absence of a current object because call syntax alone can hide whether a particular instance is involved.
- **Learning issue:** Students need to distinguish "valid Java" from the clearest call form and explain why `ClassName.method()` communicates class ownership.

**Ideate:**
- **HMW Question:** How might we make the presence or absence of a current object visible during a method call?
- **HMW Question:** How might we help students diagnose a static-context error from the data the method tries to access?
- **Activity:** Sort call cards into class-method and instance-method columns, justify the receiver for each call, and repair one call that produces a static-context compiler error.

**Prototype & Test:** Give a partner one class with a static method, an instance method, and an instance field. Have the partner predict which calls compile and explain each receiver; revise the example if the explanation relies only on memorized punctuation.

### Class Methods

A **class method** is associated with the class rather than a particular object. In Java, class methods use the `static` keyword.

```java
public static void printMessage(String message) {
    System.out.println(message);
}
```

Because `printMessage` is static, it can be called using the class name:

```java
Printer.printMessage("Hello!");
```

This does **not** require a `Printer` object.

### Static vs. Instance

| Type | Belongs to | Typical call |
| :--- | :--- | :--- |
| Class / static method | Class | `ClassName.method()` |
| Instance method | Object | `objectName.method()` |



### Example — Which calls work?

```java
public class Printer {
    public static void printMessage(String message) {
        System.out.println(message);
    }

    public void printStatus() {
        System.out.println("Printer is ready.");
    }
}
```

```java
Printer.printMessage("Hello!"); // works
Printer.printStatus();          // does NOT work

Printer p = new Printer();
p.printStatus();                // works
```

`printMessage()` is static, so it belongs to `Printer`. `printStatus()` is an instance method, so Java needs a particular `Printer` object.

If you see an error like **"non-static method ... cannot be referenced from a static context,"** check whether you are trying to call an instance method as if it were static.


````java
public class Printer {
    public static void printMessage(String message) {
        System.out.println(message);
    }

    public void printStatus() {
        System.out.println("Printer is ready.");
    }

    public static void main(String[] args) {
        Printer.printMessage("Hello!");

        Printer p = new Printer();
        p.printStatus();
    }
}
````

### Calling a static method inside its class

When the call happens inside the class that defines the static method, the class name is optional.

Both calls below work:

```java
Calculator.square(5);
square(5);
```

For example:

```java
public class Calculator {
    public static int square(int number) {
        return number * number;
    }

    public static void main(String[] args) {
        System.out.println(square(5));
        System.out.println(Calculator.square(5));
    }
}
```

Both calls produce `25`.


### Static methods and instance variables

A static method cannot directly access an instance variable.

```java
public class Student {
    String name;

    public Student(String name) {
        this.name = name;
    }

    public static void printName() {
        System.out.println(name); // ERROR
    }
}
```

The problem is that `name` belongs to an individual `Student`. A static method is not connected to one specific student.

An instance method can access it:

```java
public void printName() {
    System.out.println(name);
}
```

**Popcorn Hack (2 min):** Decide whether each call works. If it does not, explain why.

```java
public class Robot {
    String name;

    public Robot(String name) {
        this.name = name;
    }

    public void introduce() {
        System.out.println("I am " + name);
    }

    public static void beep() {
        System.out.println("Beep!");
    }
}

Robot r = new Robot("R2");
```

1. `Robot.beep();`
2. `r.beep();`
3. `Robot.introduce();`
4. `r.introduce();`

**Answers:**  
1. Works.  
2. Valid Java, but `Robot.beep()` is clearer because `beep()` is static.  
3. Does not work because `introduce()` is an instance method.  
4. Works because `r` identifies the object.

## Quick Summary - What You Learned

1. **`static`** identifies a class method.
2. Class methods belong to the **class**, not an individual object.
3. Use `ClassName.methodName()` to call a class method.
4. Inside the defining class, the class name can be omitted.
5. Instance methods need an object when called.
6. A static method cannot directly access an instance variable.
