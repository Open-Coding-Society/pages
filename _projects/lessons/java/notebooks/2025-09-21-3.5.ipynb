## Key Vocabulary

| Term | Definition | Example |
|---|---|---|
| **Method** | A named behavior defined inside a class. | `calculateArea()` |
| **Return type** | The type of value a method sends back; `void` means no value. | `int` |
| **Parameter** | A typed input named in the method header. | `int length` |
| **Argument** | The actual value supplied in a method call. | `calculateArea(8, 5)` |
| **Return statement** | Ends the method and sends a value to its caller. | `return length * width;` |
| **Method signature** | The method name and its parameter types. | `calculateArea(int, int)` |

## Concept Overview

### What is a Method?

A method is a reusable block of code that describes one behavior of a class. Instead of repeating the same calculation or print statement, you write it once and call it whenever the behavior is needed.

---

## Reading a Method Header

Read a method header from left to right:

```java
public int calculateArea(int length, int width)
```

| Part | Purpose | Example |
|---|---|---|
| Access modifier | Controls where the method can be called | `public` |
| Return type | States what comes back to the caller | `int` |
| Method name | Describes the behavior | `calculateArea` |
| Parameters | Declare the typed inputs | `int length, int width` |

The header is the first line. The code between the braces is the method **body**. If the return type is not `void`, every possible path through the method must return a compatible value.

````java
// Method Header Example
public class RectangleHeader {
    public int calculateArea(int length, int width) {
        return length * width;
    }

    public static void main(String[] args) {
        RectangleHeader rectangle = new RectangleHeader();
        int area = rectangle.calculateArea(8, 5);
        System.out.println("Area: " + area);
    }
}
RectangleHeader.main(null);
````

---

## `void` Methods and Returning Methods

A `void` method performs an action but does not send a value back. A returning method calculates or chooses a value that the caller can store and use.

- Use `void` when the goal is an action, such as printing a description.
- Use a type such as `int`, `double`, `boolean`, or `String` when the caller needs a result.
- Printing and returning are different: `System.out.println` displays a value, while `return` gives a value to the caller.

````java
// Void and Returning Method Examples
public class Shape {
    // This method performs an action and returns no value.
    public void printDimensions(int length, int width) {
        System.out.println("Dimensions: " + length + " by " + width);
    }

    // This method returns an int that the caller can store.
    public int calculateArea(int length, int width) {
        return length * width;
    }

    public static void main(String[] args) {
        Shape rectangle = new Shape();
        rectangle.printDimensions(8, 5);

        int area = rectangle.calculateArea(8, 5);
        System.out.println("Area: " + area);
    }
}
Shape.main(null);
````

## Parameters and Arguments

Parameters are the input variables declared in a method header. Arguments are the values placed in a call. Java matches arguments to parameters by position.

```java
rectangle.calculateArea(8, 5);
```

In this call, `8` is assigned to `length` and `5` is assigned to `width`. The number and types of the arguments must match the method's parameters.

````java
// Arguments are matched to parameters from left to right.
public class ArgumentOrder {
    public String describe(int length, int width) {
        return "length=" + length + ", width=" + width;
    }

    public static void main(String[] args) {
        ArgumentOrder example = new ArgumentOrder();
        System.out.println(example.describe(8, 5));
        System.out.println(example.describe(5, 8));
    }
}
ArgumentOrder.main(null);
````

## Returning a Value on Every Path

A value-returning method must return the declared type no matter which branch runs. The method below returns a `String` for all three possible comparisons.

````java
// Every path returns a String.
public class ShapeClassifier {
    public String classify(int length, int width) {
        if (length <= 0 || width <= 0) {
            return "invalid";
        }
        if (length == width) {
            return "square";
        }
        return "rectangle";
    }

    public static void main(String[] args) {
        ShapeClassifier classifier = new ShapeClassifier();
        System.out.println(classifier.classify(4, 4));
        System.out.println(classifier.classify(8, 5));
        System.out.println(classifier.classify(-1, 5));
    }
}
ShapeClassifier.main(null);
````

## Summary

````java
// A void method performs an action.
public void printMessage(String message) {
    System.out.println(message);
}

// A returning method sends a value to its caller.
public int add(int first, int second) {
    return first + second;
}
````

### Remember

- Give each method one clear behavior.
- Choose `void` only when the caller does not need a value.
- Parameters declare inputs; arguments supply the actual values.
- A non-`void` method must return the declared type on every path.
- Use descriptive method and parameter names.
- Run the homework tests before submitting.
