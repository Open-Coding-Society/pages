## 1. Reference Guide

### Key Topics

<br>

| Term | Definition | Example |
|---|---|---|
| **ArrayList** | Resizable list of object references; part of `java.util`. | `ArrayList<Integer> nums = new ArrayList<Integer>();` |
| **Generic type** | The `<E>` in `ArrayList<E>` that fixes the element type at compile time. | `ArrayList<String>` only holds `String` |
| **size()** | Method that returns the number of elements currently stored. | `nums.size()` |
| **Index** | Position of an element, from `0` to `size() - 1`. | `nums.get(0)` |
| **IndexOutOfBoundsException** | Runtime error from `get`/`set`/`remove` at an invalid index. | `nums.get(nums.size())` |

- An `ArrayList` holds **object references**, so there is no `ArrayList<int>`. Use `ArrayList<Integer>` and Java boxes each `int` for you.
- `ArrayList` lives in `java.util`, so a full program needs `import java.util.ArrayList;`.

### Array vs. ArrayList

| | Array | ArrayList |
|---|---|---|
| Size | Fixed at creation | Grows and shrinks |
| Element count | `.length` (field) | `.size()` (method) |
| Read/write | `arr[i]` | `list.get(i)` / `list.set(i, v)` |
| Add/remove | Not possible after creation | `add`, `add(index, obj)`, `remove(index)`, `remove(Object)` |

### Picking a Method

- **Add to the end** → `add(obj)`
- **Insert at a position** → `add(index, obj)`
- **Read a value** → `get(index)`
- **Replace a value** → `set(index, obj)`
- **Remove by position** → `remove(int index)`
- **Remove by value** → `remove(Object obj)`, such as `nums.remove(Integer.valueOf(7))`
- **Count the elements** → `size()`

Essential knowledge 4.8.A.5 places these method signatures on the Java Quick Reference, quoted exactly as printed (College Board, 2025, p. 111):

| Method | Quoted description |
|---|---|
| `int size()` | "returns the number of elements in the list." |
| `boolean add(E obj)` | "appends `obj` to end of list; returns `true`." |
| `void add(int index, E obj)` | "inserts `obj` at position `index` (0 <= index <= size), moving elements at position `index` and higher to the right (adds 1 to their indices) and adds 1 to size." |
| `E get(int index)` | "returns the element at position `index` in the list." |
| `E set(int index, E obj)` | "replaces the element at position `index` with `obj`; returns the element formerly at position `index`." |
| `E remove(int index)` | "removes element from position `index`, moving elements at position `index + 1` and higher to the left (subtracts 1 from their indices) and subtracts 1 from size; returns the element formerly at position `index`." |

## 4. Lesson Plan

**Learning Objective:** Declare an `ArrayList` and use `add`, `get`, `set`, `remove`, and `size` to build and change it correctly.

**Success Criteria:** You can add elements at the end or at a specific index, read and replace an element without changing the list's size, remove an element by index or by value using the correct `remove` overload, and explain why `ArrayList` uses `size()` instead of `length`.

### Tech Talk (5 minutes)

An array is created with a fixed length. A grocery list grows every time you think of something new and shrinks every time you cross something off. `ArrayList` is built for data that changes size while a program runs.

```java
ArrayList<String> groceries = new ArrayList<String>(); // empty list
groceries.add("Milk");                                 // add to the end
groceries.add(0, "Eggs");                              // insert at index 0
String first = groceries.get(0);                       // read: "Eggs"
groceries.set(1, "Oat Milk");                          // replace index 1
groceries.remove(0);                                   // remove by index
System.out.println(groceries.size());                  // count elements
```

## 5. Code Examples

### A. Adding elements.  The PlaylistBuilder class compares a fixed array to a growing ArrayList

Two `add` calls get mixed up a lot:

<pre><code class="language-java">
playlist.add("Study Beats");      // add(obj): appends to the end
playlist.add(1, "Focus Mix");     // add(index, obj): inserts at index 1, shifts the rest right
</code></pre>

Terminology:

An array's `length` is fixed when it is created. An `ArrayList` starts empty and its `size()` grows by one with every `add`.

---

````java
// CODE_RUNNER: Predict the playlist order before running, then add a fourth song at index 0
// Fixed Size vs. Growing Size
import java.util.ArrayList;

public class PlaylistBuilder {
    public static void main(String[] args) {
        // An array's length never changes
        int[] fixedScores = new int[3];
        System.out.println("fixedScores.length = " + fixedScores.length);

        // An ArrayList starts empty and grows
        ArrayList<String> playlist = new ArrayList<String>();
        System.out.println("playlist.size() = " + playlist.size());

        playlist.add("Morning Run");
        playlist.add("Study Beats");
        playlist.add(1, "Focus Mix");   // inserts, pushing "Study Beats" to index 2

        System.out.println("playlist = " + playlist);
        System.out.println("playlist.size() = " + playlist.size());
    }
}

PlaylistBuilder.main(null);
````

### B. Reading and changing elements.  The ScoreEditor class uses get and set, then steps past the last index

Common indexing mistake:

<pre><code class="language-java">
scores[1] = 95;          // array syntax, does not compile for an ArrayList
scores.set(1, 95);       // ArrayList uses set
int first = scores.get(0);   // ArrayList uses get
</code></pre>

Terminology:

`get` reads a value and `set` replaces one. Neither changes `size()`. A valid index runs from `0` to `size() - 1`, and only `add` can make a new slot.

---

````java
// CODE_RUNNER: Run it, then change get(3) to get(2) and see what prints instead of the exception
// get, set, and the boundary
import java.util.ArrayList;

public class ScoreEditor {
    public static void main(String[] args) {
        ArrayList<Integer> scores = new ArrayList<Integer>();
        scores.add(70);
        scores.add(85);
        scores.add(90);

        int first = scores.get(0);        // read, list unchanged
        int replaced = scores.set(1, 95); // replace, returns the old value

        System.out.println("first = " + first);
        System.out.println("replaced = " + replaced);
        System.out.println("scores = " + scores + ", size = " + scores.size());

        try {
            System.out.println(scores.get(3));   // valid indexes are 0, 1, 2
        } catch (IndexOutOfBoundsException e) {
            System.out.println("get(3) failed: " + e.getMessage());
        }
    }
}

ScoreEditor.main(null);
````

### C. size() and traversal.  The TaskList class loops over every element

Common length mistake:

<pre><code class="language-java">
tasks.length       // compiler error: ArrayList has no length field
tasks.size()       // ArrayList uses the size() method
</code></pre>

Terminology:

`size()` is a method because the number of elements can change while the program runs. Looping with `i < tasks.size()` stays correct even after elements are added or removed.

---

````java
// CODE_RUNNER: Run it, then change tasks.size() to tasks.length and read the compiler error
// Traversing with size()
import java.util.ArrayList;

public class TaskList {
    public static void main(String[] args) {
        ArrayList<String> tasks = new ArrayList<String>();
        tasks.add("Read");
        tasks.add("Write code");
        tasks.add("Test");

        for (int i = 0; i < tasks.size(); i++) {
            System.out.println(i + ": " + tasks.get(i));
        }
    }
}

TaskList.main(null);
````

### D. Two remove methods.  The RemoveComparison class removes by position and by value

Common remove mistake:

<pre><code class="language-java">
values.remove(1);                    // int argument: removes the element AT index 1
values.remove(Integer.valueOf(1));   // Integer argument: removes the VALUE 1
</code></pre>

Terminology:

For `ArrayList<Integer>`, an `int` argument calls `remove(int index)` and an `Integer` argument calls `remove(Object obj)`. Both shift every later element one position left and reduce `size()` by one.

---

````java
// CODE_RUNNER: Predict both lists before running, then try byValue.remove(Integer.valueOf(99)) and explain why nothing changes
// remove(int index) vs. remove(Object obj)
import java.util.ArrayList;

public class RemoveComparison {
    public static void main(String[] args) {
        ArrayList<Integer> byIndex = new ArrayList<Integer>();
        byIndex.add(1);
        byIndex.add(2);
        byIndex.add(3);
        byIndex.remove(1);                     // int: removes the element at index 1
        System.out.println("byIndex = " + byIndex);

        ArrayList<Integer> byValue = new ArrayList<Integer>();
        byValue.add(1);
        byValue.add(2);
        byValue.add(3);
        byValue.remove(Integer.valueOf(1));    // Integer: removes the value 1
        System.out.println("byValue = " + byValue);
    }
}

RemoveComparison.main(null);
````
