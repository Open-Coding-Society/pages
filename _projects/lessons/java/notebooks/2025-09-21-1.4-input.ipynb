## 1. Reference Guide

| Term | Definition / Example | College Board Requirement |
|---|---|---|
| **Assignment statement** | Stores the value of the right side in the variable on the left: `x = x + 5;` | Unit 1, Topic 1.4, 1.4.A.2, p. 35 |
| **Assignment operator** | `=` means "store," not "is equal to.": `score = 90;` | Unit 1, Topic 1.4, 1.4.A.2, p. 35 |
| **Equality operator** | `==` compares two values and gives `true` or `false`: `score == 90` | — |
| **Expression** | Code that works out to a single value with a type: `(7 + 8) / 2.0` | Unit 1, Topic 1.4, 1.4.A.3, p. 35 |
| **Input** | Data that comes from outside the program: Keyboard, a web request | Unit 1, Topic 1.4, 1.4.B.1, p. 35 |
| **Scanner** | A class that reads text input: `Scanner sc = new Scanner(System.in);` | Unit 1, Topic 1.4, 1.4.B.1, p. 35. Specific input code is outside the exam scope. |

- A variable needs a value before it can be used on the right side
- Changing a variable later does not change a copy made from it earlier

### Assignment Step by Step

| Code | What Java does | Values after this line |
| --- | --- | --- |
| `int a = 5;` | Stores 5 | `a` is `5` |
| `a = a * 2;` | Works out `5 * 2` first, then stores it | `a` is `10` |
| `int b = a;` | Copies the current value of `a` | `a` is `10`, `b` is `10` |
| `a = 1;` | Replaces `a`; `b` keeps its copy | `a` is `1`, `b` is `10` |

### Setting Up Scanner

| Step | Code |
| --- | --- |
| 1. Import the class | `import java.util.Scanner;` |
| 2. Make one Scanner | `Scanner sc = new Scanner(System.in);` |
| 3. Prompt, then read | `System.out.print("Age: ");` then `int age = sc.nextInt();` |
| 4. Close it when done | `sc.close();` |

### Scanner Methods

The Scanner class provides several methods for reading different types of input:

- **`nextInt()`** - Reads the next integer value (e.g., `42`)
- **`nextDouble()`** - Reads the next decimal number (e.g., `3.14`)
- **`next()`** - Reads the next word up to whitespace (e.g., `"Hello"`)
- **`nextLine()`** - Reads an entire line including spaces (e.g., `"Hello World"`)

### Picking a Scanner Method

- **Whole number** → `nextInt()`
- **Decimal** → `nextDouble()`
- **One word** → `next()`
- **A full line with spaces** → `nextLine()`

### The nextInt() and nextLine() Problem

`nextInt()`, `nextDouble()`, and `next()` read their value but leave the Enter key behind. The next `nextLine()` grabs that leftover and returns an empty String right away.

Oracle's docs explain why: `nextLine()` "returns the rest of the current line, excluding any line separator at the end" (Oracle, n.d.). After `nextInt()`, the rest of the line is empty.

**The fix:** add one extra `sc.nextLine();` after `nextInt()` or `nextDouble()`.

## 2. Real-Life Application
### One Missing Equals Sign That Almost Took Over Linux

In 2003, someone slipped two lines into the Linux kernel's source code, the code that runs servers, Android phones, and much of the internet. It looked like a harmless safety check. Everyone find the one character that turns this check into a takeover before you read the table:

<pre><code class="language-c">
if ((options == (__WCLONE|__WALL)) &amp;&amp; (current-&gt;uid = 0))
        retval = -EINVAL;
</code></pre>

| What It Looks Like | What It Really Does | Java Idea in This Lesson |
|:---|:---|:---|
| `current->uid == 0` would ask "is this user the administrator?" | `current->uid = 0` makes the user the administrator (user 0 is root) | `==` compares; `=` stores a value |
| A check that quietly fails | The attacker gets full control, and the program keeps running normally | An assignment works out the right side, then stores it on the left |
| Caught before release | Reviewers spotted the change before it shipped | Read every `=` as "store" and every `==` as "is equal to" |

Input is the other half of this lesson, and the College Board lists four forms of it. Every app on your phone uses at least one:

| Input Form | Real Example | Java Version in This Lesson |
|:---:|:---|:---|
| Tactile | Tapping a touchscreen | Not covered |
| Audio | Talking to a voice assistant | Not covered |
| Visual | Scanning a QR code with the camera | Not covered |
| Text | Typing your name into a form | `Scanner`: `nextLine()`, `nextInt()`, `nextDouble()` |

**The takeaway:** one character can change what a program does. Trace every `=` and every input value line by line.

## 3. Tech Talk

**Learning Objective:** Write assignment statements with expressions, say what value each variable holds, and read input with Scanner.

**Success Criteria:** You can trace the value in a variable after each line, set up a Scanner, read int, double, and String input, and fix the leftover newline after nextInt().

Assignment works right to left. Java finishes the whole right side first, then stores the result on the left.

<pre><code class="language-java">
int x = 10;
x = 20;          // 10 is gone
x = x + 5;       // right side is 20 + 5, so x is 25
String name = null;   // null means it points at no object
</code></pre>

The type is decided before the value is stored, which surprises people:

<pre><code class="language-java">
double average = (7 + 8) / 2;     // 7.0, because (7 + 8) / 2 is int math
double better = (7 + 8) / 2.0;    // 7.5
</code></pre>

`=` stores a value. `==` compares two values. Input is anything that comes from outside the program. `Scanner` is one way to read text from a keyboard.

<pre><code class="language-java">
Scanner sc = new Scanner(System.in);
int age = sc.nextInt();
</code></pre>

### Code Examples

#### A. Reading input. The InputDemoRunner class reads a String, an int, and a double with Scanner

The runner has no keyboard, so this copy reads the answers from a String. Change them and run it again.

Common reading mistake:

<pre><code class="language-java">
int age = sc.nextInt();        // reads 16 but leaves the Enter key behind
String name = sc.nextLine();   // grabs the leftover Enter, so name is ""
</code></pre>

Terminology:

`Scanner` is a **reference** type: `new Scanner(...)` makes one object, and each call like `nextLine()`, `nextInt()`, or `nextDouble()` reads the next value from it.

---

```java
// CODE_RUNNER: Run it with the simulated answers, then change them
import java.util.Scanner;

public class InputDemoRunner {
    public static void main(String[] args) {
        Scanner sc = new Scanner("Alex Rivera\n16\n2.5\n");  // simulated keyboard input
        
        // Reading a String (entire line)
        System.out.print("Enter your name: ");
        String name = sc.nextLine();
        System.out.println("Hello, " + name + "!");
        
        // Reading an integer
        System.out.print("Enter your age: ");
        int age = sc.nextInt();
        System.out.println("You entered: " + age);
        
        // Reading a double
        System.out.print("Enter a decimal: ");
        double value = sc.nextDouble();
        System.out.println("Double that is: " + (value * 2));
        
        sc.close();  // Always close to prevent resource leaks
    }
}

InputDemoRunner.main(null);
```

#### B. Full input program. The InputLessonRunner class reads three answers and prints a summary

These are the Popcorn Hack inputs. Trace the output first, then run it.

Common assignment mistake:

<pre><code class="language-java">
age + 1;          // works out 21 but stores it nowhere, so Java rejects it: not a statement
age = age + 1;    // right side first, then store: age is now 21
</code></pre>

Terminology:

**Assignment:** Java works out the whole right side first, then stores the result in the variable on the left. `(age + 1)` in the summary only prints next year's age; `age` itself stays 20.

---

```java
// CODE_RUNNER: Trace the output first, then run it to check
import java.util.Scanner;

public class InputLessonRunner {
    public static void main(String[] args) {
        System.out.println("=== User Information Program ===\n");
        Scanner sc = new Scanner("Alice Wonderland\n20\n3.85\n");  // simulated keyboard input

        // String input
        System.out.print("Enter your full name: ");
        String name = sc.nextLine();
        
        // Integer input
        System.out.print("Enter your age: ");
        int age = sc.nextInt();
        
        // Double input
        System.out.print("Enter your GPA: ");
        double gpa = sc.nextDouble();
        
        // Display results
        System.out.println("\n--- Summary ---");
        System.out.println("Name: " + name);
        System.out.println("Age: " + age + " (next year: " + (age + 1) + ")");
        System.out.println("GPA: " + gpa);
        
        sc.close();
    }
}

InputLessonRunner.main(null);
```

## 6. LxD Cycle Process

**Empathize:** Students read x = x + 5 as a math equation instead of "work out the right side, then store it." At the end of a first-year Java course, about a third of students still held a broken model of assignment (Ma et al., 2007). Mixing up = and == is one of the most common novice mistakes (Altadmri & Brown, 2015).

**Define:**
- **POV:** CSA students need a clear picture of assignment and of where input comes from, because every interactive program depends on both.
- **Learning Goal:** Students will write assignment statements and say what each variable holds, and read int, double, and String input with Scanner.

**Ideate:**
- **HMW Question:** How might we make "right side first, then store" automatic through line-by-line tracing?
- **HMW Question:** How might we show that a value can come from outside the program: a keyboard now, a web request in the project?
- **Activity:** Trace a full input program line by line, predict answers before revealing them, then use = and ==, read input, and build an average calculator.

**Prototype:**
- A reference guide on Scanner methods and the nextInt() and nextLine() problem, runnable input examples with simulated answers, a code trace, an MCQ knowledge check, three scaffolded homework hacks with self-checking output, and a grading rubric.
- Students write their trace before running the cell and revise it where it differs.
- Excellence means tracing what each variable holds after every line, avoiding integer division, explaining `=` versus `==`, and improving the first attempt rather than only producing working code.

**Test:**
- Ask peers in peer review to complete the practice without additional explanation.
- Observe whether peers work out the right side first before storing, and whether they trace before they run.
- Compare your lesson with another that is posted.
- Use the findings to revise any instruction or rubric criterion that did not guide students clearly.
- On submission, collect evidence from Popcorn traces, runner output, MCQ results, AI grading and student explanations.  
- After teaching, grading and analysis, come back and revise lesson to complete teaching cycle for continuous improvement.
