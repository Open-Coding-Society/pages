## Constructors

A constructor runs the moment you write `new`. It fills in the object's fields so it comes out ready to use — not blank.

```java
public class Student {
    String name;
    int grade;
    public Student(String name, int grade) {
        this.name  = name;
        this.grade = grade;
    }
}
Student s = new Student("Alex", 11);  // constructor fires here
```

## Constructor Anatomy

| Part | What it is |
|---|---|
| `public Student(...)` | The constructor — must have the same name as the class |
| `String name, int grade` | Parameters — values you pass in when using `new` |
| `this.name` | The field on the object being built |
| `name` (no `this.`) | The parameter that was passed in |
| `new Student(...)` | The keyword that triggers the constructor |

## 2. Lesson Plan

**Learning Objective:** Understand what a constructor is, why it exists, and how to write one.

**Success Criteria:** You can write a constructor for a given class that takes parameters and sets fields correctly.

### Tech Talk (3 minutes)

**What is a constructor?**

When you order a custom hoodie online, you fill out a form — your size, color, and name to put on the back. The moment you hit "place order," the factory takes that information and builds your hoodie with those exact values. You get back a finished product, ready to wear.

A constructor is the same idea. When you write `new Student("Alex", 11)`, you're placing an order. The constructor is the factory — it takes your inputs and builds a finished object with those values already set. You get back a ready-to-use object, not an empty one.

A class is just a blueprint — it describes what an object *will* have. A constructor is what actually builds it. It runs once, automatically, the moment you write `new`.

Without a constructor, every field is empty (`null` for text, `0` for numbers). The constructor is what fills them in.

✅ **Do this** — pass the values in through `new` and let the constructor set everything up for you in one shot:
```java
Student s = new Student("Alex", 11);
// constructor runs immediately, s.name = "Alex" and s.grade = 11 are set
```
This is correct because the object comes out fully built. You can use it right away.

❌ **Don't do this** — manually setting fields after creation is messy and easy to forget:
```java
Student s = new Student();
s.name  = "Alex";
s.grade = 11;
```

### Code Examples

**A. Simple — the problem without a constructor**

````python
public class Student {
    String name;
    int grade;
    // no constructor — nothing tells Java what values to use
}

// "Student s" is just a variable, like "int x" — except instead of holding
// a number, it holds a Student object. "s" is the name we gave it.
Student s = new Student();
System.out.println(s.name);   // we never set this, so Java defaults to null
System.out.println(s.grade);  // we never set this, so Java defaults to 0
````

**Output:**
```
null
0
```
`null` and `0` are Java's defaults — they mean "nothing was set." We created a Student but never gave it a name or grade, so Java just filled those spots with empty values.

**B. Intermediate — adding a constructor**

````python
public class Student {
    String name;
    int grade;

    // =============================================
    //  CONSTRUCTOR
    //  Purpose: guarantee every Student starts with
    //  a real name and grade, not null and 0
    // =============================================
    public Student(String name, int grade) {
        this.name  = name;
        this.grade = grade;
    }

}

Student s = new Student("Alex", 11);
System.out.println(s.name);
System.out.println(s.grade);
````

**Output:**
```
Alex
11
```
This time we got real values back because the constructor ran when we wrote `new Student("Alex", 11)`. It took `"Alex"` and `11`, stored them in the object's fields, and now `s.name` and `s.grade` actually hold something.

**C. Complex — two objects from the same constructor**

````python
// The same constructor builds every Student —
// each object gets its own copy of the fields.

Student s1 = new Student("Alex", 11);
Student s2 = new Student("Jordan", 10);

System.out.println(s1.name + " " + s1.grade);
System.out.println(s2.name + " " + s2.grade);
````

**Output:**
```
Alex 11
Jordan 10
```
