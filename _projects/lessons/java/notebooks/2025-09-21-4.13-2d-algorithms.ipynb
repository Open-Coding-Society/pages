## 1. Reference Guide

### Key Topics

<br>

| Term | Definition | Example |
|---|---|---|
| **Row-major** | Visit every seat in row 0, then row 1, and so on. | outer loop `row`, inner loop `col` |
| **Column-major** | Visit every row in column 0, then column 1, and so on. | outer loop `col`, inner loop `row` |
| **`grid.length`** | The number of rows. | `row < grid.length` |
| **`grid[row].length`** | The number of columns in that row. | `col < grid[row].length` |
| **Accumulator** | A variable that builds a result across the traversal. | `total`, `count`, `best` |
| **Sentinel** | A value that means "no data" or "not found". | `-1`, `{-1, -1}` |

- A 2D array is an array of arrays. Every algorithm here is a 1D algorithm you already know, run once per row.
- Always use `<`, never `<=`. `grid.length` is one index past the last valid row.
- `return` leaves the whole method from any depth, so it is the cleanest way to stop a search.

### Picking an Algorithm

- **Where is it?** → search, return `{row, col}`, and `return` at the first match
- **Total, average, or count** → one traversal with accumulators; guard divide-by-zero and cast to `double`
- **Best so far** → track a "champion" that starts at a value guaranteed to lose, such as `-1`
- **Per-row answer** → declare the accumulator inside the row loop so it restarts for each row

## 5. Code Examples

### A. Traversal order.  The TraversalOrder class walks the same grid row-major and column-major

Common starting mistake:

<pre><code class="language-java">
for (int col = 0; col < grid.length; col++)         // wrong: grid.length counts rows
for (int col = 0; col < grid[row].length; col++)    // right: columns in this row
</code></pre>

Terminology:

**Row-major** is the default: the row loop is on the outside. Swap which loop is outside and you get **column-major**. That swap is the entire idea of traversal order.

---

````java
// CODE_RUNNER: Predict both lines before running, then add a third row of 7, 8, 9
// Row-major vs. column-major
public class TraversalOrder {
    public static void main(String[] args) {
        int[][] grid = {
            {1, 2, 3},
            {4, 5, 6}
        };

        String rowMajor = "";
        for (int row = 0; row < grid.length; row++) {
            for (int col = 0; col < grid[row].length; col++) {
                rowMajor += grid[row][col] + " ";
            }
        }
        System.out.println("row-major:    " + rowMajor);

        String colMajor = "";
        for (int col = 0; col < grid[0].length; col++) {
            for (int row = 0; row < grid.length; row++) {
                colMajor += grid[row][col] + " ";
            }
        }
        System.out.println("column-major: " + colMajor);
    }
}

TraversalOrder.main(null);
````

### B. Finding a seat.  The SeatSearch class returns the location of a score and stops early

Common starting mistake:

<pre><code class="language-java">
return new int[] {0, 0};     // wrong: 0 is a real seat, so it can't mean "not found"
return new int[] {-1, -1};   // right: -1 is not a valid index
</code></pre>

Terminology:

`return` leaves the whole method from any depth, so the search stops at the first match. Looking for a score that is not in the grid is always the worst case, because every seat must be checked.

---

````java
// CODE_RUNNER: Predict both results before running, then search for 87 and watch how many seats are visited
// Search with an early return
import java.util.Arrays;

public class SeatSearch {
    static int visited = 0;

    public static int[] findSeat(int[][] chart, int target) {
        visited = 0;
        for (int row = 0; row < chart.length; row++) {
            for (int col = 0; col < chart[row].length; col++) {
                visited++;
                if (chart[row][col] == target) {
                    return new int[] {row, col};
                }
            }
        }
        return new int[] {-1, -1};
    }

    public static void main(String[] args) {
        int[][] scoreChart = {
            {87, 92, -1, 76, 88, 91},
            {-1, 65, 72, 89, -1, 94},
            {90, 88, 77, -1, 82, 79},
            {68, -1, 95, 91, 84, -1},
            {73, 86, 90, 88, -1, 77}
        };

        int[] found = findSeat(scoreChart, 94);
        System.out.println("94  -> " + Arrays.toString(found) + ", visited " + visited + " of 30 seats");

        int[] missing = findSeat(scoreChart, 100);
        System.out.println("100 -> " + Arrays.toString(missing) + ", visited " + visited + " of 30 seats");
    }
}

SeatSearch.main(null);
````

### C. The headcount.  The ClassSummary class counts seats and averages scores in one traversal

Common starting mistake:

<pre><code class="language-java">
double avg = totalScore / occupiedCount;            // int / int truncates first
double avg = (double) totalScore / occupiedCount;   // cast first, then divide
</code></pre>

Terminology:

`emptyCount`, `occupiedCount` and `totalScore` are **accumulators** updated in a single pass. Divide by the *occupied* seats, since empty seats never took the quiz, and guard against dividing by zero if the whole class is absent.

---

````java
// CODE_RUNNER: Predict both averages before running, then change one score and run it again
// One traversal, three accumulators
public class ClassSummary {
    public static void main(String[] args) {
        int[][] scoreChart = {
            {87, 92, -1, 76, 88, 91},
            {-1, 65, 72, 89, -1, 94},
            {90, 88, 77, -1, 82, 79},
            {68, -1, 95, 91, 84, -1},
            {73, 86, 90, 88, -1, 77}
        };

        int emptyCount = 0, occupiedCount = 0, totalScore = 0;
        for (int row = 0; row < scoreChart.length; row++) {
            for (int col = 0; col < scoreChart[row].length; col++) {
                if (scoreChart[row][col] == -1) {
                    emptyCount++;
                } else {
                    occupiedCount++;
                    totalScore += scoreChart[row][col];
                }
            }
        }

        System.out.println("empty: " + emptyCount + ", occupied: " + occupiedCount);
        System.out.println("int division: " + (totalScore / occupiedCount));
        System.out.println("with a cast:  " + ((double) totalScore / occupiedCount));
    }
}

ClassSummary.main(null);
````

### Popcorn Hack (In-Class)

2-minute challenge: analyze your own homework minutes for 3 school weeks (4 days each), then run the cell.

1. Store the minutes in an `int[][]` with one row per week
2. Find the most minutes in one day and where it happened
3. Count the days with at least 60 minutes

Replace the sample values with your own.

````java
// CODE_RUNNER: Replace the sample values with your own, then run it
// Practice #1 - Try it yourself first!
public class MyHomeworkGrid {
    public static void main(String[] args) {
        // Try writing your own code here first!

        // Sample answer:
        int[][] minutes = {
            {45, 60, 30, 75},
            {20, 90, 55, 40},
            {65, 35, 80, 50}
        };

        int best = -1, bestRow = -1, bestCol = -1, count = 0;
        for (int row = 0; row < minutes.length; row++) {
            for (int col = 0; col < minutes[row].length; col++) {
                if (minutes[row][col] > best) {
                    best = minutes[row][col];
                    bestRow = row;
                    bestCol = col;
                }
                if (minutes[row][col] >= 60) count++;
            }
        }
        System.out.println("Most: " + best + " (week " + (bestRow + 1) + ", day " + (bestCol + 1) + "), days at 60+: " + count);
    }
}

MyHomeworkGrid.main(null);
````
