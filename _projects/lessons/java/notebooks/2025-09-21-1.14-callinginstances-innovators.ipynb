## 1. Reference Guide

| Term | Definition / Example | College Board Requirement |
|---|---|---|
| **Instance method** | A method that works on one object's own data, called on that object: `myDog.bark();` | Unit 1, Topic 1.14, p. 48 |
| **Dot operator** | Connects an object to the method you call on it: `object.method()` | Unit 1, Topic 1.14, p. 48 |
| **Static (class) method** | A method that belongs to the class, called on the class name: `Math.sqrt(9)` | Unit 1, Topic 1.10 |
| **`void` method** | Does something but returns nothing, so its call is a statement: `c.add(5);` | Unit 1, Topic 1.9 |
| **Non-`void` method** | Returns a value you can store or use: `int n = c.getCount();` | Unit 1, Topic 1.9 |
| **`NullPointerException`** | Run-time error from calling a method on a `null` reference: `Dog d = null; d.bark();` | Unit 1, Topic 1.14, p. 48 |

- An instance method reads and changes the fields of the object it was called on
- Each object has its own fields, so the same call on two objects can give two different answers

### Instance Methods vs. Static Methods

| Feature | Instance method | Static (class) method |
| --- | --- | --- |
| How you call it | `myCookie.addChips(2)` | `Cookie.ovenTempF()` |
| Needs an object? | Yes | No |
| Can use the object's fields? | Yes | No |
| Common uses | Behavior tied to one object's state | Utilities and class-wide information |

### Rules for Calling Instance Methods (AP Exam Focus)

1. You need an object reference: `Dog.bark()` does not compile unless `bark` is `static`.
2. The reference cannot be `null`: calling a method on `null` throws a `NullPointerException` at run time.
3. Arguments must match the parameters in number, order, and type (or widen to the type).
4. A `void` method returns nothing: `int x = d.bark();` does not compile if `bark` is `void`.
5. A non-`void` method's result can be stored or used in an expression: `r1.getArea() + r2.getArea()`.
6. Inside the class, an instance method can call another instance method of the same object without a dot: `forward(50);`.
7. Parameter names in the method header do not matter to the caller. Only the arguments' values, order, and types do.

### Predicting a Call

- **The call is on a class name** → it compiles only if the method is `static`
- **The variable was never given an object** → it does not compile if it is a local variable that was never assigned; it throws a `NullPointerException` if it holds `null`
- **The result is stored in a variable** → the method must not be `void`, and its return type must fit the variable
- **Two objects call the same method** → each one uses its own fields

### College Board Requirements

AP CSA Unit 1, Topic 1.14 Calling Instance Methods. Quoted from the course and exam description (College Board, 2025, p. 48):

- "Instance methods are called on objects of the class. The dot operator is used along with the object name to call instance methods."

## 2. Real-Life Application
### One Order, Many Phones

When you tap "Order" in a food delivery app, the app does not call a method on "orders" in general. It calls methods on your one order object: `myOrder.addItem(...)`, `myOrder.getTotal()`. Thousands of people order at the same time, and every order keeps its own items and total. Everyone guess: what would go wrong if `getTotal()` belonged to the class instead of to each order?

| Real Case | What Happened | Java Idea in This Lesson |
|:---:|:---|:---|
| Food delivery orders | Every customer's order keeps its own items and total | Each object has its own fields, so `a.getTotal()` and `b.getTotal()` differ |
| Game characters | `player.jump()` moves one character, not every character | An instance method acts on the object it is called on |
| A missing account | The app reads a profile that was never loaded and crashes | Calling a method on `null` throws a `NullPointerException` |
| A calculator app | `Math.sqrt(9)` needs no object | Static methods are called on the class name |

**The takeaway:** an instance method call says which object to use. Make sure that object exists, then use what the method gives back.

## 3. Tech Talk

**Learning Objective:** Call instance methods on objects with the dot operator, use the values non-`void` methods return, and tell instance methods from static methods.

**Success Criteria:** You can say whether a call compiles, whether it needs an object first, what it returns, and whether a `null` reference would make it throw a `NullPointerException`.

Terminology:

**Dot operator:** the object before the dot decides whose fields the method uses.

````java
// CODE_RUNNER: Predict both lines, run it, then make a third Dog and call bark on it
public class Dog {
    private String name;

    public Dog(String name) {
        this.name = name;
    }

    public void bark() {
        System.out.println(name + " says: Woof!");
    }

    public static void main(String[] args) {
        Dog fido = new Dog("Fido");
        Dog rex = new Dog("Rex");
        fido.bark();
        rex.bark();
    }
}

Dog.main(null);
````

#### B. Arguments and return values. The Counter class mixes void and non-void calls

`add` changes the object and returns nothing. `getCount` returns a value, so its call can be stored or printed:

<pre><code class="language-java">
c.add(5);                    // void
int val = c.getCount();      // non-void: returns an int
int x = c.add(5);            // does not compile: add is void
</code></pre>

Terminology:

**Return value:** what a non-`void` method sends back to the line that called it.

---

````java
// CODE_RUNNER: Predict the three counts, run it, then call add on c2 as well
public class Counter {
    private int count;

    public void add(int x) {
        count += x;
    }

    public int getCount() {
        return count;
    }

    public static void main(String[] args) {
        Counter c1 = new Counter();
        Counter c2 = new Counter();
        c1.add(5);
        c1.add(3);

        int val = c1.getCount();
        System.out.println("c1: " + val);
        System.out.println("c2: " + c2.getCount());
        System.out.println("Total: " + (c1.getCount() + c2.getCount()));
    }
}

Counter.main(null);
````

## 6. LxD Cycle Process

**Empathize:** When students see a class for the first time, they often call its methods on the class name (`Dog.bark()`), the way they call `Math.sqrt()`. They have not yet separated what belongs to the class (static) from what belongs to one object (instance), and they try to store the result of `void` methods.

**Define:**
- **POV:** CSA students need one reliable test for when a call needs an object and what it returns, because the AP exam repeatedly shows static and instance calls that look almost the same.
- **Learning Goal:** Students will call instance methods with the dot operator, use the values non-`void` methods return, tell instance methods from static methods, and predict when a call fails to compile or throws a `NullPointerException`.

**Ideate:**
- **HMW Question:** How might we give students one question to ask about any call: "which object is this for?"
- **HMW Question:** How might we make the difference between a `void` and a non-`void` call something students check every time?
- **Activity:** A cookie-recipe analogy, then add methods to Dog and Counter, fix a call that stores a `void` result, and build a grade tracker with two independent students.

**Prototype:**
- A reference guide with seven calling rules, runnable two-object and return-value examples, a method-writing Popcorn Hack with a checker, an MCQ knowledge check, three scaffolded homework hacks with self-checking output, and a grading rubric.
- A trial run showed that students who missed the first MC question all called `deposit` on the class name, so the cookie analogy and the instance-versus-static table come before any questions, and the Popcorn checker asks whether the class-name call compiles.
- Excellence means every call is on the right object and every return value is either used or deliberately ignored.

**Test:**
- Ask peers in peer review to complete the practice without additional explanation.
- Observe whether peers call methods on objects rather than on the class name, and whether they store results only from non-`void` methods.
- Compare your lesson with another that is posted.
- Use the findings to revise any instruction or rubric criterion that did not guide students clearly.
- On submission, collect evidence from runner output, MCQ results, AI grading and student explanations.
- After teaching, grading and analysis, come back and revise lesson to complete teaching cycle for continuous improvement.
