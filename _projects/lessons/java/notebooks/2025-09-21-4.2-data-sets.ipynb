## 1. Reference Guide

### Key Topics

| Term | Definition | Example |
|---|---|---|
| Array | An object that stores a fixed number of values of the same type under one name. | `int[] scores = {85, 92, 78, 90};` |
| Element | One value stored in the array. | `92` is an element of `scores` |
| Index | The position of an element. Indices start at 0. | `scores[1]` is `92` |
| Length | How many elements the array holds. Set at creation, never changes. | `scores.length` is `4` |
| Last valid index | Always one less than the length. | `scores[scores.length - 1]` is `90` |
| ArrayIndexOutOfBoundsException | Runtime error when an index is negative or is `length` or more. | `scores[4]` |

- An array is a **reference type**: the variable holds the address of the array object, not the values themselves.
- `length` is a **field**, not a method, so it never has parentheses.

### Array Syntax

| Task | Syntax | Example | Notes |
|---|---|---|---|
| Declare | `type[] name;` | `int[] scores;` | No array exists yet |
| Create with a size | `new type[n]` | `scores = new int[4];` | Every slot gets a default value |
| Initializer list | `{v1, v2, ...}` | `int[] scores = {85, 92, 78, 90};` | Length is the number of values |
| Access | `name[i]` | `scores[2]` | Returns `78` |
| Modify | `name[i] = value;` | `scores[2] = 80;` | Replaces the old value |
| Length | `name.length` | `scores.length` | Field, no `()` |

### Picking the Right Tool

- A few unrelated values (a name, an age) → separate variables
- Many values of the same kind that get processed the same way → an array
- The first value → `arr[0]`
- The last value → `arr[arr.length - 1]`
- Every value → a loop from `0` while `i < arr.length`
- A collection that needs to grow → build a new, larger array (or use `ArrayList` later in the unit)

### Tech Talk (5 minutes)

An array is a shelf of identical cubbies numbered left to right starting at 0. The shelf has a fixed number of slots, and every slot holds the same kind of item. To find something, Java does not search the shelf; it jumps straight to the numbered slot.

The index tells Java how many slots to skip from the start. Skipping zero slots gets you the first one, which is why the first index is 0 and the last is `length - 1`.

````java
// CODE_RUNNER: Run it, then change one score and predict the new output
public class ScoreShelf {
    public static void main(String[] args) {
        int[] scores = {85, 92, 78, 90};

        System.out.println("Number of scores: " + scores.length);
        System.out.println("First score (index 0): " + scores[0]);
        System.out.println("Third score (index 2): " + scores[2]);
        System.out.println("Last score (index " + (scores.length - 1) + "): " + scores[scores.length - 1]);

        scores[2] = 80;
        System.out.println("Third score after update: " + scores[2]);
    }
}

ScoreShelf.main(null);
````

## 5. Code Examples

### A. Separate variables vs. an array. The `ScoreComparison` class stores the same four scores both ways

Three steps get mixed up a lot:

```java
int[] scores;                    // declaration: names the variable, no array yet
scores = new int[4];             // creation: builds 4 slots, all 0
int[] quiz = {85, 92, 78, 90};   // initializer list: declare, create, and fill at once
```

````java
// CODE_RUNNER: Run it, then add a fifth score to both versions and count how many lines each one needed
public class ScoreComparison {
    public static void main(String[] args) {
        // Without an array: four names, no way to loop
        int score1 = 85;
        int score2 = 92;
        int score3 = 78;
        int score4 = 90;
        System.out.println("Separate: " + score1 + " " + score2 + " " + score3 + " " + score4);

        // With an array: one name, index can be a variable
        int[] scores = {85, 92, 78, 90};
        for (int i = 0; i < scores.length; i++) {
            System.out.println("scores[" + i + "] = " + scores[i]);
        }
    }
}

ScoreComparison.main(null);
````

**Terminology:**

The index inside the brackets can be any `int` expression, including a variable like `i`. That is what lets one loop handle 4 scores or 400.

### B. Bounds checking. The `BoundsCheck` class shows a valid index next to an invalid one

````java
// CODE_RUNNER: Run it, then try scores[-1] inside the try block and see what happens
public class BoundsCheck {
    public static void main(String[] args) {
        int[] scores = {85, 92, 78, 90};

        System.out.println("Valid indices: 0 to " + (scores.length - 1));
        System.out.println("scores[3] = " + scores[3]);

        try {
            System.out.println(scores[4]);
        } catch (ArrayIndexOutOfBoundsException e) {
            System.out.println("scores[4] failed: " + e.getMessage());
        }
    }
}

BoundsCheck.main(null);
````

Common length mistake:

```java
scores.length      // arrays use a field
name.length()      // String uses a method
list.size()        // ArrayList uses size()
```

**Terminology:**

The compiler cannot catch a bad index. The error only happens at runtime, when the program actually reaches for the missing slot.

### C. Tracing a loop. The `DoubledTrace` class walks every index and prints a changed value

````java
// CODE_RUNNER: Predict every line before running, then change < to <= and explain the new output
public class DoubledTrace {
    public static void printDoubled(int[] arr) {
        for (int i = 0; i < arr.length; i++) {
            System.out.println(arr[i] * 2);
        }
    }

    public static void main(String[] args) {
        int[] scores = {85, 92, 78, 90};
        printDoubled(scores);
    }
}

DoubledTrace.main(null);
````

<details markdown="1">
<summary>Click to reveal the trace</summary>

- `i = 0` → `85 * 2` → prints `170`
- `i = 1` → `92 * 2` → prints `184`
- `i = 2` → `78 * 2` → prints `156`
- `i = 3` → `90 * 2` → prints `180`
- `i = 4` → `4 < 4` is false → loop ends

With `<=`, the loop runs once more with `i = 4`, prints all four values, then throws `ArrayIndexOutOfBoundsException`.
