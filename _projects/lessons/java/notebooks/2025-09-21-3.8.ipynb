## Topic 3.8: Scope and Access

**Big idea:** *where* you declare a variable decides *who* can see it. AP MCQs on this topic almost always test one of three things: block scope, shadowing, or `private` access.

| Concept | One-line takeaway |
|---|---|
| Scope | A variable only exists inside the `{ }` block where it is declared |
| Shadowing | A local variable/parameter with the same name hides the instance variable |
| Access modifiers | `private` = this class only, `public` = any class |


## 1. Scope of Variables

Scope determines the **visibility** of a variable. A variable is only accessible within the block of code (`{ }`) where it is declared.

- **Local variables** — declared inside a method, constructor, or loop body. They are created when the block executes and destroyed as soon as the block ends. Accessing them outside their block is a **compiler error**.
- **Parameters** — declared in the method/constructor header. They behave like local variables that live for the whole method body.
- **Instance variables (fields)** — declared directly inside the class but outside any method. They are accessible by all non-static methods of that class.

```java
public class Counter {
    private int count;              // instance variable: visible in the whole class

    public void addMany(int times) { // parameter: visible in this method only
        int added = 0;               // local variable: visible in this method only
        for (int i = 0; i < times; i++) { // i: visible in the loop only
            count++;
            added++;
        }
        System.out.println(added);   // OK
        // System.out.println(i);    // COMPILE ERROR: i is out of scope
    }
}
```


## 2. Variable Shadowing

If a local variable (or parameter) has the **exact same name** as an instance variable, the local variable **shadows** the instance variable.

- Inside that method, the plain name refers to the **local** variable, not the instance variable.
- To reach the instance variable anyway, use the `this` keyword: `this.variableName`.

```java
public class Student {
    private String name;

    public Student(String name) {
        this.name = name;  // this.name = instance variable, name = parameter
    }

    public void printName() {
        String name = "Local";
        System.out.println(name);       // Local
        System.out.println(this.name);  // the instance variable
    }
}
```


## 3. Access Modifiers

- **`private`** — access restricted to the declaring class. Other classes cannot read or write it directly; they must go through public **accessor (getter)** or **mutator (setter)** methods.
- **`public`** — any other class can freely access the variable or method.

> **Note:** You cannot put `public` or `private` on a local variable. Access modifiers are only for class-level members.

```java
public class BankAccount {
    private double balance;                       // hidden from other classes

    public double getBalance() { return balance; } // accessor

    public void deposit(double amount) {           // mutator with validation
        if (amount > 0) {
            balance += amount;
        }
    }
}

// In another class:
BankAccount acct = new BankAccount();
// acct.balance = 1000000;   // COMPILE ERROR: balance has private access
acct.deposit(50);             // OK
System.out.println(acct.getBalance()); // 50.0
```


## Common AP Exam Traps

### Trap 1: Using loop variables after the loop

```java
for (int i = 0; i < 5; i++) {
    int result = i * 2;
}
System.out.println(result); // COMPILE ERROR! 'result' does not exist here.
```

**Why:** `result` is local to the `for` loop block. It dies the moment the loop body ends. The same is true for `i`.

**Fix:** declare the variable *before* the loop so it lives in the outer block.

```java
int result = 0;
for (int i = 0; i < 5; i++) {
    result = i * 2;
}
System.out.println(result); // 8
```

### Trap 2: The shadowing bug in a constructor

```java
public class Student {
    private String name; // instance variable

    public Student(String name) {
        name = name; // BUG! Assigns the parameter to itself.
                     // The instance variable stays null.
    }
}
```

**Fix:** use `this.name = name;` to explicitly update the instance variable.

> This compiles with **no error**, which is why it shows up on the exam: the question asks what is printed, and the answer is `null` (or `0` for an `int`).

### Trap 3: Re-declaring instead of assigning

```java
public class Game {
    private int score;

    public void reset() {
        int score = 0; // declares a NEW local variable; the field is untouched
    }
}
```

**Fix:** drop the type: `score = 0;`


## Summary Checklist for the Exam

| Variable Type | Declared In | Accessible In | `public`/`private` Allowed? |
|---|---|---|---|
| Instance variable | Inside the class, outside methods | Everywhere in the class | Yes (usually `private`) |
| Parameter | In the method/constructor header | Only inside that method/constructor | No |
| Local variable | Inside a method or loop body | Only inside that specific block | No |

**Quick checks when reading MCQ code:**

1. Find the `{ }` a variable was declared in. Is it being used outside of it? → compile error.
2. Does a parameter or local share a name with a field? → is `this.` used?
3. Is another class touching a `private` member directly? → compile error.
4. Does a line start with a type (`int score = ...`) inside a method? → it is a *new* local, not the field.


## Practice MCQs

**Question 1.** What is the result of this code?

```java
int total = 0;
for (int k = 1; k <= 3; k++) {
    int sq = k * k;
    total += sq;
}
System.out.println(total + " " + k);
```

A. `14 3`  B. `14 4`  C. `14 0`  D. Compile error  E. Runtime exception

<details><summary>Answer</summary>
<b>D.</b> <code>k</code> is declared in the <code>for</code> header, so it is out of scope after the loop.
</details>

**Question 2.** What is printed?

```java
public class Dog {
    private int age;
    public Dog(int age) { age = age; }
    public int getAge() { return age; }
}
// ...
Dog d = new Dog(5);
System.out.println(d.getAge());
```

A. `5`  B. `0`  C. `null`  D. Compile error  E. Runtime exception

<details><summary>Answer</summary>
<b>B.</b> The parameter shadows the field, so <code>age = age</code> assigns the parameter to itself. The field keeps its default value, <code>0</code>.
</details>

**Question 3.** `Point` has `private int x;` and `public int getX()`. Which line, written in a *different* class, compiles?

A. `int a = p.x;`  B. `p.x = 3;`  C. `int a = p.getX();`  D. `int a = Point.x;`  E. `private int a = p.getX();` (inside a method)

<details><summary>Answer</summary>
<b>C.</b> A and B access a private field directly, D treats an instance variable as static, and E puts an access modifier on a local variable.
</details>

**Question 4.** What is printed?

```java
public class Box {
    private int size = 10;
    public void grow() {
        int size = 20;
        size++;
        this.size++;
    }
    public int getSize() { return size; }
}
// ...
Box b = new Box();
b.grow();
System.out.println(b.getSize());
```

A. `10`  B. `11`  C. `20`  D. `21`  E. Compile error

<details><summary>Answer</summary>
<b>B.</b> <code>size++</code> changes the local (20 → 21, then discarded). Only <code>this.size++</code> touches the field: 10 → 11.
</details>
