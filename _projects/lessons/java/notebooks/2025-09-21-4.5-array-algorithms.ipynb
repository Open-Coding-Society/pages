<h3>Key Topics</h3>

<table>
    <thead>
        <tr>
            <th>Term</th>
            <th>Definition</th>
            <th>Example / Context</th>
        </tr>
    </thead>
    <tbody>
        <tr>
            <td><strong>Accumulator</strong></td>
            <td>A variable declared before a loop used to build up a result, such as a sum or count.</td>
            <td><code>int sum = 0;</code> before looping to add <code>arr[i]</code> to <code>sum</code>.</td>
        </tr>
        <tr>
            <td><strong>Min / Max Algorithm</strong></td>
            <td>An algorithm that assumes the first element is the extreme value, then updates if a more extreme value is found.</td>
            <td><code>int max = arr[0];</code> rather than starting at <code>0</code> to handle negative values.</td>
        </tr>
        <tr>
            <td><strong>Flag Variable</strong></td>
            <td>A boolean variable used to record whether a certain condition was met during the traversal.</td>
            <td><code>boolean hasZero = false;</code> updated to <code>true</code> if <code>arr[i] == 0</code> is found.</td>
        </tr>
        <tr>
            <td><strong>Shifting Elements</strong></td>
            <td>Moving all elements in an array one index left or right, often requiring a temporary variable.</td>
            <td>Moving <code>arr[i+1]</code> to <code>arr[i]</code> inside a loop.</td>
        </tr>
    </tbody>
</table>

<ul>
    <li>
        <strong>Contextual Rule:</strong>
        Standard array algorithms are the building blocks of most data processing tasks.
        Always initialize search variables correctly before the loop, and return the final
        result after the loop finishes executing.
    </li>
</ul>

### Slide 3: Corrected Implementation


```java
// CODE_RUNNER: Run it, then change one score and predict the new output
// Corrected Shift Left Implementation
int[] arr = {10, 20, 30, 40};
int temp = arr[0]; // Save the first element

for (int i = 0; i < arr.length - 1; i++) {
    arr[i] = arr[i + 1]; // Shift remaining elements left
}
arr[arr.length - 1] = temp; // Place first element at the end

// Print result: [20, 30, 40, 10]
```

**Title:** Temperature Data Processor Lab

**Instructions:**
Complete the static methods in the `TemperatureAnalyzer` class below using standard 1D array traversal algorithms.


```java
// CODE_RUNNER: Run it, then change one score and predict the new output
public class TemperatureAnalyzer {

    /**
     * Counts how many times two consecutive days both had 
     * temperatures strictly greater than 90 degrees.
     */
    public static int countHeatWaves(int[] temps) {
        int count = 0;
        // TODO: Implement using a loop bound of temps.length - 1
        for (int i = 0; i < temps.length - 1; i++) {
            if (temps[i] > 90 && temps[i + 1] > 90) {
                count++;
            }
        }
        return count;
    }

    /**
     * Reverses the elements of the temps array in-place 
     * using a while loop and a swap mechanism.
     */
    public static void reverseLog(int[] temps) {
        int left = 0;
        int right = temps.length - 1;
        while (left < right) {
            int temp = temps[left];
            temps[left] = temps[right];
            temps[right] = temp;
            left++;
            right--;
        }
    }
}
```
