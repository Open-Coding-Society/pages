## 1. Reference Guide

| Term | Definition / Example | College Board Requirement |
|---|---|---|
| **Instantiation** | Creating an object from a class with `new`: `Dog myDog = new Dog("Rex");` | Unit 1, Topic 1.13, p. 46 |
| **Constructor** | Code with the same name as the class that sets up a new object: `public Dog(String name)` | Unit 1, Topic 1.13, p. 46 |
| **Overloaded constructors** | Several constructors with different parameter lists: `Dog()` and `Dog(String name)` | Unit 1, Topic 1.13, p. 46 |
| **Reference variable** | Holds the address of an object, not the object itself: `Dog d` | Unit 1, Topic 1.13, p. 46 |
| **`null`** | A reference that points to no object: `Dog d = null;` | Unit 1, Topic 1.13, p. 46 |
| **Alias** | Two variables holding the same reference, so both name one object: `Dog b = a;` | — |

- `new` creates the object and calls a constructor; the variable stores where the object is
- Java always copies the variable when you assign it or pass it to a method. For an object, the copy is the address, so both copies reach the same object

### Primitive vs. Reference Variables

| | Primitive (`int`, `double`, `boolean`) | Reference (any object) |
| --- | --- | --- |
| What the variable holds | The value itself | The address of an object |
| `b = a;` copies | The value: two separate numbers | The address: two names for one object |
| Changing through `b` | Never affects `a` | `b.setName(...)` changes the object `a` sees too |
| Passed to a method | The method gets a copy of the value | The method gets a copy of the address |

### Where the Data Lives (Beyond the AP Exam)

| | Stack | Heap |
| --- | --- | --- |
| Stores | Local variables: primitives and references | Every object |
| Freed | When the method returns | When no reference points to it any more |
| Picture it as | Your desk, with small items and notes | A playroom of toy boxes your notes point to |

### Constructor Types

| Type | Behavior | Example |
| --- | --- | --- |
| No-argument constructor | Sets default values | `new Student()` |
| Parameterized constructor | Sets values from its parameters | `new Student("Soni", 11)` |

### Predicting What a Method Changes

- **The method changes a primitive parameter** → the caller's variable does not change
- **The method changes an object's data through its parameter** (`p.age++`) → the caller sees the change, because it is the same object
- **The method assigns a new object to its parameter** (`p = new Person(...)`) → the caller's variable does not change; only the method's copy of the address moved
- **A variable is `null`** → calling a method on it throws a `NullPointerException`

## 2. Real-Life Application
### The Billion-Dollar Mistake

In 1965, computer scientist Tony Hoare added the null reference to a programming language he was designing, because it was easy to build. In a 2009 talk he called it his "billion-dollar mistake": decades of crashes in many languages, Java included, came from code that used a reference that pointed to nothing. Everyone guess: what happens when your program calls a method on a variable that was never given an object?

| Real Case | What Happened | Java Idea in This Lesson |
|:---:|:---|:---|
| Null references (1965 to today) | Code used a reference that pointed to no object, and the program crashed | `Dog d = null; d.bark();` throws a `NullPointerException` |
| A shared document link | Everyone with the link edits the same file | `Dog b = a;` makes an alias: two names, one object |
| A downloaded copy | Editing your copy does not change the original | `new Dog(a.getName())` makes a separate object |
| A house address | The address on paper is not the house | A reference variable holds where the object is |

**The takeaway:** a variable of a class type holds a reference. Know whether two variables share one object or point to separate ones, and never call a method through `null`.

## 3. Tech Talk

**Learning Objective:** Create objects with `new` and constructors, explain what a reference variable stores, and predict what changes when two variables or a method share one object.

**Success Criteria:** You can call the right constructor, tell an alias from a copy, and predict whether a change made inside a method shows up back in `main`.

A class is the cookie cutter and each object is a cookie. `new` makes a cookie and the constructor decorates it:

<pre><code class="language-java">
Dog a = new Dog("Rex");   // a new object; a holds its address
Dog b = a;                // no new object: b holds the same address
b.setName("Max");
System.out.println(a.getName());   // Max: a and b are aliases
</code></pre>

Primitives behave differently, because the variable holds the value itself:

<pre><code class="language-java">
int x = 10;
int y = x;                // a copy of the value
y = 20;
System.out.println(x);    // 10
</code></pre>

### Code Examples

#### A. Constructors. The Student class has a no-argument constructor and a parameterized one

Java picks the constructor whose parameter list matches the arguments:

<pre><code class="language-java">
new Student()               // matches Student()
new Student("Soni", 11)     // matches Student(String name, int grade)
new Student("Soni")         // no matching constructor: does not compile
</code></pre>

Terminology:

**Overloading** means one class has several constructors with the same name and different parameter lists.

---

```java
// CODE_RUNNER: Predict both lines, run it, then add a third Student
public class Student {
    private String name;
    private int grade;

    // No-argument constructor: default values
    public Student() {
        name = "New Student";
        grade = 9;
    }

    // Parameterized constructor: values from the caller
    public Student(String name, int grade) {
        this.name = name;
        this.grade = grade;
    }

    public void display() {
        System.out.println("Name: " + name + ", Grade: " + grade);
    }

    public static void main(String[] args) {
        Student s1 = new Student();
        Student s2 = new Student("Soni", 11);
        s1.display();
        s2.display();
    }
}

Student.main(null);
```

#### B. Aliases and method calls. The ReferenceDemo class compares a primitive, an alias, and a method that changes an object

Watch three cases side by side:

<pre><code class="language-java">
int b = a;              // copy of a value
Counter c2 = c1;        // copy of an address: an alias
addTen(c1);             // the method gets a copy of the address too
</code></pre>

Terminology:

**Alias:** two reference variables holding the same address. A change made through either one is a change to the one shared object.

---

```java
// CODE_RUNNER: Predict all four lines, run it, then make c2 a separate Counter
public class ReferenceDemo {
    static class Counter {
        int value;

        Counter(int value) {
            this.value = value;
        }
    }

    public static void addTen(int n) {
        n += 10;              // changes only the method's copy
    }

    public static void addTen(Counter c) {
        c.value += 10;        // changes the shared object
    }

    public static void main(String[] args) {
        int a = 5;
        int b = a;
        b = 99;
        System.out.println("a = " + a);

        Counter c1 = new Counter(5);
        Counter c2 = c1;      // alias, not a copy
        c2.value = 99;
        System.out.println("c1.value = " + c1.value);

        addTen(a);
        System.out.println("a after addTen = " + a);
        addTen(c1);
        System.out.println("c1.value after addTen = " + c1.value);
    }
}

ReferenceDemo.main(null);
```

## 6. LxD Cycle Process

**Empathize:** Students think `=` always makes an independent copy, the way it does for `int`. When two object variables end up naming the same object and a change through one shows up in the other, they assume something broke, instead of seeing the defined behavior of references.

**Define:**
- **POV:** CSA students need a picture of what a reference variable holds, because without it, the behavior of primitives and objects looks like two inconsistent rules instead of one rule (always copy the variable) applied to two kinds of values.
- **Learning Goal:** Students will create objects with constructors, explain what a reference variable stores, and predict whether a change made through an alias or inside a method affects the caller.

**Ideate:**
- **HMW Question:** How might we make something invisible, the address in a variable, visible enough that students can predict output?
- **HMW Question:** How might we separate "change the object" from "point the variable somewhere else," which look alike in code?
- **Activity:** A desk-and-playroom analogy for where values live, then predict-then-run hacks on aliases, constructors, and references passed to methods.

**Prototype:**
- A reference guide with a primitive-versus-reference table, runnable constructor and alias examples, a prediction Popcorn Hack with a checker, an MCQ knowledge check, three scaffolded homework hacks with self-checking output, and a grading rubric.
- A trial run showed that students could recite the rule but still predicted the wrong output for `reassignPerson`, so the Reference Guide now contrasts changing an object with reassigning a parameter, and the Popcorn Hack asks for both predictions.
- The old version called object parameters "pass-by-reference." Java always passes a copy of the variable; for an object, that copy is the reference. The lesson now uses that wording.

**Test:**
- Ask peers in peer review to complete the practice without additional explanation.
- Observe whether peers predict the `reassignPerson` and `tryReplace` results correctly, and whether they say "copy of the reference."
- Compare your lesson with another that is posted.
- Use the findings to revise any instruction or rubric criterion that did not guide students clearly.
- On submission, collect evidence from runner output, MCQ results, AI grading and student explanations.
- After teaching, grading and analysis, come back and revise lesson to complete teaching cycle for continuous improvement.
