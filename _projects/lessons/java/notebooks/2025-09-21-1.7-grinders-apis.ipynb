## 1. Reference Guide

| Term | Definition / Example | College Board Requirement |
|---|---|---|
| **Library** | A collection of classes someone already wrote and tested: Java's standard library | Unit 1, Topic 1.7, p. 39 |
| **API** | The documentation that says how to use a library's classes: the `String` API lists `length()` and `substring(int, int)` | Unit 1, Topic 1.7, p. 39 |
| **Package** | A named group of related classes: `java.util` holds `ArrayList` and `Scanner` | Unit 1, Topic 1.7, p. 39 |
| **`import`** | Lets you use a class from another package by its short name: `import java.util.ArrayList;` | — |
| **Attribute** | Data an object has, stored in variables: a book's `title` and `pages` | Unit 1, Topic 1.7, p. 39 |
| **Behavior** | What an object can do, defined by methods: a book's `isLong()` | Unit 1, Topic 1.7, p. 39 |

- Classes in `java.lang`, like `String` and `Math`, never need an `import`
- An API entry gives the method's name, parameter types, and return type. That is everything you need to call it correctly

### Common Java Packages

| Package | Description | Example Classes | Import needed? |
| --- | --- | --- | --- |
| `java.lang` | Core classes | `String`, `Math`, `Integer` | No |
| `java.util` | Utility classes | `ArrayList`, `Scanner`, `Random` | Yes |
| `java.io` | Input and output | `File`, `FileReader` | Yes |

### Reading an API Entry

| API Entry | Return Type | Call It Like This |
| --- | --- | --- |
| `static double sqrt(double a)` | `double` | `double r = Math.sqrt(64);` |
| `static double pow(double a, double b)` | `double` | `double p = Math.pow(3, 4);` |
| `int length()` | `int` | `int n = text.length();` |
| `String substring(int beginIndex, int endIndex)` | `String` | `String s = text.substring(0, 8);` |
| `boolean add(E e)` (ArrayList) | `boolean` | `list.add("red");` |
| `int size()` (ArrayList) | `int` | `int n = list.size();` |

### Attributes vs. Behaviors

| Concept | Definition | Example |
| --- | --- | --- |
| **Attributes** | What an object HAS: data stored in variables | `name`, `grade`, `gpa` |
| **Behaviors** | What an object DOES: actions defined by methods | `displayInfo()`, `updateGPA()` |

### Picking the Right Call

- **The entry says `static`** → call it on the class name: `Math.sqrt(64)`
- **The entry has no `static`** → call it on an object: `text.length()`
- **The return type is not `void`** → store or use the result
- **The class is not in `java.lang`** → add an `import` at the top of the file

## 2. Real-Life Application
### Eleven Lines That Broke Thousands of Projects

In March 2016, a developer removed a tiny library called `left-pad` from npm, the package registry for JavaScript. It had about eleven lines of code that added spaces to the left of a string. Thousands of projects, including the build tools behind big websites, used it, so their builds started failing within hours. Everyone guess: why would huge projects depend on a library that small instead of writing those eleven lines themselves?

| Real Case | What Happened | Java Idea in This Lesson |
|:---:|:---|:---|
| `left-pad` (2016) | Projects used a tiny library through its API, so removing it broke them all at once | Using a library means depending on its API |
| Square roots | Nobody writes their own square root routine | `Math.sqrt(64)` is already written and tested |
| Lists that grow | Java's `ArrayList` handles resizing for you | `import java.util.ArrayList;`, then call `add` and `size` |
| A school app | A student record has data and actions | Attributes like `gpa`, behaviors like `updateGPA()` |

**The takeaway:** libraries save you from rewriting tested code, and the API tells you exactly how to call them. Read the documentation before you write your own version.

#### A. Library classes without an import. The LibraryDemo class calls Math and String from their API entries

Every call below comes from an API entry in the Reference Guide. `Math` methods are `static`, so they are called on the class name. `String` methods are called on a `String` object:

<pre><code class="language-java">

text.substring(0, 8)          // String substring(int beginIndex, int endIndex)
</code></pre>

Terminology:

**java.lang** is the package Java imports for you, which is why `String` and `Math` need no `import` line.

```java
// CODE_RUNNER: Predict each line from its API entry, then run it
public class LibraryDemo {
    public static void main(String[] args) {
        // Math: static methods, called on the class name
        System.out.println("Square root of 16: " + Math.sqrt(16));
        System.out.println("3 to the 4th: " + Math.pow(3, 4));
        System.out.println("Absolute value of -5: " + Math.abs(-5));

        // String: methods called on a String object
        String text = "Computer Science";
        System.out.println("Length: " + text.length());
        System.out.println("First 8 chars: " + text.substring(0, 8));
        System.out.println("Index of Science: " + text.indexOf("Science"));
    }
}

LibraryDemo.main(null);
```

#### B. Attributes, behaviors, and an import. The StudentRecord class keeps its courses in an ArrayList

`StudentRecord` has three attributes and four behaviors. One attribute is an `ArrayList`, which lives in `java.util`, so the file needs an `import`. Every `ArrayList` call (`add`, `size`, `get`) comes from its API.

<pre><code class="language-java">
import java.util.ArrayList;   // without this line, ArrayList does not compile
</code></pre>

Terminology:

**Attributes** are the variables an object keeps. **Behaviors** are the methods that read or change them.

---

```java
// CODE_RUNNER: Label each member as an attribute or a behavior, then run it
import java.util.ArrayList;

public class StudentRecord {
    // Attributes: what a student record HAS
    private String name;
    private double gpa;
    private ArrayList<String> courses;

    public StudentRecord(String name, double gpa) {
        this.name = name;
        this.gpa = gpa;
        this.courses = new ArrayList<>();
    }

    // Behaviors: what a student record DOES
    public void enroll(String course) {
        courses.add(course);           // ArrayList API: boolean add(E e)
    }

    public void updateGPA(double newGPA) {
        gpa = newGPA;
    }

    public boolean isHonorRoll() {
        return gpa >= 3.5;
    }

    public void displayInfo() {
        System.out.println(name + " - GPA " + gpa + " - " + courses.size() + " courses");  // int size()
        System.out.println("First course: " + courses.get(0));                              // E get(int index)
    }

    public static void main(String[] args) {
        StudentRecord bob = new StudentRecord("Bob", 3.4);
        bob.enroll("AP CSA");
        bob.enroll("Physics");
        bob.updateGPA(3.8);
        bob.displayInfo();
        System.out.println("Honor roll: " + bob.isHonorRoll());
    }
}

StudentRecord.main(null);
```

| API Entry | What It Does |

## 6. LxD Cycle Process

**Empathize:** Students write their own sorting, text, or random-number code from scratch, even though Java already has tested classes that do it. They treat API documentation as optional reading instead of as the main way to learn what a class can do, and they use "library," "API," and "package" as if they meant the same thing.

**Define:**
- **POV:** CSA students need to be comfortable reading API documentation, because real Java work and the AP exam's provided classes assume you can use a class without reading its source code.
- **Learning Goal:** Students will call library methods from their API entries, explain how libraries, APIs, and packages relate, and tell a class's attributes from its behaviors.

**Ideate:**
- **HMW Question:** How might we teach students to trust and read documentation instead of rewriting what already exists?
- **HMW Question:** How might we make the return type in an API entry something students check before they call a method?
- **Activity:** Read API entries like a menu, predict a call's output from its return type, then use new entries to finish a class and edit a list.

**Prototype:**
- A reference guide with an API-entry table, runnable library and attributes-and-behaviors examples, a documentation Popcorn Hack with a checker, an MCQ knowledge check, three scaffolded homework hacks with self-checking output, and a grading rubric.
- A trial run showed students mixing up "API," "library," and "package," so the three terms sit side by side in the Reference Guide, next to the attributes and behaviors table.
- Excellence means calling an unfamiliar method correctly from its documentation on the first try and citing the entry used.

**Test:**
- Ask peers in peer review to complete the practice without additional explanation.
- Observe whether peers look up the API entry before calling a method, and whether they predict `81.0` rather than `81` for `Math.pow`.
- Compare your lesson with another that is posted.
- Use the findings to revise any instruction or rubric criterion that did not guide students clearly.
- On submission, collect evidence from runner output, MCQ results, AI grading and student explanations.
- After teaching, grading and analysis, come back and revise lesson to complete teaching cycle for continuous improvement.
