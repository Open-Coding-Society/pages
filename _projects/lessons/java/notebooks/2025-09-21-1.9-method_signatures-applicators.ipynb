### Introduction
In Java, a method’s signature determines which method is called. Understanding what is and is not part of a signature is essential for reading APIs and writing overloaded methods.


## Quick Reference

* **Signature (AP scope)** = method name + parameter list only. `add(int, int)` — not the return type, not the access modifier, not parameter names.
* **Cannot overload by return type alone.** Same name + same parameters + different return type = compile error.
* **Parameter** = variable in the header (`int a`). **Argument** = value passed at the call (`add(3, 4)`).
* **Overloading** = same name, different parameter list (type, count, or order).
* **Resolution:** Java auto-widens (`int` → `double`) but never auto-narrows (needs a cast).

### Tech Talk

```
public static int add(int a, int b)
```
| Part | Example | Note |
| :--- | :--- | :--- |
| Access modifier | `public` | Not part of the signature |
| Return type | `int` | Not part of the signature |
| Method name | `add` | **Part of the signature** |
| Parameter list | `(int a, int b)` | **Part of the signature** |

Signature: `add(int, int)`. Two methods can share a name only if their parameter lists differ — that's overloading.

### Example — overloading and resolution

````python
public class SigDemo {
    public static int add(int a, int b)    { return a + b; }     // add(int, int)
    public static double add(double a, double b) { return a + b; } // add(double, double)
    public static int add(int a)           { return a + 1; }     // add(int)

    public static void main(String[] args) {
        System.out.println(add(3, 4));   // 7   -> add(int, int)
        System.out.println(add(3.0, 4)); // 7.0 -> add(double, double), 4 widens
        System.out.println(add(5));      // 6   -> add(int)
    }
}
````
