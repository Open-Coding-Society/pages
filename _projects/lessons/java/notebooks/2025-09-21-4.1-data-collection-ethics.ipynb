<h2>2. LxD Cycle Process</h2>

<h3>Empathize</h3>
<p>
    Students often think code syntax is purely objective and neutral, assuming that because a program compiles without syntax errors, its outcomes must be fair.
</p>

<h3>Define</h3>

<ul>
    <li>
        <strong>POV:</strong> CSA students need to see how "fitness" in computer science means checking if data is suitable for the purpose of the algorithm, because training a model on a subset of the population means it will fail when applied to everyone else.
    </li>
    <li>
        <strong>Learning Goal:</strong> Students will analyze data collection practices, identify algorithmic bias, and propose ethical corrections to a flawed data collection pipeline.
    </li>
</ul>

<h3>Ideate</h3>

<ul>
    <li>
        <strong>HMW Question:</strong> How might we get students to audit a dataset to ensure equity before deploying the model?
    </li>
    <li>
        <strong>HMW Question:</strong> How might we show that perfectly bug-free Java code can still cause real-world harm?
    </li>
    <li>
        <strong>Activity:</strong> Discuss an interactive case study about a flawed fitness app and break into pairs to analyze a university admissions algorithm.
    </li>
</ul>

<h3>Prototype</h3>

<ul>
    <li>
        A slide deck on core concepts, an interactive case study, a Socratic discussion activity, and a student homework assignment handout.
    </li>
    <li>
        Students write a 300-word reflection analyzing the root cause of algorithmic failure and detailing their developer response.
    </li>
</ul>

<h3>Test</h3>

</ul>

<h4>Slide 2: Core Concepts</h4>

<ul>
    <li>
        <strong>Privacy &amp; Data Ownership:</strong>
        Who owns user data once collected? What risks exist when sensitive data is breached or sold?
    </li>
    <li>
        <strong>Algorithmic Bias:</strong>
        Automated systems in hiring, lending, and law enforcement reinforcing historical biases.
    </li>
    <li>
        <strong>Data Set Fitness:</strong>
        Using incomplete, unrepresentative, or skewed datasets leads to incorrect conclusions.
    </li>
    <li>
        <strong>Teacher Script:</strong>
        "When we talk about 'fitness' in computer science, we mean: Is this data suitable for the purpose of this algorithm? If you train a model on data that only represents a subset of the population, your algorithm will fail when applied to everyone else."
    </li>
</ul>

<h4>Slide 3: Interactive Case Study — "The Flawed Fitness App"</h4>

<ul>
    <li>
        <strong>Scenario:</strong>
        A tech company launches an AI health app to predict cardiac risk. The dataset used to train the algorithm consisted entirely of 20-something male professional athletes.
    </li>
    <li>
        <strong>Class Discussion Questions:</strong>
        <ol>
            <li>What happens when this software evaluates an elderly patient or a female patient?</li>
            <li>Where did the failure occur—in the Java logic, or in the data collection phase?</li>
            <li>Who is legally and ethically responsible for this mistake?</li>
        </ol>
    </li>
</ul>
