## 1. Reference Guide

| Term | Definition / Example | College Board Requirement |
|---|---|---|
| **Comment** | Text for people that the compiler ignores | Unit 1, Topic 1.8, p. 40 |
| **Single-line comment** | `//` comments out the rest of one line: `int x = 5; // starting value` | Unit 1, Topic 1.8, p. 40 |
| **Block comment** | `/* */` comments out everything between the markers, across lines | Unit 1, Topic 1.8, p. 40 |
| **Javadoc comment** | `/** */` before a class or method, used to build API documentation | Unit 1, Topic 1.8, p. 40 |
| **Precondition** | What must be true before a method is called. The method does not have to check it | Unit 1, Topic 1.8, p. 40 |
| **Postcondition** | What is always true after the method runs: the return value or the object's new state | Unit 1, Topic 1.8, p. 40 |

- Commented-out code does not run, so a method inside `/* */` does not exist as far as the compiler is concerned
- Comments explain why the code does something; the code already shows what it does

### Javadoc Template and Tags

<pre><code class="language-java">
/**
 * One sentence saying what the method does.
 *
 * Precondition: what the caller must make true first.
 * Postcondition: what is true when the method returns.
 *
 * @param name what the parameter means and its allowed values
 * @return what the method gives back
 */
</code></pre>

| Tag | Describes |
| --- | --- |
| `@param` | One parameter: its meaning and allowed values |
| `@return` | The value the method returns |
| `@throws` | An exception the method might throw |
| `@author`, `@version` | Who wrote a class and which version it is (class comments) |

### Documentation Best Practices

| Do | Don't |
| --- | --- |
| Be specific: "returns the average of the scores, from 0.0 to 100.0" | Be vague: "processes the data" |
| Document edge cases: what happens with 0, negatives, or an empty input | Document a getter whose name already says everything |
| Update the comment when the code changes | Let the comment describe old code |

### Deciding Whether to Document

- **The method has rules for its inputs** → write them as preconditions
- **The method changes an object or can return different kinds of answers** → write the postcondition for each case
- **The name and signature already say everything** (`getName()`) → a comment adds nothing
- **The code is commented out** → it does not run; delete it or bring it back

AP CSA Unit 1, Topic 1.8 Documentation with Comments. Quoted from the course and exam description (College Board, 2025, p. 40):

- "Comments are written for both the original programmer and other programmers to understand the code and its functionality, but are ignored by the compiler and are not executed when the program is run. Three types of comments in Java include /* */, which generates a block of comments; //, which generates a comment on one line; and /** */, which are Javadoc comments and are used to create API documentation."

## 2. Real-Life Application
### The Comment That Became Famous

The Unix operating system's source code from the 1970s contains a comment that programmers still quote: "You are not expected to understand this." It sat above a few lines that switched the computer between running programs. Dennis Ritchie, one of its authors, later explained that the lines relied on a subtle trick that was hard to explain, and later versions of Unix replaced it. Everyone guess: if those lines had a bug, how would the next programmer know what they were supposed to do?

| Real Case | What Happened | Java Idea in This Lesson |
|:---:|:---|:---|
| Unix source code (1975) | A comment admitted the code was hard to follow, but did not say what it had to do | A good comment states the rules: preconditions and postconditions |
| Java's own API pages | Every page on Oracle's Java documentation site is generated from Javadoc comments | `/** */` plus `@param` and `@return` |
| A team project | A teammate calls your method with a negative number | A precondition tells them that is not allowed |
| AP free-response questions | Each method comes with a comment that states its preconditions and what it must do | Reading the comment is part of answering the question |

**The takeaway:** code says how. Documentation says what the method promises and what it needs, so the next person can use it without guessing.

## 3. Tech Talk

**Learning Objective:** Write single-line, block, and Javadoc comments, and describe a method with its preconditions and postconditions.

**Success Criteria:** You can document a method with `@param` and `@return`, state its preconditions and postconditions, and explain why a simple getter does not need a comment.

Java has three kinds of comments, and the compiler ignores all of them:

<pre><code class="language-java">
int x = 5;     // single line: the rest of the line is a comment
/* block: everything between the
   markers is a comment */
/** Javadoc: describes the class or method right below it */
</code></pre>

A method's documentation is a contract. The **precondition** is the caller's side: what must be true first. The **postcondition** is the method's side: what will be true after it runs.

### Code Examples

#### A. A method's contract. The GradeCalculator class documents its parameters, return value, and conditions

The comment says what the method needs and promises. Anyone can call `finalGrade` without reading its body:

<pre><code class="language-java">
@param homework the average homework score, from 0.0 to 100.0
@return the weighted final grade, from 0.0 to 100.0
</code></pre>

Terminology:

**Javadoc** is a `/** */` comment placed right before a method or class. Tools read it to build pages like Java's own API documentation.

---

```java
// CODE_RUNNER: Read the Javadoc, predict the grade, then run it
public class GradeCalculator {
    /**
     * Calculates a weighted final grade: homework 30%, tests 50%, participation 20%.
     *
     * Precondition: every score is from 0.0 to 100.0.
     * Postcondition: the result is from 0.0 to 100.0.
     *
     * @param homework the average homework score
     * @param tests the average test score
     * @param participation the participation score
     * @return the weighted final grade
     */
    public static double finalGrade(double homework, double tests, double participation) {
        return homework * 0.3 + tests * 0.5 + participation * 0.2;
    }

    public static void main(String[] args) {
        System.out.println("Final grade: " + finalGrade(80.0, 90.0, 100.0));
    }
}

GradeCalculator.main(null);
```

#### B. When to document. The BankAccount class documents withdraw and leaves getBalance alone

`withdraw` has rules, two possible outcomes, and a change to the object, so it gets a full contract. `getBalance` is obvious from its name, so it gets none:

<pre><code class="language-java">
public double getBalance() { return balance; }   // no comment needed
</code></pre>

Terminology:

**Postcondition** describes every outcome. Here that means both the `true` path (the money is taken) and the `false` path (nothing changes).

---

```java
// CODE_RUNNER: Predict both withdrawals from the postcondition, then run it
public class BankAccount {
    private double balance;

    public BankAccount(double startingBalance) {
        balance = startingBalance;
    }

    /**
     * Takes money out of the account if there is enough.
     *
     * Precondition: amount is greater than 0.
     * Postcondition: if amount is no more than the balance, the balance goes down by
     * amount and the method returns true; otherwise the balance is unchanged and the
     * method returns false.
     *
     * @param amount the amount to withdraw
     * @return true if the money was withdrawn, false if the balance was too low
     */
    public boolean withdraw(double amount) {
        if (amount > balance) {
            return false;
        }
        balance -= amount;
        return true;
    }

    public double getBalance() {
        return balance;
    }

    public static void main(String[] args) {
        BankAccount account = new BankAccount(500.0);
        System.out.println("Withdraw 100: " + account.withdraw(100.0) + ", balance " + account.getBalance());
        System.out.println("Withdraw 900: " + account.withdraw(900.0) + ", balance " + account.getBalance());
    }
}

BankAccount.main(null);
```

## 6. LxD Cycle Process

**Empathize:** Students treat comments as an afterthought, added at the end to satisfy a rubric. They skip documentation on the methods that need it and over-document getters that do not. They also confuse a precondition with code that checks the precondition.

**Define:**
- **POV:** CSA students need to see documentation as a contract between the method and its caller, because AP free-response questions and real teammates rely on it to know what a method needs and promises.
- **Learning Goal:** Students will write `//`, `/* */`, and Javadoc comments, state preconditions and postconditions, and decide when documentation is needed.

**Ideate:**
- **HMW Question:** How might we get students to document why and what a method promises, instead of restating the code?
- **HMW Question:** How might we show that code and its documentation have to agree?
- **Activity:** Rewrite a useless comment as a Javadoc, bring back commented-out code, then document an enrollment rule and fix the code so it matches.

**Prototype:**
- A reference guide with a Javadoc template, runnable contract and when-to-document examples, a documentation Popcorn Hack with a checker, an MCQ knowledge check, three scaffolded homework hacks with self-checking output, a reflection, and a grading rubric.
- Peer review found that an earlier example did not compile and that preconditions appeared in prose long before they appeared in real Javadoc, so every example now compiles and shows its contract inside the comment.
- Excellence means a specific, accurate contract, with code that does what the comment promises.

**Test:**
- Ask peers in peer review to complete the practice without additional explanation.
- Observe whether peers write the precondition and postcondition before they edit code, and whether they leave simple getters uncommented.
- Compare your lesson with another that is posted.
- Use the findings to revise any instruction or rubric criterion that did not guide students clearly.
- On submission, collect evidence from runner output, MCQ results, AI grading and student explanations.
- After teaching, grading and analysis, come back and revise lesson to complete teaching cycle for continuous improvement.
