## 1. Reference Guide

| Term | Definition / Example | College Board Requirement |
|---|---|---|
| **Casting** | Writing a type in parentheses to convert a value: `(int) 5.9`, `(double) 7` | Unit 1, Topic 1.5, p. 36 |
| **Widening** | `int` to `double` happens automatically, with no lost data: `double y = 5;` stores `5.0` | Unit 1, Topic 1.5, p. 36 |
| **Narrowing** | `double` to `int` needs a cast, and the decimal part is dropped: `(int) 5.9` is `5` | Unit 1, Topic 1.5, p. 36 |
| **Truncation** | Dropping the decimal part, toward zero: `(int) -5.9` is `-5` | Unit 1, Topic 1.5, p. 36 |
| **Rounding** | Add 0.5 before the cast for positives, subtract 0.5 for negatives: `(int) (5.5 + 0.5)` is `6` | Unit 1, Topic 1.5, p. 36 |
| **`Integer.MAX_VALUE` / `MIN_VALUE`** | The largest and smallest `int`: `2147483647` and `-2147483648` | Unit 1, Topic 1.5, p. 37 |
| **Integer overflow** | A result outside the `int` range wraps around: `Integer.MAX_VALUE + 1` is `-2147483648` | Unit 1, Topic 1.5, p. 37 |

- A cast applies only to the value right after it, so `(double) (7 / 2)` is `3.0` but `(double) 7 / 2` is `3.5`
- Overflow does not crash and gives no error. The program keeps running with the wrong number

### Type Casting Rules

| Direction | Name | Behavior | Example |
| --- | --- | --- | --- |
| `int` to `double` | Widening (automatic) | No loss of data | `int x = 5; double y = x;` gives `5.0` |
| `double` to `int` | Narrowing (needs a cast) | Truncates toward zero | `double d = 5.9; int x = (int) d;` gives `5` |

### Integer Division vs Floating-Point Division

| Situation | Result | Example |
| --- | --- | --- |
| Both operands `int` | `int`, decimals dropped | `7 / 2` is `3` |
| At least one `double` | `double` | `7.0 / 2` is `3.5` |
| Cast before dividing | The cast happens first | `(double) 7 / 2` is `3.5` |
| Cast after dividing | Too late, the decimal is already gone | `(double) (7 / 2)` is `3.0` |

### Truncation vs Rounding

| Operation | Behavior | Example |
| --- | --- | --- |
| `(int) x` | Truncate toward zero | `(int) 5.9` is `5`, `(int) -5.9` is `-5` |
| `(int) (x + 0.5)` | Round a non-negative number | `(int) (5.4 + 0.5)` is `5`, `(int) (5.5 + 0.5)` is `6` |
| `(int) (x - 0.5)` | Round a negative number | `(int) (-5.4 - 0.5)` is `-5`, `(int) (-5.5 - 0.5)` is `-6` |

### Integer Range

| Type | Minimum | Maximum | Notes |
| --- | --- | --- | --- |
| `int` | `-2147483648` | `2147483647` | 32 bits, -2^31 to 2^31 - 1 |
| `double` | very large negative | very large positive | about 15 significant digits |

### Predicting a Mixed-Type Expression

- **Both values are `int`** → integer division, decimals dropped
- **A cast sits right before a value** → that value converts first, then the math happens
- **A cast sits before parentheses** → the math inside happens first, then the result converts
- **An `int` result would pass 2147483647** → it wraps to a negative number

### College Board Requirements

AP CSA Unit 1, Topic 1.5 Casting and Range of Variables. Quoted from the course and exam description (College Board, 2025, pp. 36-37):

- "The casting operators (int) and (double) can be used to convert from a double value to an int value (or vice versa)... Casting a double value to an int value causes the digits to the right of the decimal point to be truncated."
- "If an expression would evaluate to an int value outside of the allowed range, an integer overflow occurs. The result is an int value in the allowed range but not necessarily the value expected."

The Java language specification gives the rule behind truncation: a floating-point value converted to an integer "is rounded to an integer value V using the round toward zero rounding policy" (Gosling et al., 2023, section 5.1.3). That is why `(int) 5.9` is `5`, and why rounding needs the `+ 0.5` step.

## 2. Real-Life Application
### The Video That Ran Out of Numbers

In December 2014, YouTube announced that one music video had been watched more than 2,147,483,647 times. That number is `Integer.MAX_VALUE`. Its view counter was a 32-bit integer, and one more view would have wrapped the count to a negative number. YouTube switched the counter to a 64-bit integer before that happened. Everyone guess: what would the counter have shown after view number 2,147,483,648?

| Real Case | What Happened | Java Idea in This Lesson |
|:---:|:---|:---|
| YouTube view counter (2014) | A 32-bit counter reached its limit, so YouTube made it 64 bits wide | `Integer.MAX_VALUE + 1` wraps to `-2147483648` |
| Boeing 787 power units (2015) | The U.S. Federal Aviation Administration ordered 787 operators to restart the planes' generator control units at least every 248 days, because a 32-bit counter of hundredths of a second would overflow and shut the generators down | 2^31 hundredths of a second is about 248 days. Overflow gives no error, it just gives a wrong value |
| Grade averages | `int` points divided by `int` assignments drops the decimal | `(double) total / count` keeps it |
| Star ratings | 4.6 stars shown as 4 instead of 5 | `(int) 4.6` truncates; `(int) (4.6 + 0.5)` rounds |

**The takeaway:** know the range of your type and decide when the conversion happens. Neither overflow nor truncation gives an error message.

## 3. Tech Talk

**Learning Objective:** Predict the result of casting between `int` and `double`, explain integer division and truncation, round with a cast, and recognize integer overflow.

**Success Criteria:** You can trace a mixed-type expression, place a cast so the division keeps its decimal, round positive and negative numbers, and say when an `int` will overflow.

Java checks types strictly. When types mix, the result depends on what each value is at the moment the operator runs.

<pre><code class="language-java">
System.out.println(7 / 2);              // 3    (int / int)
System.out.println((double) 7 / 2);     // 3.5  (7 becomes 7.0 first)
System.out.println((double) (7 / 2));   // 3.0  (7 / 2 is already 3)
System.out.println((int) 3.9);          // 3    (truncated, not rounded)
System.out.println((int) (3.9 + 0.5));  // 4    (rounded)
</code></pre>

Overflow is silent:

<pre><code class="language-java">
int max = Integer.MAX_VALUE;            // 2147483647
System.out.println(max + 1);            // -2147483648
</code></pre>

### Code Examples

#### A. Where the cast goes. The DivisionAndCasting class shows integer division, cast placement, truncation, and rounding

Students often put the cast in the wrong place:

<pre><code class="language-java">
double avg = (double) (sum / count);   // too late: sum / count already dropped the decimal
double avg = (double) sum / count;     // sum becomes a double first, so the decimal stays
</code></pre>

Terminology:

**Narrowing** a `double` to an `int` always truncates toward zero. To round, shift the value by 0.5 first, then cast.

---

```java
// CODE_RUNNER: Predict each line, run it, then move one cast and predict again
public class DivisionAndCasting {
    public static void main(String[] args) {
        int sum = 10, count = 3;

        System.out.println(sum / count);              // 3: int / int
        System.out.println(sum % count);              // 1: the remainder
        System.out.println((double) sum / count);     // 3.3333333333333335: cast first
        System.out.println((double) (sum / count));   // 3.0: cast after the division

        double d = 5.7;
        System.out.println((int) d);                  // 5: truncated
        System.out.println((int) (d + 0.5));          // 6: rounded
        System.out.println((int) (-d - 0.5));         // -6: rounding a negative
    }
}

DivisionAndCasting.main(null);
```

#### B. Range and overflow. The OverflowDemo class shows what happens past the edges of int

An `int` has 32 bits. Going one past either end wraps around to the other end, with no error.

<pre><code class="language-java">
Integer.MAX_VALUE + 1   // -2147483648
Integer.MIN_VALUE - 1   //  2147483647
</code></pre>

Terminology:

**Integer overflow:** the true answer does not fit in an `int`, so Java keeps a value that does fit, which is usually far from the true answer.

---

```java
// CODE_RUNNER: Predict the two wrap-around lines, run it, then change the coin values
public class OverflowDemo {
    public static void main(String[] args) {
        int max = Integer.MAX_VALUE;
        int min = Integer.MIN_VALUE;
        System.out.println("Max int: " + max);
        System.out.println("Min int: " + min);

        System.out.println("Max + 1: " + (max + 1));   // wraps to the minimum
        System.out.println("Min - 1: " + (min - 1));   // wraps to the maximum

        // A game wallet that is almost full
        int coins = 2000000000;
        coins = coins + 500000000;
        System.out.println("Coins: " + coins);         // negative: overflow
    }
}

OverflowDemo.main(null);
```

## 6. LxD Cycle Process

**Empathize:** Students make casting errors without seeing what is happening. Their code compiles but prints the wrong answer, because they do not expect integer division, truncation instead of rounding, or overflow. They treat a cast as a magic symbol rather than a conversion that happens at one specific moment.

**Define:**
- **POV:** CSA students need a clear rule for when a conversion happens, because a wrong cast gives a wrong answer with no error message.
- **Learning Goal:** Students will predict the results of casts between `int` and `double`, explain integer division and truncation, round with a cast, and recognize overflow in a 32-bit `int`.

**Ideate:**
- **HMW Question:** How might we get students to think about the type of each value at the moment each operator runs?
- **HMW Question:** How might we make silent errors like overflow visible before they ship?
- **Activity:** Predict casting results, run them to check, then fix averages, rounding, and a wallet that overflows.

**Prototype:**
- A reference guide, runnable examples for cast placement and overflow, a prediction Popcorn Hack with a checker, an MCQ knowledge check, three scaffolded homework hacks with self-checking output, and a grading rubric.
- A trial run showed that introducing widening, narrowing, division, rounding, and overflow at once was too much, so the lesson goes from cast placement to rounding to overflow, and each one has its own hack.
- Excellence means predicting the exact output, including where the cast takes effect, rather than only producing working code.

**Test:**
- Ask peers in peer review to complete the practice without additional explanation.
- Observe whether peers write predictions before they press run, and whether they put the cast before or after the division.
- Compare your lesson with another that is posted.
- Use the findings to revise any instruction or rubric criterion that did not guide students clearly.
- On submission, collect evidence from runner output, MCQ results, AI grading and student explanations.
- After teaching, grading and analysis, come back and revise lesson to complete teaching cycle for continuous improvement.
