## 1. Reference Guide

### Key Topics

<br>

| Term | Definition | Example |
|------|------------|---------|
| **Class** | Blueprint that groups related data into one type. | `class Submission { ... }` |
| **Instance variable** | A field each object keeps its own copy of. | `int submitterId;` |
| **Constructor** | Sets an object's fields when it is created with `new`. | `new Submission(1, 101, 201, null)` |
| **Static** | Belongs to the class itself, not to any one object, so it is used with the class name instead of `new`. | `public static int countPending(...)` |
| **`ArrayList`** | A resizable list of object references. | `List<Submission> all = new ArrayList<>();` |
| **Traversal** | Visiting every element of a collection. | `for (Submission s : all)` |
| **Selection** | Choosing what to do with an `if` / `else if`. | `if (role == Role.ADMIN)` |
| **Role** | Which kind of user is viewing the data. | `STUDENT`, `TEACHER`, `ADMIN` |

- A `Double grade` of `null` means **not graded yet**, which a primitive `double` cannot express
- Filtering never changes the original list; it builds a **new** list of the matches

### Submissions Page Tabs

The live Submissions page (`/submissions`) has three tabs. Each tab is the same list of submissions, filtered a different way:

| Tab | Who sees it | Filter rule |
|-----|-------------|-------------|
| My Submissions | Every logged-in user | Keep only submissions where `submitterId` matches the viewer |
| Assignments You Created | Any teacher who authored an assignment | Keep only submissions whose assignment is owned by the viewer |
| All Submissions | Admins only | Keep every submission, no filter |

### Picking a Filter Rule

- **"What did I turn in?"** → compare `submitterId` to the viewer
- **"Who turned in my assignment?"** → compare `assignmentOwnerId` to the viewer
- **"Show me everything"** → no comparison, admins only
- **"Is it graded?"** → check whether `grade` is `null`

`enum` is outside the AP CSA Java subset. This lesson uses it for `Role` because it matches how the real backend names roles, but it will not be on the exam; a `String` or `int` constant would work the same way.

## 5. Code Examples

### A. Submission basics. The Submission class stores who turned it in, who owns the assignment, and the grade

Three parts of a class get mixed up a lot:

<pre><code class="language-java">
int submitterId;                               // instance variable: the data each object keeps
Submission(int id, ...) { this.id = id; ... }  // constructor: fills in the fields
new Submission(1, 101, 201, null)              // object: one real submission
</code></pre>

Terminology:

`grade` is a `Double`, not a `double`. `Double` is a **reference** type, so it can be `null`, which the site uses to mean "not graded yet."

---

```java
// CODE_RUNNER: Run it, then give the pending submission a grade and predict the new output
public class SubmissionBasics {
    static class Submission {
        int id;
        int submitterId;
        int assignmentOwnerId;
        Double grade;

        Submission(int id, int submitterId, int assignmentOwnerId, Double grade) {
            this.id = id;
            this.submitterId = submitterId;
            this.assignmentOwnerId = assignmentOwnerId;
            this.grade = grade;
        }
    }

    public static void main(String[] args) {
        Submission graded = new Submission(2, 102, 201, 0.9);
        Submission pending = new Submission(1, 101, 201, null);

        System.out.println("Submission " + graded.id + " by student " + graded.submitterId
                + " for teacher " + graded.assignmentOwnerId + ", grade: " + graded.grade);
        System.out.println("Submission " + pending.id + " by student " + pending.submitterId
                + " for teacher " + pending.assignmentOwnerId + ", grade: " + pending.grade);
        System.out.println("Is submission " + pending.id + " graded? " + (pending.grade != null));
    }
}
SubmissionBasics.main(null);
```

### B. Role filtering. The SubmissionRoles class filters one list three different ways

`visibleSubmissionsFor` walks a `List<Submission>` with an enhanced `for` loop and uses `if` / `else if` selection to decide whether each submission belongs in the result, based on the caller's `Role`.

Common filtering mistake:

<pre><code class="language-java">
allSubmissions.remove(submission);   // changes the list while looping: ConcurrentModificationException
visible.add(submission);             // build a new list of the matches instead
</code></pre>

Terminology:

**Security boundary:** this method only decides what a role is *shown*. On the real Open Coding Society backend, every submissions API endpoint must also check the caller's role and identity on the server before returning data. Filtering only in the browser would let anyone request the admin endpoint directly and see every student's submissions.

---

```java
// CODE_RUNNER: Run the three role checks, then change a submitterId or assignmentOwnerId and predict the new counts.
import java.util.ArrayList;
import java.util.List;

public class SubmissionRoles {
    enum Role { STUDENT, TEACHER, ADMIN }

    static class Submission {
        int id;
        int submitterId;
        int assignmentOwnerId;
        Double grade;

        Submission(int id, int submitterId, int assignmentOwnerId, Double grade) {
            this.id = id;
            this.submitterId = submitterId;
            this.assignmentOwnerId = assignmentOwnerId;
            this.grade = grade;
        }
    }

    public static List<Submission> visibleSubmissionsFor(
            List<Submission> allSubmissions, Role role, int userId) {
        List<Submission> visible = new ArrayList<>();
        for (Submission submission : allSubmissions) {
            if (role == Role.ADMIN) {
                visible.add(submission);
            } else if (role == Role.TEACHER && submission.assignmentOwnerId == userId) {
                visible.add(submission);
            } else if (role == Role.STUDENT && submission.submitterId == userId) {
                visible.add(submission);
            }
        }
        return visible;
    }

    public static void main(String[] args) {
        List<Submission> allSubmissions = new ArrayList<>();
        allSubmissions.add(new Submission(1, 101, 201, null));
        allSubmissions.add(new Submission(2, 102, 201, 0.9));
        allSubmissions.add(new Submission(3, 101, 202, null));
        allSubmissions.add(new Submission(4, 103, 202, 1.0));

        System.out.println("Student 101 sees: " + visibleSubmissionsFor(allSubmissions, Role.STUDENT, 101).size());
        System.out.println("Teacher 201 sees: " + visibleSubmissionsFor(allSubmissions, Role.TEACHER, 201).size());
        System.out.println("Admin sees: " + visibleSubmissionsFor(allSubmissions, Role.ADMIN, 0).size());
    }
}
SubmissionRoles.main(null);
```

### Popcorn Hack (In-Class)

2-minute challenge: put yourself in the submissions list, predict what each tab shows, then run the cell.

1. Pick your own student ID for `myId`
2. Pick a teacher ID for `teacherId`
3. Add two submissions where **you** are the submitter: one for `teacherId`'s assignment and one for a different teacher
4. Before running, write down how many rows each tab will show
5. Run it and compare with your prediction
6. Read the Popcorn checker under the runner. When every check passes, copy its line into your submission notes

Replace the sample values with your own.

```java
// CODE_RUNNER: Replace the sample IDs with your own, predict the three counts, then run it
// Practice #1 - Try it yourself first!
import java.util.ArrayList;
import java.util.List;

public class PopcornRoles {
    static class Submission {
        int submitterId;
        int assignmentOwnerId;

        Submission(int submitterId, int assignmentOwnerId) {
            this.submitterId = submitterId;
            this.assignmentOwnerId = assignmentOwnerId;
        }
    }

    public static void main(String[] args) {
        int myId = 500;       // your student ID
        int teacherId = 900;  // a teacher's ID

        List<Submission> submissions = new ArrayList<>();
        submissions.add(new Submission(101, teacherId));  // someone else's work

        // Try writing your own two submissions here first!

        // Sample answer:
        submissions.add(new Submission(myId, teacherId));
        submissions.add(new Submission(myId, 901));

        int mine = 0;
        int teachers = 0;
        for (Submission submission : submissions) {
            if (submission.submitterId == myId) {
                mine++;
            }
            if (submission.assignmentOwnerId == teacherId) {
                teachers++;
            }
        }

        System.out.println("My Submissions tab: " + mine);
        System.out.println("Assignments You Created tab (teacher " + teacherId + "): " + teachers);
        System.out.println("All Submissions tab: " + submissions.size());
    }
}

PopcornRoles.main(null);
```
