## 1. Reference Guide

| Concept                           | Key idea                                                            | Java example                                | Important note                                          |
| --------------------------------- | ------------------------------------------------------------------- | ------------------------------------------- | ------------------------------------------------------- |
| **Instance variable**             | Belongs to an individual object                                     | `private String name;`                      | Every object gets its own copy                          |
| **Class variable**                | Belongs to the class and is shared by all objects                   | `private static int numberOfStudents;`      | Declared with `static`                                  |
| **`static`**                      | Makes a variable/method belong to the class rather than an instance | `public static int count;`                  | Access class members through the class when appropriate |
| **Public class variable**         | A class variable accessible outside its class                       | `public static int players;`                | Access with `Game.players`                              |
| **Class method**                  | A method belonging to the class                                     | `public static int doubleValue(int x)`      | Can be called without creating an object                |
| **Static method access**          | Static methods can access static/class members                      | `count++`                                   | They **cannot directly access instance variables**      |
| **Dot operator**                  | Used to access public class members                                 | `MathHelper.doubleValue(5)`                 | Typically `ClassName.member`                            |
| **`final` variable**              | A variable whose value cannot be reassigned after initialization    | `public static final int DAYS_IN_WEEK = 7;` | Often used for constants                                |
| **Shared state**                  | All instances see the same class-variable value                     | `Student.numberOfStudents`                  | Changing it affects what all objects see                |
| **Constructor + static variable** | A constructor can update shared class state                         | `numberOfStudents++;`                       | Useful for tracking how many objects have been created  |

## 5. Code Examples

### Instance vs. Class Variables

Recall that an *instance variable belongs to an individual object*, so let's say we have:
```java
public class Student {
    private String name;
    private int grade;
}
```
and we define 3 students
```java
Student s1 = new Student();
Student s2 = new Student();
Student s3 = new Student();
```

Each student gets their own `name` and `grade`.

Class variables are different in that they *belong to the class itself*:
```java
public class Student {
    private String name;
    private int grade;

    public static int numberOfStudents = 0;
}
```

Now, each student `s1`, `s2`, `s3`, shares one copy of `numberOfStudents`.

For example, we could make a class:
```java
public class Student {
    private String name;

    public static int numberOfStudents = 0;

    public Student(String n) {
        name = n;
        numberOfStudents++;
    }
}
```
and create a few objects:
```java
Student a = new Student("Alice");
Student b = new Student("Bob");
Student c = new Student("Charlie");
```
which makes `Student.numberOfStudents` `3`.

### Static

We use `static` to denote that a variable or method belongs to the class itself rather than to an individual object.

This would be an instance variable:
```java
private String name;
```
and this would be a class variable:
```java
private static int numberOfStudents;
```

### Public and Private

If we want some class variable to be accessible outside of the class, we can declare it as `public`:
```java
public class Game {
    public static int players = 0;
}
```
which means we can do:
```java
System.out.println(Game.players); // get # of players
Game.players = 5; // modify # of players
```


### Class Methods

Similarly, we can define a method as `static` to denote it belongs to a class:
```java
public class MathHelper {

    public static int doubleValue(int x) {
        return x * 2;
    }
}
```

Now we can call it without creating a `MathHelper` object:
```java
int result = MathHelper.doubleValue(5);

System.out.println(result); // 10
```

Class methods can access class variables, other class methods, its parameters, and local variables, but NOT instance variables:
```java
public class Counter {

    private static int count = 0;

    public static void increase() {
        count++;
    }

    public static int getCount() {
        return count;
    }

    public static void reset() {
        count = 0;
    }

    public static void resetAndIncrease() {
        reset();
        increase();
    }
}
```

We CANNOT do this
```java
public class Student {
    private String name;

    public static void printName() {
        System.out.println(name); // ERROR
    }
}
```

### `final` Variables

We define a variable as `final` if we want it to be immutable.
```java
public static final int DAYS_IN_WEEK = 7;
```
