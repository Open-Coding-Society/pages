## Abtraction Key Ideas

**Abstraction:** Reduces complexity, focuses on main purpose while reducing unneccessary details.

**Data Abstraction:** Uses the name of the variable without exposing the actual structure or representation.

**Procedural Abstraction:** Hides what the function actually does, leaving only a name to explain details, effectively removing any unneccessary details.

**Method Decomposition:** Breaks large behaviors into smaller methods to hide the steps of any sub-tasks.

## 2. Lesson Plan

### Tech Talk

Abstraction means working with the important idea while leaving lower-level details inside the implementation. In Java, this happens in both the data a class stores and the behaviors its methods provide.

**Data abstraction** answers: *What information does this object represent?* A class might represent a library book with a title and availability state. Code using the object can work with those concepts without needing to know every storage decision behind them.

**Procedural abstraction** answers: *What operation does this code provide?* A method such as `calculateTotal()` gives the program a named behavior. The calling code can use that behavior without reproducing the steps that make it happen.

**Method decomposition** takes a behavior that is doing too many things and divides it into smaller methods with clear purposes. **Parameters** make those methods general enough to work with different inputs.

The AP CSA connection is class design: first identify the state and behaviors required by a specification, then implement those abstractions in Java.


### Code Examples

#### Data Abstraction

```java
public class Book {
    private String title;
    private boolean checkedOut;

    public Book(String title) {
        this.title = title;
        this.checkedOut = false;
    }
}
```

Both `title` and `checkedOut` represent the object state. Code that works with the `Book` class does not need to know what the actual state is.

#### Procedural Abstraction

```java
public static double calculateTotal(double price, int quantity) {
    return price * quantity;
}
```

The method named `calculateTotal` clearly names the operation so that code that works with the method knows what it does but not neccessarily how it does it. The parameters allow the same abstraction to work with a variety of prices and quantities.

#### Method Decomposition

```java
public class ScoreTracker {
    private int score;
    private int target;

    public ScoreTracker(int target) {
        this.score = 0;
        this.target = target;
    }

    public void processTurn(int points) {
        addPoints(points);
        if (hasReachedTarget()) {
            reportCompletion();
        }
    }

    private void addPoints(int points) {
        score += points;
    }

    private boolean hasReachedTarget() {
        return score >= target;
    }

    private void reportCompletion() {
        System.out.println("Target reached: " + score);
    }
}
```

`processTurn` is a high-level procedural abstraction, as it uses helper methods, `addPoints`, `hasReachedTarget`, and `reportCompletion` in order to hide the smaller details in implementation while still describing what is going on.

## 3. Lesson Hacks

### Popcorn Hack

**Task:** The code below mixes the calculation directly into `main`. Create a procedural abstraction named `calculateAverage` that accepts an array of scores and returns the average. Then use the method from `main`.

```java
public class PopcornAbstraction {
    public static void main(String[] args) {
        double[] scores = {80, 90, 100};
        double total = 0;

        for (double score : scores) {
            total += score;
        }

        double average = total / scores.length;
        System.out.println(average);
    }
}
```

### Homework Hack

**Task:** Refactor the following Java class so its responsibilities are represented through abstractions. Preserve the behavior of the program.

Your refactor should:

1. Keep `currentPoints` and `goalPoints` as instance variables that represent the object's state.
2. Replace the repeated scoring logic with a method named `addPoints` that accepts the number of points as a parameter.
3. Extract the goal check into a boolean method named `hasReachedGoal`.
4. Make `takeTurn` use those abstractions instead of performing every operation itself.

```java
public class PracticeGame {
    private int currentPoints = 0;
    private int goalPoints = 50;

    public void takeTurn(int earnedPoints) {
        currentPoints = currentPoints + earnedPoints;

        if (currentPoints >= goalPoints) {
            System.out.println("Goal reached");
        } else {
            System.out.println("Keep going");
        }
    }

    public void bonusTurn(int earnedPoints) {
        currentPoints = currentPoints + earnedPoints;

        if (currentPoints >= goalPoints) {
            System.out.println("Goal reached");
        } else {
            System.out.println("Keep going");
        }
    }
}
```
