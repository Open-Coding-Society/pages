## A capstone that continues across three trimesters

In CSA week 9, continue working in teams on an OpenCS capstone features that can grow across all three trimesters. 


## From community projects to enduring system contributions

Students enter this course having built a community or nonprofit project in the prior year's APCSP course. CSA extends that experience into system-level work.


| Stage | Project role | Individual evidence |
| --- | --- | --- |
| CSA1 - Trimester 1 | Establish a team role, design a feature, and build an initial Java/Web implementation. POJO, JPA, and APIs were introduced in week 4, building on APCSP Python experience. | Design plan, code contribution, integration tests, review ticket, and demonstration. |
| CSA2 - Trimester 2 | Strengthen CS112 foundations and introduce AP-aligned search, sort, and analysis within the evolving project. | CodeRunner competence, tested feature extensions, and explanations of revised design choices. |
| DS2 - Trimester 3 | Extend the project through differentiated CS112/CS113 implementation layers or the ML/AI pathway. | Selected structure or algorithm implementations, correctness and efficiency evidence, presentations, and reflection. |



## Sprint teams and feature portfolios

Choose or continue a team role on a team project.

1. **[Gamification / Onboarding]({{site.baseurl}}/capstone/toolchain-trail/):** Gamified lessons, GameBuilder, AI NPCs, game analytics, wayfinding, and teams. Design and test an interaction or progression feature; explain its state and data flow.
2. **[Admin / Security]({{site.baseurl}}/capstone/ocs-admin-security/):** Deployment, mentor management, DevOps, a code-builder UI, table reduction, and cybersecurity. Plan an administrative workflow and verify the relevant permissions, deployment behavior, or data changes.
3. **[Communication / Design]({{site.baseurl}}/capstone/submissions/):** UX, SASS, calendar, direct messages, and microblogging. Define a user flow, integrate the interface with its service, and test both success and error feedback.
4. **[AI / Assignments]({{site.baseurl}}/capstone/submissions/):** Student and teacher dashboards, student-success insights, articulation evidence, and analytics. Define the assignment data flow and validate the records, calculations, or feedback your feature produces.

## Team plan and beginning-to-end contribution

Create a team issue that connects ideation, design, development, integration, test, and retrospective. 

## Communication and wayfinding

Teacher and peer review support improvement. Keep a personal review ticket or blog that chronicles Sprint 1-3 accomplishments and links your team issues, code, tests, and feedback.

- **Temporal wayfinding:** Compare your progress and completed work with the calendar, announcements, and agreed timeline.
- **Social wayfinding:** Make your role visible, communicate dependencies, and support teammates inside and outside class.
- **Analytical wayfinding:** Use rubric feedback and evidence to explain what you know, what improved, and what needs further work. Activity counts alone do not demonstrate competence.

## Demonstration and reflection

Before CSA2 begins in week 13, every team completes **two mentor reviews and a community N@tM presentation**. Each review or presentation centers on the project, a working demonstration, and a discussion of **how it is made**: design choices, code, data flow, integration, and test evidence. This mentor-review and community-presentation cycle continues throughout the year as the capstone evolves.

### Grading guideline

- **55% - Minimum evidence:** Initial work is present but major requirements or explanations remain incomplete.
- **75% - Mostly complete:** Most requirements are demonstrated, with identified gaps in implementation, integration, or validation.
- **85% - Complete:** Required work is integrated, demonstrated, and supported by individual evidence.
- **90% - Thorough evidence:** Complete work is supported by clear design reasoning, relevant tests, feedback-driven revision, and reflection. This is the maximum in this guideline, not a claim of perfection.
