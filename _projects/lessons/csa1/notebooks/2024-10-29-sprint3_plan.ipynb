## A capstone that continues across three trimesters

In CSA week 9, continue working in teams on an OpenCS capstone that can grow across all three trimesters. This is not a one-sprint final assignment: preserve the project, its design decisions, and evidence of your contributions as you extend it.

## From community projects to enduring system contributions

Students enter this course having built a community or nonprofit project in the prior year's APCSP course. CSA extends that experience into system-level work: features are designed to integrate into the OCS system rather than remain isolated, short-term projects. Once integrated, they become foundations that future students can maintain and extend, helping successive cohorts build on prior success year after year.

Demonstrate how your feature fits the existing system: identify dependencies and interface contracts, test integration, document its design and operation, and record known limitations and next steps for a future maintainer. Distinguish a proposed feature or prototype from work that has actually been integrated. A lasting system contribution still requires individual evidence of the concepts you implemented and can explain.

| Stage | Project role | Individual evidence |
| --- | --- | --- |
| CSA1 - Trimester 1 | Establish a team role, design a feature, and build an initial Java/Web implementation. POJO, JPA, and APIs were introduced in week 4, building on APCSP Python experience. | Design plan, code contribution, integration tests, review ticket, and demonstration. |
| CSA2 - Trimester 2 | Strengthen CS112 foundations and introduce AP-aligned search, sort, and analysis within the evolving project. | CodeRunner competence, tested feature extensions, and explanations of revised design choices. |
| DS2 - Trimester 3 | Extend the project through differentiated CS112/CS113 implementation layers or the ML/AI pathway. | Selected structure or algorithm implementations, correctness and efficiency evidence, presentations, and reflection. |

Project participation is a context for learning, not proof of college credit or individual mastery. Link the actual design, code, tests, and explanations that demonstrate a concept. Not every student is expected to complete every advanced CS113 layer.


## Sprint teams and feature portfolios

Choose or continue a team role. The feature lists below describe opportunities; your team issue must identify the specific feature you will deliver, its acceptance checks, and each member's responsibilities.

1. **[Gamification / Onboarding]({{site.baseurl}}/capstone/toolchain-trail/):** Gamified lessons, GameBuilder, AI NPCs, game analytics, wayfinding, and teams. Design and test an interaction or progression feature; explain its state and data flow.
2. **[Admin / Security]({{site.baseurl}}/capstone/ocs-admin-security/):** Deployment, mentor management, DevOps, a code-builder UI, table reduction, and cybersecurity. Plan an administrative workflow and verify the relevant permissions, deployment behavior, or data changes.
3. **[Communication / Design]({{site.baseurl}}/capstone/submissions/):** UX, SASS, calendar, direct messages, and microblogging. Define a user flow, integrate the interface with its service, and test both success and error feedback.
4. **[AI / Assignments]({{site.baseurl}}/capstone/submissions/):** Student and teacher dashboards, student-success insights, articulation evidence, and analytics. Define the assignment data flow and validate the records, calculations, or feedback your feature produces.

Communication / Design and AI / Assignments share the assignments-system capstone link. Document their separate responsibilities and shared interface contracts in team issues. Some capstone features are live and others are proposals: identify the current stage of your selected feature rather than claiming planned work is implemented.

A feature called analytics, AI, or wayfinding does not by itself demonstrate a CS113 algorithm or data structure. Name the actual structure or algorithm, explain why it fits, and supply code and test evidence when making that connection.


## Team plan and beginning-to-end contribution

Create a team issue that connects ideation, design, development, integration, and validation. For the selected feature, record:

1. The user need, feature scope, and acceptance criteria.
2. A design artifact before coding: a model, API contract, pseudocode, UML, or data-flow diagram appropriate to the feature.
3. Assigned responsibilities and the interfaces between members' work.
4. Implementation links and an integration pull request, with a clear explanation of your contribution.
5. A test plan with inputs and expected results, followed by actual results. Include normal, boundary, and failure cases relevant to the feature.
6. Review feedback, the changes made in response, and remaining limitations.

Use existing code as a starting point, but distinguish what you observed, reused, changed, or implemented yourself. A group deployment does not replace each member's explanation and demonstration.

## Communication and wayfinding

Teacher and peer review support improvement. Keep a personal review ticket or blog that chronicles Sprint 1-3 accomplishments and links your team issues, code, tests, and feedback.

- **Temporal wayfinding:** Compare your progress and completed work with the calendar, announcements, and agreed timeline.
- **Social wayfinding:** Make your role visible, communicate dependencies, and support teammates inside and outside class.
- **Analytical wayfinding:** Use rubric feedback and evidence to explain what you know, what improved, and what needs further work. Activity counts alone do not demonstrate competence.

## Demonstration and reflection

Before CSA2 begins in week 13, every team completes **two mentor reviews and a community N@tM presentation**. Each review or presentation centers on the project, a working demonstration, and a discussion of **how it is made**: design choices, code, data flow, integration, and test evidence. This mentor-review and community-presentation cycle continues throughout the year as the capstone evolves.

For each event, retain the review ticket or presentation record, feedback received, and the changes made or planned in response. Each student should be able to explain their own contribution; attendance or a group demonstration alone does not establish individual competence.

Prepare the project for the sprint's closing review and N@tM presentation. Demonstrate the feature, trace a request through its relevant layers, explain your own code and tests, and identify a limitation. Record audience, peer, and teacher feedback and propose the next iteration. The week-9 plan establishes this evidence; presentations and summative reflection occur at the sprint's scheduled close, not as a new week-9 deadline.

### Grading guideline

- **55% - Minimum evidence:** Initial work is present but major requirements or explanations remain incomplete.
- **75% - Mostly complete:** Most requirements are demonstrated, with identified gaps in implementation, integration, or validation.
- **85% - Complete:** Required work is integrated, demonstrated, and supported by individual evidence.
- **90% - Thorough evidence:** Complete work is supported by clear design reasoning, relevant tests, feedback-driven revision, and reflection. This is the maximum in this guideline, not a claim of perfection.
