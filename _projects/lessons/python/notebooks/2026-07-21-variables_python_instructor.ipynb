### Tech Talk (3 minutes)

- A **variable** abstracts storage behind a name. It represents one current value, which can itself be a list. Use meaningful names such as `round_score`; names are case-sensitive and cannot begin with a digit or be a keyword.
- A **data type** describes the value: Python uses `int` (`10`), `float` (`1.5`), `str` (`"Nova"`) and `bool` (`True`). A Boolean represents true/false. JavaScript uses `number`, `string` and `boolean`; Python lists/JavaScript arrays group values in 3.2.
- `"10"` is text; `10` is a number. Choose numbers for arithmetic and strings for names or IDs such as `"007"`. Python `input()` returns text: `int("10") + 5` produces `15`. In JavaScript, `Number("10") + 5` does the same.
- Python/JavaScript use `=` to assign. Python `==` and JavaScript `===` compare values. AP pseudocode uses `←` to assign and `=` to compare.

**Initialize before use:** `score = 10` gives a starting value; `score = score + 5` reads that value and replaces it with 15. The latest assignment wins. Values carry types; Python names and JavaScript `let` variables can be reassigned to values of another type. Keep each variable’s purpose clear.

### Code Examples

#### A. Simple: Name and store values

````python
player_name = "Nova"
player_id = "007"
round_score = 10
game_active = True

print(player_name)
print(player_id)
print(round_score)
print(game_active)
````

**Output:** `Nova`, `007`, `10`, `True`. The ID is text so its leading zeros remain.

#### B. Intermediate: Trace an update

````python
score = 10
saved_score = score
score = score + 5
saved_score = saved_score + 2
print(score, saved_score)
````

**Output:** `15 12`. Both start at 10, then update independently. AP writes these updates as `score ← score + 5` and `savedScore ← savedScore + 2`; the final values match.

#### C. Complex: Recalculate a total

Predict before clicking **Run**. Change the bonus and run again.

````python
# CODE_RUNNER: 3.1C - Recalculate a stored total
player_name = "Nova"
round_one = 10
round_two = 20
round_three = 15

total_score = round_one + round_two + round_three
print(player_name, total_score)

round_two = round_two + 5
print("Stored total:", total_score)

total_score = round_one + round_two + round_three
print("Updated total:", total_score)
````

**Output:** `Nova 45`, `Stored total: 45`, `Updated total: 50`. The total stores a result, not a live formula.

**JavaScript comparison:** `let` allows reassignment; `const` prevents reassignment of that binding. JavaScript uses `console.log()` for output and lowercase `true`/`false`.

````python
%%js
// CODE_RUNNER: 3.1C JS - Recalculate a stored total
const playerName = "Nova";
const roundOne = 10;
let roundTwo = 20;
const roundThree = 15;

let totalScore = roundOne + roundTwo + roundThree;
console.log(playerName, totalScore);

roundTwo = roundTwo + 5;
console.log("Stored total:", totalScore);

totalScore = roundOne + roundTwo + roundThree;
console.log("Updated total:", totalScore);
````
