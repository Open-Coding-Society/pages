### **Part 1 — The Basics: What is a Math Expression?**

**Why this matters:** Every program that calculates a price, a score, a distance, or a time is built on math expressions. A math expression combines numbers and operators so the computer evaluates them into a single new value — without expressions, a program can only store numbers, it can never actually *do* anything with them.

Python uses the same basic operators as a calculator:


| Operator | Meaning | Description | Usage Example |
| :---: | :--- | :--- | :--- |
| **`+`** | Addition | Adds two values together. | `10 + 5` evaluates to `15` |
| **`-`** | Subtraction | Subtracts the second value from the first. | `10 - 5` evaluates to `5` |
| **`*`** | Multiplication | Multiplies two values. | `10 * 5` evaluates to `50` |
| **`/`** | Division | Divides the first value by the second. | `10 / 5` evaluates to `2.0` |
| **`**`** | Exponentiation | Raises the first value to the power of the second. | `2 ** 3` evaluates to `8` |
| **`//`** | Floor Division | Divides and rounds down to the nearest whole number. | `10 // 3` evaluates to `3` |
| **`%`** | Modulus | Divides and returns only the remainder. | `10 % 3` evaluates to `1` |

### **Going Deeper: Mathematical Expressions**

Most expressions you write use **infix notation** — the operator sits between the numbers, like `5 + 3`. Computers don't have to evaluate math that way.

RPN removes the need for parentheses and order-of-operations rules — every expression can be evaluated left to right using an algorithm.  Create a data structure stack (like plates), first-in, first-out.

Step | Input Queue | Operation | Stack
-- | -- | -- | --
1 | 5 3 - 2 * | Push 5 | [5]
2 | 3 - 2 * | Push 3 | [5, 3]
3 | - 2 * | Pop 3, pop 5 → 5 - 3 → push 2 | [2]
4 | 2 * | Push 2 | [2, 2]
5 | * | Pop 2, pop 2 → 2 × 2 → push 4 | [4]
6 | (empty) | Result | [4]

This is important as we often think the human representation of a problem is simple, but computer algorithms are required behind the scenes for everything we do on computers.  RPN is still used today in some scientific and financial calculators because it’s fast and never ambiguous.

**2. Consider the following expression:**

According to the standard order of operations used in AP CSP, what value is stored in `result`?
A) 14
B) 16
C) 10
D) 20

**3. In the AP CSP pseudo-code, which of the following expressions evaluates to 3.5?**
A) `7 MOD 2 + 2`
B) `7 / 2`
C) `14 MOD 4 / 2`
D) `3 + 1 / 2 * 0`

---

### Answer Key & Explanations

**1. Answer: B (a = 8, b = 6, c = 14)**
*   `a ← 4`, `b ← 6`, `c ← 2` (Initial values)
*   `a ← b + c` (a becomes 6 + 2 = 8)
*   `b ← a - c` (b becomes 8 - 2 = 6)
*   `c ← a + b` (c becomes 8 + 6 = 14)

**2. Answer: A (14)**
*   According to the AP CSP reference sheet, `MOD`, `*`, and `/` have the same precedence and are evaluated left to right before addition `+`.
*   Step 1: `17 MOD 5` evaluates to 2 (the remainder of 17 divided by 5).
*   Step 2: `2 * 2` evaluates to 4.
*   Step 3: `10 + 4` evaluates to 14.

**3. Answer: B (7 / 2)**
*   Unlike AP CSA where `7 / 2` would truncate to the integer `3`, AP CSP evaluates division to an exact decimal. Therefore, `7 / 2` evaluates exactly to `3.5`.
