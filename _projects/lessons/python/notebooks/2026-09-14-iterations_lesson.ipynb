# Iterations: RSFSC Fire Response

An iteration is a repeated action. In programming, loops let us repeat steps without writing the same code many times.

This lesson explores how the RSFSC (Rapid Safety Fire Suppression Crew) uses loops to check danger zones, spray water, and stop fires safely.

## The big idea

Iteration means repeating an action until a goal is reached.

In Python, loops give us a clean way to do that:

- `for` loops: repeat a known number of times or move through a list
- `while` loops: repeat until a condition becomes false
- `break`: stop the loop early
- `continue`: skip the current item and keep going

In real life, firefighters and emergency teams use this same idea every time they check zones, spray water, and watch for danger levels.

## RSFSC emergency response story

The RSFSC (Rapid Safety Fire Suppression Crew) gets a call about a growing fire. Instead of writing the same action over and over, the crew uses loops.

- A `for` loop checks every fire zone.
- A `while` loop keeps spraying water until the temperature drops.
- A `break` stops the loop when the main fire is found.
- A `continue` skips safe zones and keeps going.

Loops help the crew stay organized, repeat the plan safely, and finish the job faster.

````python
# CODE_RUNNER: RSFSC uses a for loop to inspect all fire zones and report their risk.
fire_zones = ["North Gate", "Warehouse", "Forest Line", "Boiler Room"]

for zone in fire_zones:
    print(f"Checking {zone}...")
    if zone == "Forest Line":
        print("High risk: send backup crew immediately.")
    else:
        print("Routine inspection complete.")
````

## `for` loops: inspect each fire zone

A `for` loop is best when you already know the work to do or you are iterating over a collection. RSFSC can check a list of zone names, temperatures, or danger levels without rewriting the same code.

```python
for zone in ["North Gate", "Warehouse", "Forest Line", "Boiler Room"]:
    print(f"Checking {zone}...")
```

The loop variable `zone` changes each time, so the same block of code can inspect a different area every iteration.

````python
# CODE_RUNNER: RSFSC keeps spraying water until the heat drops below 30.
heat = 85
spray_count = 0

while heat > 30:
    heat -= 15
    spray_count += 1
    print(f"Spraying water... heat is now {heat}.")

print(f"Fire contained after {spray_count} spray cycles.")
````

## `while` loops: keep spraying until the fire is contained

A `while` loop keeps running as long as a condition stays true. This is perfect for emergency response when the crew keeps working until the fire is under control.

```python
heat = 85
spray_count = 0

while heat > 30:
    heat -= 15
    spray_count += 1
    print(f"Spraying water... heat is now {heat}.")

print(f"Fire contained after {spray_count} spray cycles.")
```

The key is to make the condition eventually become false. If it never does, the loop runs forever.

````python
# CODE_RUNNER: RSFSC should skip safe zones and stop once the severe fire is found.
fire_levels = [7, 4, 2, 9, 1]

for level in fire_levels:
    if level <= 2:
        print(f"Zone is stable: {level}")
        continue
    print(f"Crew responds to level {level}")
    if level >= 8:
        print("Severe fire detected — focus on the main fire line.")
        break
````

## `break` and `continue` in an emergency

- `break` stops the whole loop immediately.
- `continue` skips the current item and keeps checking the rest.

RSFSC uses this logic when a fire is already contained or when a zone is safe.

```python
fire_levels = [7, 4, 2, 9, 1]
for level in fire_levels:
    if level <= 2:
        print(f"Zone is stable: {level}")
        continue
    print(f"Crew responds to level {level}")
    if level >= 8:
        print("Severe fire detected — focus on the main fire line.")
        break
```

This pattern is useful when the crew should skip harmless zones and stop once the worst fire is addressed.

## Quick challenge

Try to predict the output before you run the code.

A fire crew checks several zones. If the reading is 3 or higher, the zone is still active. If it is below 3, the zone is stable.

Use the loop below to count active fire zones.

````python
# CODE_RUNNER: RSFSC needs one final count. Count the fire zones that are still active and print the total.
alerts = [3, 2, 5, 1, 8, 4, 0]
active_zones = 0

for alert in alerts:
    if alert >= 3:
        active_zones += 1
        print(f"Alert at {alert}: fire still active")
    else:
        print(f"Alert at {alert}: zone is stable")

print(f"Total active fire zones: {active_zones}")
````
