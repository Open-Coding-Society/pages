# Iterations in Python

An iteration is a repeated action. In programming, loops let us repeat steps without writing the same code over and over.

This lesson focuses on the core Python patterns you use every time you process lists, check conditions, and stop a loop at the right moment.

## The big idea

Iteration means repeating an action until a goal is reached.

In Python, loops give us a clean way to do that:

- `for` loops: repeat a known number of times or move through a list
- `while` loops: repeat until a condition becomes false
- `break`: stop the loop early
- `continue`: skip the current item and keep going

These patterns are useful whenever you need to check a set of values, keep testing until a condition changes, or filter data before moving on.

## A simple real-world example

A monitoring system might check several room temperatures, only keep the high alerts, and stop when it finds the first critical reading.

- A `for` loop checks each temperature in a list.
- A `while` loop keeps cooling until the temperature falls under a safe threshold.
- A `break` stops the loop when the first critical warning is found.
- A `continue` skips safe readings so the program stays focused.

This is the same reasoning behind many real systems: repeat the work, filter the data, and stop when the condition is satisfied.

````python
%%python

# CODE_RUNNER: Use a for loop to inspect a list of sensor readings.
readings = [34, 42, 31, 48, 29, 37]

for reading in readings:
    print(f"Checking reading: {reading}")
    if reading >= 40:
        print("High alert: investigate this reading.")
    else:
        print("Reading is within the normal range.")
````

## `for` loops: inspect each item in a list

A `for` loop is best when you already know the work to do or you are iterating over a collection. You can check a list of values, names, temperatures, or scores without rewriting the same code.

```python
for value in [30, 42, 18, 55]:
    print(f"Checking {value}...")
```

The loop variable `value` changes each time, so the same block of code can inspect a different item every iteration.

````python
%%py

# CODE_RUNNER: RSFSC keeps spraying water until the heat drops below 30.
heat = 85
spray_count = 0

while heat > 30:
    heat -= 15
    spray_count += 1
    print(f"Spraying water... heat is now {heat}.")

print(f"Fire contained after {spray_count} spray cycles.")
````

## `while` loops: keep spraying until the fire is contained

A `while` loop keeps running as long as a condition stays true. This is perfect for emergency response when the crew keeps working until the fire is under control.

```python
heat = 85
spray_count = 0

while heat > 30:
    heat -= 15
    spray_count += 1
    print(f"Spraying water... heat is now {heat}.")

print(f"Fire contained after {spray_count} spray cycles.")
```

The key is to make the condition eventually become false. If it never does, the loop runs forever.

````python
# CODE_RUNNER: RSFSC should skip safe zones and stop once the severe fire is found.
fire_levels = [7, 4, 2, 9, 1]

for level in fire_levels:
    if level <= 2:
        print(f"Zone is stable: {level}")
        continue
    print(f"Crew responds to level {level}")
    if level >= 8:
        print("Severe fire detected — focus on the main fire line.")
        break
````

## `break` and `continue` in an emergency

- `break` stops the whole loop immediately.
- `continue` skips the current item and keeps checking the rest.

RSFSC uses this logic when a fire is already contained or when a zone is safe.

```python
fire_levels = [7, 4, 2, 9, 1]
for level in fire_levels:
    if level <= 2:
        print(f"Zone is stable: {level}")
        continue
    print(f"Crew responds to level {level}")
    if level >= 8:
        print("Severe fire detected — focus on the main fire line.")
        break
```

This pattern is useful when the crew should skip harmless zones and stop once the worst fire is addressed.

## Quick challenge

Try to predict the output before you run the code.

A fire crew checks several zones. If the reading is 3 or higher, the zone is still active. If it is below 3, the zone is stable.

Use the loop below to count active fire zones.

````python
# CODE_RUNNER: RSFSC needs one final count. Count the fire zones that are still active and print the total.
alerts = [3, 2, 5, 1, 8, 4, 0]
active_zones = 0

for alert in alerts:
    if alert >= 3:
        active_zones += 1
        print(f"Alert at {alert}: fire still active")
    else:
        print(f"Alert at {alert}: zone is stable")

print(f"Total active fire zones: {active_zones}")
````

### Tech Talk 1: `for` loops

A `for` loop repeats once for each item in a sequence. It is ideal when you know the data you are looping over or when you want to process every item in a list.

```python
for score in [70, 80, 90]:
    print(score)
```

### Tech Talk 2: `while` loops

A `while` loop continues as long as a condition stays true. It is useful when you do not know ahead of time how many times the loop should run.

```python
count = 1
while count <= 5:
    print(count)
    count += 1
```

### Tech Talk 3: `break` and `continue`

- `break` ends the loop immediately.
- `continue` skips the current item and goes to the next one.

```python
for number in range(1, 8):
    if number == 3:
        continue
    if number == 6:
        break
    print(number)
```
