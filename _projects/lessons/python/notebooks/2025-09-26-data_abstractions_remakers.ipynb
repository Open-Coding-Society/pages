### Tech Talk · 3 minutes

**Data abstraction** lets a name and operations stand for a collection, without requiring code to manage its storage details. In [3.1]({{ '/python/variables' | relative_url }}), each level had a separate variable. Here, `level_stars = [10, 20, 15]` represents all levels' star counts.

A list is ordered, with an **element** at each **index**. Values may repeat and types may mix, such as `["Forest", 10]`; this counter keeps every element numeric so addition makes sense. Lists are also called arrays in some languages.

**From College Board:** Topic 3.2 connects list/string representations with explaining how abstraction reduces program complexity ([AAP-1.C/D, pp. 65–66](https://apcentral.collegeboard.org/media/pdf/ap-computer-science-principles-course-and-exam-description.pdf#page=72)). Merely storing a list is insufficient evidence: explain which repeated edits it eliminates.

## 3. Reference Guide

| Operation | Python | College Board pseudocode |
| --- | --- | --- |
| Create a collection | `level_stars = [10, 20, 15]` | `level_stars ← [10, 20, 15]` |
| Empty project | `level_stars = []` | `level_stars ← []` |
| Read the second level | `level_stars[1]` | `level_stars[2]` |
| Replace its count | `level_stars[1] = 25` | `level_stars[2] ← 25` |
| Count levels | `len(level_stars)` | `LENGTH(level_stars)` |
| Add at the end | `level_stars.append(12)` | `APPEND(level_stars, 12)` |
| Insert at position two | `level_stars.insert(1, 8)` | `INSERT(level_stars, 2, 8)` |
| Remove position two | `level_stars.pop(1)` | `REMOVE(level_stars, 2)` |

**Index map · Star Trail's sample levels**

| Level label | Forest | Cavern | Summit |
| --- | --- | --- | --- |
| Stars | 10 | 20 | 15 |
| Python index | 0 | 1 | 2 |
| AP index | 1 | 2 | 3 |

For length `n`, nonnegative Python indices are `0` through `n−1`; AP indices are `1` through `n`. Empty lists have no readable element. An invalid AP index terminates the program; Python raises `IndexError` for an out-of-range index ([AP list rules](https://apcentral.collegeboard.org/media/pdf/ap-computer-science-principles-exam-reference-sheet.pdf#page=4)).

A **string** is an ordered character sequence: Python `"Forest"[0]` is `"F"`. Strings cannot be edited character by character; lists can be changed. **Copying:** Python `other = level_stars` shares the same list; `level_stars.copy()` makes a separate shallow copy, sufficient for these numbers. AP `other ← level_stars` copies the list ([Python lists](https://docs.python.org/3/tutorial/datastructures.html), [AP assignment](https://apcentral.collegeboard.org/media/pdf/ap-computer-science-principles-exam-reference-sheet.pdf#page=4)).

### A. Simple · Inspect the level sequence

````python
level_stars = [10, 20, 15]
level_name = "Forest"
print(level_stars[0])
print(level_stars[1])
print(len(level_stars))
print(level_name[0])
````

**Output:** `10`, `20`, `3`, `F`. Python index 1 is the second level; length counts all levels.

### B. Intermediate · Edit and add a level

````python
level_stars = [10, 20, 15]
level_stars[1] = level_stars[1] + 5
level_stars.append(12)
print(level_stars)
print(len(level_stars))
````

**Output:** `[10, 25, 15, 12]`, then `4`. Replacement keeps the length; appending increases it. AP updates the same level using index `2`.

### C. Complex · Total the whole game

The supplied loop visits each level's star count. Test four levels, then an empty `[]` project.

````python
# CODE_RUNNER: UESL 3.2 - Total the level collection
level_stars = [10, 20, 15]
total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars
print(len(level_stars))
print(total_stars)
````

**Both versions output:** `3`, then `45`. Read these as **level count → total stars**. With `[10, 20, 15, 12]`, expect `4`, `57`; with `[]`, expect `0`, `0`.

<div class="ocs__callout"><strong>Maker checkpoint:</strong> Adding a level changes the collection. The traversal stays the same. A three-element hardcoded sum would still need editing. Reset <code>total_stars</code> before every calculation.</div>

<template data-uesl-pseudocode>
level_stars ← [10, 20, 15] // Store one star count per level.
total_stars ← 0 // Start a fresh total for this playtest.
FOR EACH stars IN level_stars // Visit the levels' counts in order.
{ // Begin the instructions for each visit.
    total_stars ← total_stars + stars // Add this level's stars.
} // Finish the visit, then continue through the remaining levels.
DISPLAY(LENGTH(level_stars)) // Display the number of levels: 3.
DISPLAY(total_stars) // Display the combined star count: 45.
</template>
<template data-uesl-pseudocode>
level_stars ← [6, 9, 12] // Create three levels' star counts.
level_stars[2] ← level_stars[2] + 3 // Add three stars to the second level; AP starts at 1.
APPEND(level_stars, 5) // Add a fourth level with five stars.
DISPLAY(level_stars) // Show all counts: [6, 12, 12, 5].
DISPLAY(LENGTH(level_stars)) // Show the new level count: 4.
</template>

<script>
// Keep this lesson's two source choices inside the existing OCS runners.
document.addEventListener('DOMContentLoaded', () => {
  const examples = document.querySelectorAll('template[data-uesl-pseudocode]');
  document.querySelectorAll('.code-runner-container').forEach((box, index) => {
    const editor = box.querySelector('.CodeMirror').CodeMirror;
    const select = box.querySelector('.languageSelect');
    const starters = {
      python: JSON.parse(box.dataset.defaultCode).trim(),
      pseudocode: examples[index].content.textContent.trim()
    };
    const languageKey = box.dataset.storageKey + '_uesl_language';
    const savedLanguage = localStorage.getItem(languageKey);
    let current = savedLanguage === 'pseudocode' ? 'pseudocode' : 'python';
    const drafts = { ...starters };
    const savedCode = localStorage.getItem(box.dataset.storageKey);
    if (savedLanguage && savedCode !== null) drafts[current] = savedCode;
    select.replaceChildren(new Option('Python', 'python'), new Option('Pseudocode', 'pseudocode'));
    select.setAttribute('aria-label', 'Example language');
    const show = () => {
      select.value = current;
      editor.setValue(drafts[current]);
      editor.setOption('mode', current);
      editor.setOption('lineWrapping', true);
    };
    show();
    select.addEventListener('change', () => {
      drafts[current] = editor.getValue();
      current = select.value;
      show();
      box.querySelector('.output-content').textContent = 'Click Run to playtest this version.';
      box.querySelector('.execTime').textContent = '';
    });
    box.querySelector('[data-hook="save"]').addEventListener('click', () => localStorage.setItem(languageKey, current));
    box.querySelector('[data-hook="clear"]').addEventListener('click', () => {
      localStorage.removeItem(languageKey);
      Object.assign(drafts, starters);
      current = 'python';
      show();
    });
  });
});
</script>
