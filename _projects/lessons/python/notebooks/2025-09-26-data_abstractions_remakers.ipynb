### Tech Talk (3 minutes)

- A **list** (JavaScript: array) is an ordered collection of **elements**, each at a distinct **index**. Values may repeat or have different types, such as `["Nova", 10]`; keep this tracker’s scores numeric for addition.
- A **string** is an ordered character sequence: `"Nova"[0]` is `"N"`. Python/JavaScript strings cannot be changed element by element.
- For length `n`, usual Python/JavaScript indices run **0 through n−1**; AP uses **1 through n**. An empty list has no readable element. AP stops with an error outside its bounds; reading past the end gives Python `IndexError` or JavaScript `undefined`.

| Operation | Python | JavaScript | AP pseudocode |
| --- | --- | --- | --- |
| Create | `scores = [10, 20]` | `const scores = [10, 20];` | `scores ← [10, 20]` |
| Empty | `scores = []` | `const scores = [];` | `scores ← []` |
| First element | `scores[0]` | `scores[0]` | `scores[1]` |
| Element count | `len(scores)` | `scores.length` | `LENGTH(scores)` |
| Add at the end | `scores.append(12)` | `scores.push(12)` | `APPEND(scores, 12)` |

### Code Examples

#### A. Simple: Read an element

````python
scores = [10, 20, 15]
player_name = "Nova"
print(scores[0])
print(scores[1])
print(len(scores))
print(player_name[0])
````

**Output:** `10`, `20`, `3`, `N`. Index 1 selects the second score; length counts all scores.

#### B. Intermediate: Update and append

````python
scores = [10, 20, 15]
scores[1] = scores[1] + 5
scores.append(12)
print(scores)
print(len(scores))
````

**Output:** `[10, 25, 15, 12]`, then `4`. Replacing an element keeps the length; appending increases it. AP updates the second score with `scores[2] ← scores[2] + 5`.

#### C. Complex: Total any number of rounds

The supplied loop means “add each score to the total.” Change the list to four scores, then `[]`, and run again.

````python
# CODE_RUNNER: 3.2C - Total every score in the collection
scores = [10, 20, 15]
total = 0
for score in scores:
    total = total + score
print("Rounds:", len(scores))
print("Total:", total)
````

**Output:** `Rounds: 3`, `Total: 45`. With `[]`, both are 0. Reset `total` before each calculation. The list represents the data; the loop processes it without naming every position.

**JavaScript comparison:** A `const` array can change its contents; the binding cannot be reassigned to another array.

````python
%%js
// CODE_RUNNER: 3.2C JS - Total every score in the collection
const scores = [10, 20, 15];
let total = 0;
for (const score of scores) {
    total = total + score;
}
console.log("Rounds:", scores.length);
console.log("Total:", total);
````

**Copy versus sharing:** `other = scores` shares the Python/JavaScript collection, so element changes affect both names. For an independent flat copy, use Python `scores.copy()` or JavaScript `[...scores]`. AP `other ← scores` copies the list.
