### Cycle 1 Tech Talk: What is a Library? (2 minutes)

A **library** is a collection of procedures that someone already wrote and tested. In Python, one file of reusable code is called a **module**, and a group of modules is a **package**.

Python comes with a **standard library**, so modules like `math`, `random`, `datetime`, `statistics`, and `json` are ready to use with no installing.

**Three ways to import:**

~~~python
import math                     # use as math.sqrt(16)
from random import randint      # use as randint(1, 6)
import statistics as stats      # use as stats.mean([1, 2, 3])
~~~

**How do you know what a procedure does?** You read its **API**, which tells you the procedure's name, its parameters, and what it returns. You don't need to know how `random.randint` works inside. You only need to know it returns a random integer between `a` and `b`, including both. This is **procedural abstraction**.

* ✅ **Do this:** `average = statistics.mean(scores)`
* ❌ **Don't do this:** write your own loop to add and divide when a tested procedure already exists

**Where is the documentation?** Look at [docs.python.org](https://docs.python.org/3/library/), or ask Python directly with `help(math.sqrt)` or `print(math.sqrt.__doc__)`.

### Cycle 3 Tech Talk: pip and requirements.txt (2 minutes)

Standard library modules come with Python. **External libraries** like `Flask` and `pandas` do not. You install them with **pip**, Python's package installer:

~~~bash
pip install flask
~~~

A real project needs many libraries, so projects list them in a file called **`requirements.txt`**. Here is part of the OCS Flask backend's file:

~~~text
Flask
Flask_Login
SQLAlchemy
PyJWT
pandas
scikit-learn
python_dotenv
~~~

When you clone a project, one command installs everything, usually inside a virtual environment (`venv`) so each project keeps its own packages:

~~~bash
python3 -m venv venv
source venv/bin/activate
pip install -r requirements.txt
~~~

**Why it matters:**

* **Teammates** get the same libraries you have, so the project runs on every computer.
* **Deployment** servers, such as the OCS Docker backend, read the same file.
* **Versions** can be set with `Flask==3.0.0` (exactly this version) or `Flask>=3.0` (this version or newer).

**Watch out:** the name you `pip install` is not always the name you `import`:

| pip install | import |
| --- | --- |
| `python-dotenv` | `import dotenv` |
| `scikit-learn` | `import sklearn` |
| `PyJWT` | `import jwt` |

If a library is missing, Python shows `ModuleNotFoundError`. That usually means you forgot `pip install -r requirements.txt`, or your `venv` is not active.

### Popcorn Hack 3: Check a requirements.txt (5 minutes)

**Task:** Finish `parse_requirements` so it skips blank lines and comments and removes version numbers. The program then uses `importlib` to check which packages are installed on the code runner.

```python
# CODE_RUNNER: Popcorn Hack 3 - Finish the requirements.txt checker using importlib

import importlib.util

requirements_txt = '''
# Web framework
Flask>=3.0
requests

# Data and machine learning
pandas==2.2.2
scikit-learn
python-dotenv
'''

# The name you pip install is not always the name you import
IMPORT_NAMES = {
    "Flask": "flask",
    "scikit-learn": "sklearn",
    "python-dotenv": "dotenv",
}

def parse_requirements(text):
    '''Return a list of package names from requirements.txt text.'''
    packages = []
    for line in text.splitlines():
        line = line.strip()
        # TODO 1: Skip blank lines and lines that start with "#"

        # TODO 2: Remove version specifiers such as "==2.2.2" or ">=3.0"
        name = line
        packages.append(name)
    return packages

def is_installed(package):
    '''Return True if Python can find the package on this computer.'''
    import_name = IMPORT_NAMES.get(package, package)
    try:
        return importlib.util.find_spec(import_name) is not None
    except (ValueError, ModuleNotFoundError):
        return False

for package in parse_requirements(requirements_txt):
    if is_installed(package):
        print(f"{package!r:20} installed")
    else:
        print(f"{package!r:20} missing -> pip install {package}")

# TODO 3: Print is_installed("json"). Why is json installed even though it is not in requirements.txt?
```
