### Tech Talk 1: Random SFI Record IDs

A backend QA script can randomly choose which SFI automotive record to test. `random.randint(a, b)` selects an integer from the full inclusive range.

```python
# CODE_RUNNER: Tech Talk 1 - Random SFI Record ID

import random

sfi_record_id = random.randint(1, 4)
print("SFI record selected for QA:", sfi_record_id)
```

`random.randint(a, b)` can return any whole number from `a` through `b`, including both endpoints.

For `random.randint(1, 4)`, the possible SFI test record IDs are `1`, `2`, `3`, and `4`.

### Tech Talk 2: Random SFI Part Records

Instead of choosing only an ID, Python can randomly select a complete automotive part record from a list.

```python
# CODE_RUNNER: Tech Talk 2 - Random SFI Part

import random

sfi_parts = [
    "Replacement Flywheels and Clutch Assemblies",
    "Multiple Disc Clutch Assemblies",
    "Racing Flywheel Record"
]

selected_part = random.choice(sfi_parts)
print("SFI part selected for testing:", selected_part)
```

The result must be one item from the SFI part list, but the same part can be selected again on the next run.

The same random-record-ID idea looks like this in the three forms:

| Form | Random SFI record ID from 1 through 4 |
| --- | --- |
| Python | `random.randint(1, 4)` |
| JavaScript | `Math.floor(Math.random() * 4) + 1` |
| College Board pseudocode | `RANDOM(1, 4)` |

### Tech Talk 3: Use Random Values in Backend Decisions

A QA script can randomly choose a backend action and then use that result in an `if` statement.

```python
# CODE_RUNNER: Tech Talk 3 - Random SFI Backend Action

import random

actions = ["GET/search", "POST/create"]

selected_action = random.choice(actions)

if selected_action == "GET/search":
    print("QA test: search for an SFI part record")
else:
    print("QA test: validate an SFI part before creating it")
```

Random does **not** mean every possible result appears once before a repeat. Repeats are normal.
