# SASS Inputs — PAW Patrol

## 1. LxD Cycle Process

### Empathize
I noticed a lot of students styling inputs by hand — writing inline `style="border: 2px solid pink; padding: 10px;"` or inventing one-off classes like `class="pinkbox"` instead of reaching for `smallInput`, `mediumInput`, `largeInput`, or `gradientInput`.

It works visually, but it breaks the moment the site theme changes, and every form ends up looking different.

### Define

**POV:** CSP students need a shared, class-based way to size and style inputs because hand-written inline styles create inconsistent forms and make the site's design system impossible to maintain.

**Learning Goal:** Students will understand how to size and style `<input>` elements using PAW Patrol's size classes (`smallInput`, `mediumInput`, `largeInput`) and style class (`gradientInput`) instead of inline styles or custom one-off classes.

### Ideate

**HMW Question:** How might we teach students to reach for the existing input classes before writing any custom CSS?

**Activity:** Refactor a form built with inline styles and made-up classes into clean, class-based PAW Patrol inputs.

### Prototype & Test

I ran this with my project team first. They found the original draft asked them to build an entire multi-field form from scratch, which ran long and lost momentum. I cut it down to a single quick refactor task, documented below.

## 2. Lesson Plan

**Learning Objective:** By the end of this lesson, you will be able to size and style `<input>` elements using PAW Patrol's size and style classes instead of inline styles or custom classes.

**Success Criteria:** You can take an input built with inline styles or made-up classes and rewrite it using the correct `smallInput` / `mediumInput` / `largeInput` and `gradientInput` classes — matching the site's design system exactly.

### Tech Talk (3 minutes)

We are using a global PAW Patrol input system. You don't need to write CSS to size or style an input — you tell the browser what role the input plays, and the class system handles the look.

**The Rule:** Use the built-in size and style classes. Let the system handle the look.

- ✅ Do this: `<input type="text" class="mediumInput" placeholder="Email">`
- ❌ Don't do this: `<input type="text" style="width: 300px; padding: 8px;" placeholder="Email">`

Why do we do this? It keeps every form on the site consistent, keeps your HTML clean, and means a single theme change updates every input at once.

### Code Examples

#### A. Simple: Default Input

````python
%%html
<!-- No special styling needed — clean and minimal. -->
<input type="text" placeholder="default input">
````

#### B. Intermediate: Size Classes

````python
%%html
<!-- Don't eyeball a width with inline styles. Use a size class. -->
<input type="text" class="smallInput" placeholder="First Name">
<input type="text" class="largeInput" placeholder="Search...">
````

#### C. Complex: Combo Classes

````python
%%html
<!-- Combine size and style classes for a stylized, sized input. -->
<input type="text" class="largeInput gradientInput" placeholder="Username">
````
