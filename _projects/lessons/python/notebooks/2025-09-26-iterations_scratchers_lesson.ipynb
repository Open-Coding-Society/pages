### **Part 1 — The Basics: What is a `for` Loop?**

A `for` loop repeats a block of code for each item in a collection. It is useful when you already know exactly what you want to check.

Python uses `for` loops all the time when processing lists, scores, names, readings, and other repeated data.

```python
for value in [10, 20, 30]:
    print(value)
```

The loop variable changes each time, so one small block of code can inspect many values without repeating itself.

### **Part 2 — The Basics: What is a `while` Loop?**

A `while` loop keeps running as long as a condition is true. It is useful when you do not know how many times the loop needs to repeat ahead of time.

```python
count = 1
while count <= 5:
    print(count)
    count += 1
```

The key idea is that something inside the loop must eventually change the condition or the loop could run forever.

### **Popcorn Hack 2: Try a `while` Loop**

Run the code below and notice how the loop changes the temperature each pass.

````python
# CODE_RUNNER: Popcorn Hack 2 - While loop practice
temperature = 85
cooldown_count = 0

while temperature > 30:
    temperature -= 15
    cooldown_count += 1
    print(f"Cooling down... temperature is now {temperature}.")

print(f"Stabilized after {cooldown_count} cooling cycles.")
````

### **Part 3 — `break` and `continue`**

`continue` skips the rest of the current loop pass and moves to the next item.
`break` stops the loop completely.

This is useful when a program should ignore harmless values and stop once it finds the important one.

```python
scores = [10, 7, 18, 5, 25, 9, 30]
cutoff = 10

for score in scores:
    if score < cutoff:
        continue
    print(score)
    if score >= 25:
        break
```
