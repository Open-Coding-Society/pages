# 3.08 Iterations

Iteration means repeating a set of steps. In Python, loops help programs process lists, keep checking a condition, skip values that do not matter, and stop when the important case has been found.

This lesson uses short CodeRunner examples so you can run each loop, read the output, and connect the result back to the control flow.

## The Big Idea

Loops let a program repeat work without copying the same instructions over and over.

You will practice four core ideas:

- `for` loops repeat through a known sequence, such as a list of readings.
- `while` loops repeat until a condition changes.
- `continue` skips the current item and keeps checking the rest.
- `break` stops a loop early when the program has found what it needs.

## Scenario

The examples use a simple emergency-response theme. A program checks sensor readings, cooling temperatures, and fire-zone alerts.

Run each example, then explain which loop decision caused each line of output.

```python
# CODE_RUNNER: Use a for loop to inspect a list of sensor readings.
readings = [34, 42, 31, 48, 29, 37]

for reading in readings:
    print(f"Checking reading: {reading}")
    if reading >= 40:
        print("High alert: investigate this reading.")
    else:
        print("Reading is within the normal range.")
```

## `for` Loops: Inspect Each Item

A `for` loop is best when you already have a collection to examine. The loop runs once for each item, using the same logic every time.

In the runner above, `reading` changes on every pass through the loop. One block of code can inspect many values.

```python
# CODE_RUNNER: Keep checking until the temperature drops below 30.
temperature = 85
cooldown_count = 0

while temperature > 30:
    temperature -= 15
    cooldown_count += 1
    print(f"Cooling down... temperature is now {temperature}.")

print(f"Stabilized after {cooldown_count} cooling cycles.")
```

## `while` Loops: Repeat Until Safe

A `while` loop keeps running as long as its condition is true. This is useful when you do not know exactly how many repetitions will be needed.

The important rule: make sure something inside the loop eventually changes the condition. Otherwise, the loop can run forever.

```python
# CODE_RUNNER: SRFSC should skip safe zones and stop once the severe fire is found.
fire_levels = [7, 4, 2, 9, 1]

for level in fire_levels:
    if level <= 2:
        print(f"Zone is stable: {level}")
        continue
    print(f"Crew responds to level {level}")
    if level >= 8:
        print("Severe fire detected - focus on the main fire line.")
        break
```

## `break` and `continue`

Loop-control statements help a program make better decisions while it repeats.

- `continue` skips the rest of the current loop pass and moves to the next item.
- `break` exits the loop completely.

In the runner above, safe zones are skipped, while a severe fire stops the scan and becomes the top priority.

## Quick Challenge

Before you run the next example, predict the output.

A fire crew checks several alert readings. If a reading is `3` or higher, the zone is still active. If it is below `3`, the zone is stable.

Use the loop below to count how many fire zones are still active.

```python
# CODE_RUNNER: SRFSC needs one final count. Count the fire zones that are still active and print the total.
alerts = [3, 2, 5, 1, 8, 4, 0]
active_zones = 0

for alert in alerts:
    if alert >= 3:
        active_zones += 1
        print(f"Alert at {alert}: fire still active")
    else:
        print(f"Alert at {alert}: zone is stable")

print(f"Total active fire zones: {active_zones}")
```
