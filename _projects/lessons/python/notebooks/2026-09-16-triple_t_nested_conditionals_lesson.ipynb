# 3.7 Nested Conditionals

## Learning Objective

By the end of this lesson you can write a nested conditional that handles **every** combination of its conditions, and prove that it does by counting leaves.

## Success Criteria

- Given two conditions, you can state how many outcomes exist *before* writing any code.
- Your conditional produces output for every combination of inputs, not just the one you were picturing.
- You can point at any input and name which branch it lands in.

> **One rule for the whole lesson:** an `if` with no `else` is a question with a missing answer, and Python will never tell you it's missing.

**One scenario for the whole lesson.** Every example, hack, and translation on this page is the same program: **a login screen.** Someone types an email and a password, and the program decides what happens next. We start with one condition and add one at a time, so you can watch the decision tree grow instead of learning five unrelated programs.


## 1. Reference Guide

Everything on this page in one place. Come back here during the hacks.

### The rule

| Term | Meaning |
|---|---|
| **Outcome** | One combination of the conditions. `n` conditions produce **2<sup>n</sup>** outcomes. |
| **Leaf** | A branch that does something and asks nothing further. |
| **The check** | Leaves should equal outcomes. If they don't, a case is missing. |
| **Guard** | A condition checked first because it makes the others meaningless. Legal to have fewer leaves **only if you can name which combinations it absorbs.** |

| Conditions | Outcomes | Leaves you need |
|---:|---:|---|
| 1 | 2 | 2 |
| 2 | 4 | 4 |
| 3 | 8 | 8, or fewer with a justified guard |

### The same nested conditional in three languages

| | Python | JavaScript | College Board pseudocode |
|---|---|---|---|
| Assign | `x = True` | `let x = true;` | `x <- true` |
| Output | `print(...)` | `console.log(...)` | `DISPLAY(...)` |
| Grouping | indentation | `{ }` | `{ }` |
| Else-if | `elif` | `else if` | `ELSE IF` |
| And / or / not | `and` `or` `not` | `&&` `\|\|` `!` | `AND` `OR` `NOT` |
| Booleans | `True` `False` | `true` `false` | `true` `false` |
| Nesting shown by | indentation | `{ }` | `{ }` |

### Mistakes this lesson is about

| Mistake | What it looks like | Fix |
|---|---|---|
| Missing branch | Program prints nothing, raises nothing, exits 0 | Count leaves against 2<sup>n</sup> |
| `and` instead of nesting | Can't tell "only a" from "only b" from "neither" | Nest when the failures need different answers |
| Guard in the wrong place | Asking about something that doesn't exist yet | Check the thing that makes others meaningless **first** |
| `else` at the wrong indent | Attaches to a different `if`, silently changing the logic | Line each `else` up with its own `if` |

## 4. Lesson Plan

### Tech Talk (3 minutes)

A nested conditional is a conditional inside a conditional. The outer answer decides which question you ask next.

In Python the nesting **is** the indentation. You can read the depth without reading a single condition:

    if outer:
        if inner:        <- two levels in
            ...
        else:
            ...
    else:
        ...

**The rule:** `n` conditions produce 2<sup>n</sup> possible combinations of inputs.

| Conditions | Possible outcomes |
|---:|---:|
| 1 | 2 |
| 2 | 4 |
| 3 | 8 |

A **leaf** is a branch with nothing conditional inside it — a place the program stops and does something. You need one leaf per outcome, or a stated reason you have fewer.

**Why counting matters here.** In most languages a missing case is just wrong. In Python it's wrong *and silent*: `if x:` with no `else`, and `x` false, prints nothing, raises nothing, exits 0.

**The thing that trips people up.** `and` collapses four outcomes into two. `if a and b:` gives you "both true" and "everything else" — it can't tell "only `a`" from "only `b`" from "neither." When those need different answers, you nest.


## 5. Code Examples


### Example A — One condition, and the branch that isn't there

The first version of the login screen. If the password is right, let them in.

Run it before reading on.


```python
# CODE_RUNNER: 3.7A - One condition: run it, what prints?

password_correct = False

if password_correct:
    print("Welcome back!")
```

**No output. No error.** Skimming past it, you'd assume it worked.

One condition means 2 possible inputs. Count the leaves: there is **1**. The other case falls off the end of the program and disappears — so someone types the wrong password and the screen just sits there. You've hit this on a real website, and you probably clicked the button three more times wondering if it was broken.

One `else` fixes the silence:


```python
# CODE_RUNNER: 3.7A - One else: now every input gets an answer

password_correct = False

# 1 condition -> 2 outcomes -> 2 leaves
if password_correct:
    print("Welcome back!")
else:
    print("That password doesn't match. Try again.")
```

### Example B — Add a second condition: four leaves

A correct password isn't automatically enough. If you're signing in from a **device the account has never seen before**, a real login sends you a verification code first.

So add a second condition, `new_device`. Two conditions, and all four combinations need genuinely different actions.


```python
# CODE_RUNNER: 3.7B - Two conditions: change the values until all four leaves print

password_correct = True
new_device = True

print(f"password_correct={password_correct}, new_device={new_device}")
print("The login screen says:")

# 2 conditions -> 4 outcomes -> 4 leaves
if password_correct:
    if new_device:
        # leaf 1
        print("Password accepted, but we don't recognize this device.")
        print("Enter the 6-digit code we just emailed you.")
    else:
        # leaf 2
        print("Welcome back! You're signed in.")
else:
    if new_device:
        # leaf 3
        print("That password doesn't match. Try again.")
        print("We also emailed the account owner about a sign-in attempt from a new device.")
    else:
        # leaf 4
        print("That password doesn't match. Try again.")
```

#### Count the leaves

2 conditions gives 2<sup>2</sup> = **4 outcomes**. The code has **4 leaves**. They match, so coverage is complete.

| `password_correct` | `new_device` | Leaf | What the login screen does |
|---|---|---:|---|
| `True` | `True` | 1 | Right password, unknown device — send a verification code |
| `True` | `False` | 2 | Sign in |
| `False` | `True` | 3 | Reject, and warn the owner about the new device |
| `False` | `False` | 4 | Reject |

Leaf 1 is the row that carries the lesson. The password was **correct** and the answer is still not "let them in." A right password from a device nobody has ever seen is exactly what a stolen password looks like.

Now imagine writing this as `if password_correct and not new_device:` instead. That merges leaves 1, 3 and 4 into one "everything else" — so a correct password from an unknown device gets treated identically to a wrong password. The attacker and the real owner locked out of their own laptop get the same screen.

**Try it:** change the two values at the top and re-run until you've printed all four leaves.


### The same decision in JavaScript

Identical logic to Example B — same four leaves, same order. Only the punctuation changes: braces instead of indentation, `else if` instead of `elif`, `&&` instead of `and`.

Run it and compare the output to the Python version above.


```python
%%js
// CODE_RUNNER: 3.7B JS - The same four leaves in JavaScript

let passwordCorrect = true;
let newDevice = true;

console.log(`passwordCorrect=${passwordCorrect}, newDevice=${newDevice}`);

// 2 conditions -> 4 outcomes -> 4 leaves
if (passwordCorrect) {
    if (newDevice) {
        console.log("Password accepted, but we don't recognize this device.");
    } else {
        console.log("Welcome back! You're signed in.");
    }
} else {
    if (newDevice) {
        console.log("That password doesn't match. We emailed the owner about a new device.");
    } else {
        console.log("That password doesn't match. Try again.");
    }
}
```

### The same decision in College Board pseudocode

This is the form the AP exam uses. `<-` is assignment, `DISPLAY` is output, and braces group a block. The leaves are the same four.


    passwordCorrect <- true
    newDevice <- true

    IF (passwordCorrect)
    {
        IF (newDevice)
        {
            DISPLAY("Password accepted, but we don't recognize this device.")
        }
        ELSE
        {
            DISPLAY("Welcome back! You're signed in.")
        }
    }
    ELSE
    {
        IF (newDevice)
        {
            DISPLAY("That password doesn't match. We emailed the owner.")
        }
        ELSE
        {
            DISPLAY("That password doesn't match. Try again.")
        }
    }

Trace it the same way: outer `IF` first, inner `IF` second, four leaves. Every Python runner on this page also has a **Pseudocode** tab if you want to try rewriting one yourself.

**What to notice across all three.** Python uses indentation, JavaScript uses braces, pseudocode uses braces and `DISPLAY`. None of that changes the decision. Counting outcomes and counting leaves works identically in every one — which is why the exam can ask about nested conditionals without caring what language you write.


### Example C — Add a third condition, and why five leaves is enough

One more real requirement. Before any of this, does an account with that email even exist? If it doesn't, asking whether the password matches is meaningless — there's nothing to match against.

Three conditions means 2<sup>3</sup> = **8 combinations**. The code below has only **5 leaves**, and that's correct, not a bug.


```python
# CODE_RUNNER: 3.7C - Three conditions: a guard absorbs four combinations

account_exists = True
password_correct = True
new_device = False

# 3 conditions -> 8 outcomes -> 5 leaves (the guard absorbs 4; explained below)
if not account_exists:
    # leaf 1 - with no account, the other two describe nothing
    print("If that email has an account, we've sent it a sign-in link.")
else:
    if password_correct:
        if new_device:
            print("Enter the 6-digit code we just emailed you.")        # leaf 2
        else:
            print("Welcome back! You're signed in.")                    # leaf 3
    else:
        if new_device:
            print("Wrong password. We emailed the owner about it.")     # leaf 4
        else:
            print("That password doesn't match. Try again.")            # leaf 5
```

#### Why 5 leaves covers 8 combinations

| Leaf | Combinations covered | Why they collapse |
|---|---:|---|
| 1 — no account | 4 | With no account, `password_correct` and `new_device` describe nothing that exists |
| 2 — right password, new device | 1 | |
| 3 — right password, known device | 1 | |
| 4 — wrong password, new device | 1 | |
| 5 — wrong password, known device | 1 | |
| **Total** | **8** | |

Fewer leaves than 2<sup>n</sup> is fine **when you can name which combinations collapsed and why**. That sentence is the whole difference between a guard and a missing branch. If you can't name them, you didn't collapse a case — you forgot one.

Checking first the thing that makes the other checks meaningless is called a **guard**. Order matters: move the guard below the password check and you're asking whether an account that doesn't exist typed the right password.

> **Why leaf 1 is worded so vaguely.** Notice it doesn't say "no account with that email." Real login screens avoid that on purpose — if the site tells you an email isn't registered, an attacker can feed it a list of addresses and learn exactly which ones have accounts. "If that email has an account, we've sent it a link" answers the user without answering the attacker.
