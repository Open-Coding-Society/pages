### Lesson (Tech Talk)

When you order a drink at a café, the register calculates tax, prints a receipt, and sends your order to the barista. You don’t compute sales tax or printer codes by hand — the register just runs a program. That’s calling a procedure: the steps are saved, and you run them whenever you need them.

#### What is calling a procedure?
A procedure (a function in Python) is a named set of re-usable code instructions.

- A procedure (a function in Python) is a named set of re-usable code instructions.
- Reusability: You can call a procedure multiple times with different drink names or prices without rewriting the code.
- Definition vs. Calling: Defining creates the procedure (def); calling executes it with parentheses ().
- Execution: Calling allows you to run the same code instructions whenever you need them without retyping them.

#### Passing Inputs (Parameters)
You can make procedures more flexible by giving them parameters—input variables that hold data passed into the procedure when you call it.

- Parameters act as temporary placeholders inside the procedure definition.
- Arguments are the actual values passed into the procedure when you call it.

This allows a single procedure to run the same set of instructions on completely different data without rewriting the code.

#### Calling Procedures Inside Procedures (Decomposition)
In computer science, large problems are broken down into smaller steps—a process called decomposition.
- An outer procedure handles the main task (like printing a full customer receipt).
- A helper procedure handles a specific sub-task (like calculating tax or applying discounts) and returns a result to the outer procedure.

Each step can become its own procedure, and one procedure can call another one. This allows programs to become easier to read, use, and debug.

### Key Vocabulary & Concept Definitions

| Term | Definition | Example |
| :--- | :--- | :--- |
| **Procedure (Function)** | A named block of reusable code instructions. | `def print_receipt():` |
| **Parameter** | Variable listed in the procedure definition that receives data. | `def make_drink(size):` |
| **Argument** | Actual value sent into the procedure during a call. | `make_drink("Large")` |
| **Decomposition** | Breaking a complex program into smaller sub-tasks. | Creating `add_tax()` and `print_receipt()` |
| **Helper Procedure** | A procedure called inside another procedure to complete a specific sub-task. | `add_tax()` inside `print_receipt()` |

#### B. Complex: Decomposition with Helper Procedures
Demonstrates breaking down a inventory check task into a helper procedure (check_stock) and an outer procedure (place_supply_order).

```python
# CODE_RUNNER: Complex Decomposition Example
# Helper Procedure: Checks if current inventory is below reorder threshold
def check_stock(current_amount, minimum_required):
    return current_amount < minimum_required

# Outer Procedure: Decides whether to place a reorder
def place_supply_order(ingredient_name, current_stock, threshold):
    needs_reorder = check_stock(current_stock, threshold)
    
    if needs_reorder:
        print("ALERT: Low on " + ingredient_name + ". Placing reorder now!")
    else:
        print("STOCK OK: Sufficient " + ingredient_name + " available.")

# Test calls
place_supply_order("Oat Milk (Cartons)", 3, 10)
place_supply_order("Coffee Beans (Bags)", 15, 5)
```

#### Popcorn Hack 2: Decomposition & Helper Procedures
**Concept:** Break down a complex problem into smaller helper procedures that can be called inside an outer main procedure.

**Task:**

2. Complete checkout(item_name, price, discount) to call apply_discount inside it and print the final receipt message.

```python
# CODE_RUNNER: Pseudocode Template (COMPLETED EXAMPLE) of Popcorn Hack 2

PROCEDURE apply_discount(price, discount)
{
    RETURN price - discount
}

PROCEDURE add_tax(price)
{
    RETURN price * 1.08
}

PROCEDURE checkout(item_name, original_price, discount_amount)
{
    discounted_price <- apply_discount(original_price, discount_amount)
    final_total <- add_tax(discounted_price)
    DISPLAY(item_name)
    DISPLAY(" | Final Total: $")
    DISPLAY(final_total)
}

checkout("Espresso", 4.00, 0.50)
checkout("Muffin", 3.50, 1.00)
```

```python
# CODE_RUNNER: Popcorn Hack 2 - Decomposition & Helper Procedures

# Helper Procedure: Calculates discounted price
def apply_discount(price, discount):
    # TODO: Return the price minus discount
    return price - discount

# Helper Procedure: Adds 8% sales tax
def add_tax(price):
    return round(price * 1.08, 2)

# Outer Procedure: Handles checkout flow using helpers
def checkout(item_name, original_price, discount_amount):
    # TODO: Call apply_discount here
    discounted_price = apply_discount(original_price, discount_amount)
    
    # Call add_tax helper
    final_total = add_tax(discounted_price)
    
    print("Item: " + item_name + " | Final Total: $" + str(final_total))

# Test calls
checkout("Espresso", 4.00, 0.50)
checkout("Muffin", 3.50, 1.00)
```
