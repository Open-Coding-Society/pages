# Authoring skeleton - backend focus (CS112 backend integration)

## Three short topics
1. Request to controller to repository to database: who owns what.
2. Entity, DTO, and JSON contract (field names, types, required values).
3. Success and error responses a client can rely on.

## PBL context
Connect a browser interaction to a Spring endpoint and persist or retrieve project data. Observe an existing controller/entity pair, then extend it; do not build the frontend.

## Backend layers to author
- Read an existing controller method and trace one request end to end.
- Add a field or endpoint to an entity and expose it through JSON.
- Persist a record and retrieve it with a repository query.
- Document the API contract (method, path, body, status codes).

## Evidence checklist to author
Call the endpoint with a valid request, a missing field, and an unknown id; record status code and body for each. Show that valid stored data is unchanged after a failed request.

## Publication checklist
Complete examples, a CodeRunner exercise, and tests before removing the draft/hide flags, enabling submissions, or confirming the course and week.
