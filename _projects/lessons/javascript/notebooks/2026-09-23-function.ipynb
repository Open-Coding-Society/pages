<h2 class="glowing-text">What are functions?</h2>

> Functions are reusable blocks of code designed to perform specific tasks. Instead of writing the same pieces of code multiple times, you put it inside a function, which you can call whenever and can choose what goes in it.


Functions allow you to take a piece of code, and almost put it into a box so you can reuse it a lot of times. It also let you modularize code. For example, let's look at the code below.

````python
%%js

// CODE_RUNNER: Check out this LONG piece of code!


function rockPaperScissors() {
   let myChoice = "rock";
   let opponentChoice = "scissors";
   let message = "Congratulations! You beat " + opponentChoice + " using " + myChoice + "!";
   document.getElementById("output1").innerText += message;

   myChoice = "paper";
   opponentChoice = "paper";
   message = "It's a tie because the opponent chose " + opponentChoice + " and you chose " + myChoice + "!";
   document.getElementById("output1").innerText += "\n" + message;

   myChoice = "scissors";
   opponentChoice = "rock";
   message = "Aw man! You got beat by " + opponentChoice + " when you chose " + myChoice + "!";
   document.getElementById("output1").innerText += "\n" + message;

   myChoice = "rock";
   opponentChoice = "paper";
   message = "Aw man! You got beat by " + opponentChoice + " when you chose " + myChoice + "!";
   document.getElementById("output1").innerText += "\n" + message;
}
````

This code, while it works, has a lot of code that is reused. We can modularize this code as shown below.

````python
%%js
// CODE_RUNNER: Check out this WAY shorter version of the Rock Paper Scissors game!

function playRockPaperScissors(myChoice, opponentChoice){
       let message;
       if (myChoice === opponentChoice) {
           message = "It's a tie because the opponent chose " + opponentChoice + " and you chose " + myChoice + "!";
       } else if (
           (myChoice === "rock" && opponentChoice === "scissors") ||
           (myChoice === "paper" && opponentChoice === "rock") ||
           (myChoice === "scissors" && opponentChoice === "paper")
       ) {
           message = "Congratulations! You beat " + opponentChoice + " using " + myChoice + "!";
       } else {
           message = "Aw man! You got beat by " + opponentChoice + " when you chose " + myChoice + "!";
       }
       document.getElementById("output2").innerText += "\n" + message;
   }


   playRockPaperScissors("rock", "scissors");
   playRockPaperScissors("paper", "paper");
   playRockPaperScissors("scissors", "rock");
   playRockPaperScissors("rock", "paper");
````

<h2 class="glowing-text"><strong>Recursion</strong></h2>

Recursion is where a function continuously calls itself. View the example below for a factorial recursion function:

````python
%%js

//CODE_RUNNER: Check out this recursive function!

function factorial(n) {
    // If n is 0 or 1, the factorial is 1.
    // This condition prevents infinite recursion.
    if (n === 0 || n === 1) {
        return 1;
    }
    // Recursive case: the function calls itself with a smaller value of n.
    else {
        return n * factorial(n - 1);
    }
}
````

For example we run factorial(5), it will do:
- factorial(5) = 5 * factorial(4), which calls:
- factorial(4) = 4 * factorial(3)
- factorial(3) = 3 * factorial(2)
- factorial(2) = 2 * factorial (1)
- factorial(1) = 1, where it stops
- So the final answer will be 5 * (4 * (3 * (2 * (1)))) which is 120
