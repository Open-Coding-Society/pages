## Introduction

**Imagine this**: You're deciding what to do on a Saturday afternoon.

First, you check whether you have tickets for a movie.

**IF** you have movie tickets:
- You go to the theater, but then you need to check one more thing - did you bring snacks?
  - **IF** you brought snacks, you enjoy the movie with popcorn and a drink.
  - **ELSE** (if you have tickets but no snacks), you buy snacks at the theater.

**ELSE** (if you do not have movie tickets):
- You plan an outdoor activity, **BUT** you still need to check the weather.
  - **IF** it is sunny, you go to the park.
  - **ELSE**, you stay home and play a board game.

See how you're making decisions **INSIDE** other decisions? That's a **nested conditional**!


## Nested Conditional Code

Now let's look at an example of nested conditional code using the weekend-plans analogy above.

~~~js
// Decide what to do on Saturday afternoon
if (hasMovieTickets) {
    // First decision: We have tickets, so go to the theater
    goToTheater();

    // Nested decision INSIDE the movie-ticket condition
    if (hasSnacks) {
        console.log("Enjoying the movie with snacks");
    } else {
        buySnacks();
        console.log("Buying snacks at the theater");
    }

} else {
    // First decision: No tickets, so plan an outdoor activity

    // Nested decision INSIDE the no-tickets condition
    if (isSunny) {
        goToPark();
        console.log("Going to the park");
    } else {
        playBoardGame();
        console.log("Staying home to play a board game");
    }
}
~~~

In the code example above, there are 2 nested conditionals: there's one in the if statement, and one in the else statement. If you have movie tickets, the snack check inside the if block will run. If you do not have movie tickets, we skip the if block entirely and go to the else statement, where a weather check will run.
