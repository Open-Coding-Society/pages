## 2. Key Vocab

| Type | Meaning | Example | Notes |
| --- | --- | --- | --- |
| `object literal` | Keeps related information together | `const player = { name: "Alex" };` | Uses `{ }` |
| `property` | One piece of information in an object | `health: 100` | Has a name and value |
| `key` | The name of a property | `health` | Comes before `:` |
| `value` | The information stored in a property | `100` | Comes after `:` |
| `dot notation` | Gets information from an object | `player.health` | Uses a dot |


## 3. Lesson Plan & Introduction

Imagine you are making a game. Your player has a name, health, and coins. You could make a different variable for everything, but that gets confusing when you add more players.

An **object literal** lets you put the player's information in one place.

~~~js
const player = {
    name: "Alex",
    health: 100,
    coins: 20
};

console.log(player.name);   // Alex
console.log(player.health); // 100
~~~

The words `name`, `health`, and `coins` are called **properties**. We use a dot to get their values. We can also change them while the game is running.

~~~js
player.health = player.health - 10;
console.log(player.health); // 90
~~~

This could happen when the player gets hit by an enemy. Object literals are useful because you can keep all the information about a game character together.

### Popcorn Hack 2: Pick Up a Game Item

In games, players can pick up items like coins or potions. An object can store the item's name and what it does.

**Directions:**
1. Change the item name to something you would put in a game.
2. Change how many coins it gives the player.
3. Change the coin calculation so the player gets the coins from the item.
4. Run the code and see how many coins the player has now.


```javascript
%%js
const player = {
    name: "Alex",
    coins: 5
};

const item = {
    name: "Coin Bag",   // Change this item name
    coins: 10          // Change this number
};

console.log("Picked up: " + item.name);

// Add the coins from the item to the player
// Use player.coins and item.coins
player.coins = player.coins + 0;

console.log("Coins now: " + player.coins);
```

## 5. Summary

An object literal stores related values together using `{ }`. Each value has a property name. We can use dot notation like `player.health` to read or change it. In games, this can help us store player health, coins, items, and other information without making lots of separate variables.
