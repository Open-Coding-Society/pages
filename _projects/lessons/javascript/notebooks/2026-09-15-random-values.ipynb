## 1. Reference Guide

### Key Vocabulary

| Term | Definition | Example |
| --- | --- | --- |
| **Random value** | A value selected from a set/range according to a random process. | A game where you are spinning a wheel → random value could be **250 coins** |
| **Inclusive range** | Both endpoints can be selected. | A random integer from 1 through 4 can return 1 or 4 |
| **Possible output** | Any result the expression is allowed to produce. | 1, 2, 3, or 4 |
| **Random choice** | Selects one item from an array. | `parts[Math.floor(Math.random() * parts.length)]` |
| **QA simulation** | Repeated test runs used to exercise different program paths. | random record + random API action |

### JavaScript Random Quick Reference

```javascript
const recordId = Math.floor(Math.random() * 4) + 1;
const part = parts[Math.floor(Math.random() * parts.length)];
```

### Reasoning Rules

- `Math.random()` returns a number from 0 (inclusive) to 1 (exclusive).
- `Math.floor(Math.random() * (max - min + 1)) + min` creates an integer from `min` through `max`, inclusive.
- A value can repeat on consecutive runs.
- Seeing a few outputs does not prove those are the only possible outputs.
- Choosing one array element keeps the chosen object intact.
- Before using a random index/range, verify every generated value is valid.
- For reproducible debugging, use a seeded random-number generator; JavaScript's built-in `Math.random()` cannot be seeded directly.

## 3. College Board Requirements

AP CSP **Topic 3.15: Random Values** asks students to write expressions that generate possible values and evaluate expressions to determine all possible results. The exam reference sheet uses `RANDOM(a, b)`, which returns an integer from `a` to `b`, inclusive.

Connect the pseudocode to JavaScript:

| College Board pseudocode | JavaScript |
| --- | --- |
| `RANDOM(1, 4)` | `Math.floor(Math.random() * 4) + 1` |
| possible outputs: 1,2,3,4 | possible outputs: 1,2,3,4 |
| each execution may differ | repeated JavaScript runs may differ |

JavaScript uses `Math.random()` with array indexes to choose one complete record from structured test data.

### College Board Pseudocode

AP CSP uses `RANDOM(a, b)` in its pseudocode. It returns a random integer from `a` through `b`, **inclusive**.

### Random Values: College Board vs JavaScript

| College Board pseudocode | JavaScript |
| --- | --- |
| `RANDOM(1, 4)` | `Math.floor(Math.random() * 4) + 1` |
| list positions begin at 1 | JavaScript array indexes begin at 0 |
| `LENGTH(parts)` | `parts.length` |
| `parts[index]` | `parts[index - 1]` when using a 1-based pseudocode index |

### SFI QA Selection in College Board Pseudocode

```text
recordIndex ← RANDOM(1, LENGTH(parts))
actionNumber ← RANDOM(1, 4)

selectedPart ← parts[recordIndex]

IF (actionNumber = 1)
{
    DISPLAY("GET/search")
}
ELSE
{
    IF (actionNumber = 2)
    {
        DISPLAY("POST/create")
    }
    ELSE
    {
        IF (actionNumber = 3)
        {
            DISPLAY("PUT/update")
        }
        ELSE
        {
            DISPLAY("DELETE/remove")
        }
    }
}
```

### Possible Results

If `LENGTH(parts) = 3`:

- `recordIndex` can be 1, 2, or 3.
- `actionNumber` can be 1, 2, 3, or 4.
- There are `3 × 4 = 12` possible record/action pairs.
- A pair may repeat on consecutive runs.

### JavaScript Equivalent

```javascript
const selectedPart = parts[Math.floor(Math.random() * parts.length)];
const actions = [
    "GET/search",
    "POST/create",
    "PUT/update",
    "DELETE/remove"
];
const selectedAction = actions[Math.floor(Math.random() * actions.length)];
```

## 4. Lesson Plan

**Learning Objective:** Generate random values in JavaScript, determine every possible result, and apply randomness to backend-style testing.

**Success Criteria:** You can choose a valid random range, explain whether endpoints are included, distinguish random choice from guaranteed rotation, and use random results safely in program logic.

### Lesson Theme — SFI Backend QA Test Garage

**Theme:** You are a QA engineer stress-testing an SFI automotive safety-spec backend.

The lesson keeps one story from start to finish:

- **Records:** SFI automotive components.
- **Random record selection:** chooses which part gets tested.
- **Random API action:** chooses search, create, update, or remove behavior.
- **Existing spec numbers:** create realistic duplicate conditions.
- **Repeated runs:** exercise different backend paths.

Randomness is not added just for entertainment. It represents a real testing strategy:

> “Can the backend handle many valid combinations of records and actions without producing an invalid state?”

The final simulator combines the full theme into repeated QA runs.

### Tech Talk (short segments between checks)

JavaScript's `Math.random()` generates a decimal from 0 (inclusive) to 1 (exclusive). Scale and round it to create integers or choose array elements:

| JavaScript | Purpose | Example |
| --- | --- | --- |
| `Math.floor(Math.random() * (max - min + 1)) + min` | Random integer from `min` through `max`, inclusive | `Math.floor(Math.random() * 4) + 1` |
| `items[Math.floor(Math.random() * items.length)]` | Random item from an array | `parts[Math.floor(Math.random() * parts.length)]` |
| repeated calls | Independent random selections | the same result may repeat |

A random result is **not** the same as cycling through all values. Repeats are normal.
