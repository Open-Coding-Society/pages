### Introduction to Mathematical Expressions

A mathematical expression is a combination of numbers, variables, and operators that evaluates to a value. In JavaScript, expressions are used to calculate results, store values, and help make decisions in programs.

For example:

`let total = 1 + 2 * 3;`

This expression follows the order of operations, so it is evaluated as:

`1 + (2 * 3) = 7`

> Key idea: Mathematical expressions in programming are very similar to the expressions you use in math class. The main difference is that we often use variables and JavaScript syntax.

---

### How Arithmetic Operations Work

Each math operator in JavaScript does a specific job:

- `+` adds two values together
- `-` subtracts one value from another
- `*` multiplies values
- `/` divides one value by another
- `%` gives the remainder after division
- `**` raises a number to a power

The operator works on the values around it and produces one result.

```javascript
let a = 10;
let b = 3;

console.log(a + b); // 13
console.log(a - b); // 7
console.log(a * b); // 30
console.log(a / b); // 3.333...
console.log(a % b); // 1
console.log(a ** b); // 1000
```

Think of an expression like a recipe:

- JavaScript reads the numbers and symbols
- it finds the operation to do
- it calculates the result
- then it uses that result in the rest of the code

For example:

```javascript
let x = 8;
let y = 2;
let result = x + y * 3;

console.log(result); // 14
```

This is not `(8 + 2) * 3` because JavaScript does multiplication before addition unless parentheses are used.

---

### Key Mathematical Operators

| Operator | Symbol | Example | Result | Example use case |
|----------|--------|---------|--------|------------------|
| Addition | `+` | `2 + 3` | `5` | `console.log(2 + 3);` |
| Subtraction | `-` | `5 - 2` | `3` | `console.log(5 - 2);` |
| Multiplication | `*` | `4 * 3` | `12` | `console.log(4 * 3);` |
| Division | `/` | `10 / 2` | `5` | `console.log(10 / 2);` |
| Modulo | `%` | `7 % 3` | `1` | `console.log(7 % 3);` |
| Exponentiation | `**` | `2 ** 3` | `8` | `console.log(2 ** 3);` |

You can also use comparison operators in `if` statements, such as `>`, `<`, `==`, `<=`, and `>=`.

---

### Variables in Expressions

You can use variables to store values and build expressions:

```javascript
let x = 5;
let y = 3;
let total = x + y;

console.log(total);
```

Here, `x` and `y` are just placeholders for numbers. The expression `x + y` is evaluated using the values stored inside those variables.

---

### Order of Operations in Code

JavaScript follows the same order of operations as regular math: PEMDAS.

1. Parentheses
2. Exponents
3. Multiplication / Division / Modulo
4. Addition / Subtraction

This means JavaScript evaluates expressions in a predictable order, so the result is always based on the same rules.

#### Example 1

```javascript
let answer = 10 + 4 * 2;
console.log(answer); // 18
```

Why? Because multiplication happens before addition:

`4 * 2 = 8`, then `10 + 8 = 18`

#### Example 2

```javascript
let answer2 = (10 + 4) * 2;
console.log(answer2); // 28
```

Why? The parentheses tell JavaScript to do the addition first:

`10 + 4 = 14`, then `14 * 2 = 28`

#### Example 3

```javascript
let num = 9 - 3 + 2;
console.log(num); // 8
```

This is evaluated left to right because subtraction and addition are the same priority level:

`9 - 3 = 6`, then `6 + 2 = 8`

#### Example 4

```javascript
let remainder = 10 % 3 + 4;
console.log(remainder); // 5
```

JavaScript does modulo before addition:

`10 % 3 = 1`, then `1 + 4 = 5`

> Important: If you want JavaScript to do something in a different order, use parentheses. Parentheses override the normal order and tell the computer to calculate that part first.

---

### Quick Summary

When you read code with math, always ask:

- What is being calculated first?
- Are there parentheses?
- Is there multiplication, division, or modulo before addition and subtraction?

The order matters, and understanding it helps you predict exactly what your program will do.
