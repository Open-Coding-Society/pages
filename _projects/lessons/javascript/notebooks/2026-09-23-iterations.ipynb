## 1. Reference Guide

| Term | Definition | Example |
| --- | --- | --- |
| **Iteration** | One pass through the body of a loop | `i` goes 1, 2, 3... |
| **Initialization** | Sets the counter once, before the loop starts | `let i = 0` |
| **Condition** | Checked before each pass. `false` ends the loop | `i < 5` |
| **Increment** | Updates the counter after each pass | `i++` or `i += 2` |
| **Infinite loop** | A loop whose condition never becomes false | `while (true) { }` |
| **Key** | The name of a property in an object | `name` in `{name: "Casey"}` |

### Picking a Loop

| Goal | Loop |
| --- | --- |
| Repeat a set number of times | `for (let i = 0; i < 5; i++)` |
| Visit every item in an array | `for (let i = 0; i < arr.length; i++)` |
| Visit every key of an object | `for (let key in obj)` |
| Repeat until something changes | `while (condition)` |

Count by 2s with `i += 2`. Count down with `i--`.

## 2. LxD Cycle Process

**Empathize:** Students copy and paste the same line five times instead of writing a loop, and when they do write one it runs one time too many, too few, or never stops. Even students who pass programming tests often hold wrong mental models of how loops execute (Bayman & Mayer, 1983).

**Define:**

* **POV:** Students need to see every loop as "start, check, change," because a missing part causes the bug.
* **Learning Goal:** Students will write `for`, `for...in`, and `while` loops over numbers, arrays, and objects, and control where a loop starts, stops, and how it counts.

**Ideate:** How might we get students to ask "when does this stop?" before they run a loop? Activity: loop over games and player stats, then finish a self-checking coin total loop.

**Prototype:** A reference guide, two live examples, two popcorn hacks, and a homework hack that checks itself with PASS/FIX lines.

**Test:** Peers finish the popcorn hacks without extra explanation while we watch whether they set the condition on purpose or guess. We revise any unclear step, then revise again after teaching and grading.

### Tech Talk (5 minutes)

An **iteration** repeats a block of code. Every loop needs a **start**, a **condition**, and a **change**.

```javascript
for (let i = 1; i <= 5; i++) {   // start; condition; change
    console.log("Level " + i);
}

let level = 1;                   // start
while (level <= 5) {             // condition
    console.log("Level " + level);
    level++;                     // change
}
```

Both print Level 1 to Level 5. Use `for` when you know how many times to repeat, and `while` when you only know the rule for stopping. If the change is missing, you get an **infinite loop**.

## 5. Code Examples

### Example A: Arrays and Objects

A `for` loop visits each array element by index `i`, stopping at `videoGames.length`. `for...in` visits each **key** of an object, and `playerStats[key]` gets its **value**.

**Try it:** add two of your own games and change the stats to your own gamer profile.

````python
%%js
// CODE_RUNNER: Run the loops, then add two of your own video games and change the stats to your own gamer profile

let videoGames = ["Minecraft", "Roblox", "Fortnite"];

for (let i = 0; i < videoGames.length; i++) {
    console.log(videoGames[i]);
}

const playerStats = { // Define an object
    name: "Sriansh",
    level: 14,
    favoriteGame: "Greenville",
};

for (let key in playerStats) { // Notice we are using for... in... instead of for... of...
    console.log(key + ": " + playerStats[key]);
}
````

### Example B: while Loops

A **`while` loop** repeats **as long as its condition is true**. Use it when you only know the rule for when to stop. It still needs a start, a condition, and a change inside the loop.

**Try it:** change the starting value, the condition, and the increment (e.g., `coins += 2`), then make it count down.

````python
%%js
// CODE_RUNNER: Run the loop, then change the starting value, the condition, and the increment

let coins = 0;

while (coins < 5) {
  console.log("Coins collected:", coins);
  coins += 1;
}
````
