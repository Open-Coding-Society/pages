## JavaScript Conditionals

## 1. Reference Guide

### Key Vocabulary

| Term | Meaning | Example |
|---|---|---|
| Condition | A question that is true or false. | `coins >= 10` |
| `if` | Runs code when its condition is true. | `if (coins >= 10) { ... }` |
| `else` | Runs code when the `if` condition is false. | `else { ... }` |
| Code block | Code inside curly braces `{ }`. | `{ console.log("Hello"); }` |

A game can use a condition to decide whether a player has enough coins to buy something.

## 2. LxD Cycle

**Define:** Players need a simple way to understand how a game makes decisions.

**Learning Goal:** Use an `if...else` statement to check whether a player has enough coins.

**Ideate:** How can a game respond differently when a player has enough coins and when they do not?

**Activity:** Use player names and coin totals to display a simple game message.

### 3. CS111 Requirements

This lesson practices introductory programming control structures by using a condition to choose which code runs. Students will read, change, and write a simple JavaScript `if...else` statement.

## 4. Lesson Plan

**Learning Objective:** Use `if` and `else` to make a simple decision about a player's coins.

**Success Criteria:** You can identify the condition and write a message for both outcomes.

### Tech Talk

A player wants to buy a power-up that costs 10 coins. The game checks `coins >= 10`. If the condition is true, the player can buy it. Otherwise, the `else` block runs.

## 5. Code Example

Run the code. Change the player's name or coin total, then run it again to see how the decision changes.

### A. Checking a Player's Coins

```javascript
%%js
// CODE_RUNNER: Change the name or coin total and run the code.
const playerName = "Alex";
const coins = 8;

if (coins >= 10) {
  console.log(playerName + " can buy a power-up!");
} else {
  console.log(playerName + " needs more coins.");
}
```
