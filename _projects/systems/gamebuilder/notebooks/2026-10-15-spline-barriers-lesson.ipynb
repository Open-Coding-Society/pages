## 1. Reference Guide

### Key Vocabulary

| Term | Definition | Example |
| --- | --- | --- |
| **Spline barrier** | A smooth, open boundary that blocks player movement | A winding cave wall |
| **Control point** | A coordinate that shapes the barrier curve | <code>{ x: 0.25, y: 0.6 }</code> |
| **Normalized coordinate** | A position from 0 to 1 relative to the game canvas | x: 0 is left; x: 1 is right |
| **Point sequence** | Control points stored in order along the barrier | An array of x/y objects |
| **Spline** | A smooth curve interpolated through the control points | A riverbank-shaped boundary |
| **Collision** | Runtime behavior that prevents a player crossing the barrier | Hidden barriers still block movement |

### Barrier Structure

A barrier stores its identity, visibility, and ordered points.

| Part | What it is | Example |
| --- | --- | --- |
| Name | Label used to identify the barrier | Riverbank |
| Visibility | Whether the barrier is drawn | visible: true |
| Point | One normalized x/y position | { x: 0.2, y: 0.4 } |
| Points | Ordered list; at least two points are required | [{x: 0.2, y: 0.4}, {x: 0.8, y: 0.7}] |
| Curve | Smooth open spline formed from the points | One continuous boundary |

### Barrier Grammar

| Structure | What it does |
| --- | --- |
| { x: 0.2, y: 0.4 } | Defines one point in normalized canvas coordinates |
| points: [...] | Keeps points in curve order |
| points.push(point) | Adds a point to the end of the sequence |
| points.pop() | Removes the most recently added point |
| visible: false | Hides the curve without turning off collision |
| At least two points | Minimum needed to finish an open spline barrier |

### Choosing a Barrier Setup

| Goal | What to do |
| --- | --- |
| Make a straight-ish boundary | Use two points |
| Follow a winding edge | Add more points in order along the curve |
| Place points on any canvas size | Use normalized values from 0 through 1 |
| Temporarily hide a barrier | Set visibility off; collision remains active |
| Undo the last placement | Remove the last point from the sequence |

### Tech Talk (5 minutes)

A **spline barrier** is an open, smooth curve that blocks player movement. The game stores an ordered list of control points instead of many separate wall segments.

* **The Rule:** Use at least two points, keep each x/y value between 0 and 1, and list points in order along the boundary.
* **Do this:** Use three ordered normalized points to form a winding boundary.
* **Don't do this:** store screen-pixel coordinates such as { x: 400, y: 160 } when the project expects normalized coordinates.

Why? Normalized positions scale with the logical canvas. The game smooths the points into a curve for drawing and collision. Setting visible to false hides the drawing but does not remove the collision boundary.

~~~javascript
const barrier = {
  name: "Riverbank",
  visible: true,
  points: [
    { x: 0.1, y: 0.3 },
    { x: 0.5, y: 0.7 },
    { x: 0.9, y: 0.3 }
  ]
};
~~~

Read the point list from one end to the other. Moving one point changes the curve; it does not create a separate wall.

## 5. Code Examples

### A. Simple: A Barrier Object

A barrier groups its name, visibility, and points in one object. Each point is an object with normalized x and y values.

**Try it:** change the barrier name and move the middle control point.

```python
%%js
// CODE_RUNNER: Change the barrier name and move the middle point, then run

const barrier = {
  name: "Riverbank",
  visible: true,
  points: [
    { x: 0.1, y: 0.3 },
    { x: 0.5, y: 0.7 },
    { x: 0.9, y: 0.3 }
  ]
};

console.log(barrier.name);
console.log(barrier.points);
```

### B. Simple: Normalized Coordinates

Normalized positions use the logical canvas rather than fixed screen pixels. A value of 0 starts at the left or top edge; 1 reaches the right or bottom edge.

**Try it:** change the values but keep every coordinate between 0 and 1.

```python
%%js
// CODE_RUNNER: Change the point positions while keeping coordinates from 0 to 1

const points = [
  { x: 0.15, y: 0.2 },
  { x: 0.5, y: 0.75 },
  { x: 0.85, y: 0.35 }
];

for (let i = 0; i < points.length; i++) {
  console.log("Point " + (i + 1) + ":", points[i].x, points[i].y);
}
```

### C. Intermediate: Add and Undo Points

Points are stored in order. Adding one appends it to the end; undo removes the most recently added point. A finished barrier must keep at least two points.

**Try it:** add one point, run the code, then use pop() to undo the last point.

```python
%%js
// CODE_RUNNER: Add a point, then undo the last point with pop()

const points = [
  { x: 0.1, y: 0.3 },
  { x: 0.5, y: 0.7 }
];

points.push({ x: 0.9, y: 0.4 });
console.log("After adding:", points);

points.pop();
console.log("After undo:", points);
```

### D. Intermediate: Visibility and Collision

Visibility controls whether the curve is drawn. It is separate from collision, so a hidden barrier can still block the player.

**Try it:** set visible to false. The last line should still report that the barrier blocks movement.

```python
%%js
// CODE_RUNNER: Hide the barrier and observe that collision remains enabled

const barrier = {
  name: "Garden Edge",
  visible: false,
  points: [
    { x: 0.2, y: 0.2 },
    { x: 0.5, y: 0.6 },
    { x: 0.8, y: 0.3 }
  ]
};

console.log("Drawn:", barrier.visible);
console.log("Collision boundary exists:", barrier.points.length >= 2);
```

### E. Complex: Validate a Barrier

Before finishing a barrier, confirm it has a name, at least two points, and normalized coordinates.

**Try it:** make one coordinate larger than 1, run the check, then correct it.

```python
%%js
// CODE_RUNNER: Edit the barrier, then run its validation checks

const barrier = {
  name: "Forest Path",
  visible: true,
  points: [
    { x: 0.1, y: 0.2 },
    { x: 0.5, y: 0.7 },
    { x: 0.9, y: 0.3 }
  ]
};

const hasName = barrier.name.length > 0;
const hasEnoughPoints = barrier.points.length >= 2;
const normalized = barrier.points.every(point =>
  point.x >= 0 && point.x <= 1 && point.y >= 0 && point.y <= 1
);

console.log((hasName ? "PASS" : "FIX") + "  barrier has a name");
console.log((hasEnoughPoints ? "PASS" : "FIX") + "  barrier has at least two points");
console.log((normalized ? "PASS" : "FIX") + "  every x/y value is from 0 to 1");
```

### Homework Hack: Build and Check a Barrier

**Task:** create a barrier object with a name, visibility, and at least three control points. Test it with your own shape, then change one point and run it again.

**Goal:** bring the lesson together: choose the point order, use normalized coordinates, and explain the difference between drawing and collision.

**Hint:** a curve that looks smooth still needs valid data: at least two points and every coordinate between 0 and 1.

Edit the code and press **Run**. The check lines mark each rule **PASS** or **FIX**. Keep going until every line says **PASS**.

```python
%%js
// CODE_RUNNER: Build a barrier and edit it until every check says PASS

const wall = {
  name: "Castle Wall",
  visible: true,
  points: [
    { x: 0.15, y: 0.25 },
    { x: 0.5, y: 0.65 },
    { x: 0.85, y: 0.3 }
  ]
};

const hasName = wall.name.trim().length > 0;
const enoughPoints = wall.points.length >= 3;
const normalized = wall.points.every(point =>
  Number.isFinite(point.x) && Number.isFinite(point.y) &&
  point.x >= 0 && point.x <= 1 && point.y >= 0 && point.y <= 1
);

console.log("Barrier:", wall.name);
console.log((hasName ? "PASS" : "FIX") + "  barrier has a name");
console.log((enoughPoints ? "PASS" : "FIX") + "  barrier has at least three points");
console.log((normalized ? "PASS" : "FIX") + "  every coordinate is normalized from 0 to 1");
console.log((wall.visible === false || wall.visible === true ? "PASS" : "FIX") + "  visibility is a boolean");
console.log("Hidden barriers still participate in collision.");
```
